# <center>OpenAICodex快速入门实战（下）</center>

&emsp;&emsp;上一课已经把 Codex App 的基础使用、项目委派、<b>计划模式</b>、Git、生图、AGENTS.md、多智能体、Computer Use、自动化和 <b>Skill</b> 串了一遍。本课继续往真实工作场景里推进，不再从安装和基础界面讲起，而是关注 Codex 能不能帮我们完成更具体的交付任务。

&emsp;&emsp;本课会围绕五类结果展开：数据看板、信息图、PPT、录制自动化和 Goal 长任务。它们看起来属于不同工作场景，但背后有一条共同路线：先让 Codex 理解任务和材料，再让它给出方案，关键事实先核验，产物生成后继续局部修改，最后把可复用的方法沉淀下来。

&emsp;&emsp;Codex 在本课里不是单一的代码生成工具，而是围绕材料、目标和结果持续工作的任务执行工具。我们会看到它读表、整理指标、生成网页；也会看到它根据图片用途选择生成方式，把长文稿整理成 PPT，把一次手动操作变成可复用 <b>Skill</b>，再把长目标任务拆成可检查的执行过程。

&emsp;&emsp;这五类任务的关键差异不在按钮，而在判断顺序。数据任务要先核验指标，图片任务要先判断生成方式，PPT 任务要先提炼主线，录制自动化要先检查对象和流程，<b>Goal 模式</b>要先写清目标、约束、工具和<b>验收标准</b>。

&emsp;&emsp;第一章先从数据看板开始。数据类任务最容易出现的问题不是页面不好看，而是指标算错、字段关联错、结论看起来合理但没有依据。因此本章会把“核验”放在生成之前，先让 Codex 成为数据分析流程的一部分，而不是直接让它替我们下结论。


## <center>第一章 用 Codex 做数据看板</center>

&emsp;&emsp;真实运营复盘里，数据往往不是一张整理好的汇总表，而是几份分散的原始表。订单表记录每笔交易，用户表补充用户维度，渠道投放表记录每个月的广告花费和流量表现。经营看板要把这些表连起来，才能回答“这季度生意做得怎么样、钱花得值不值”。

&emsp;&emsp;本章走的是一条稳妥流程：先看清三张表，再让 Codex 设计看板方案；方案出来后不急着执行，先核验关键指标；指标能对上以后，再用可视化插件生成 HTML 看板，并通过注释做局部精修。

### 1.1 浏览数据表格

&emsp;&emsp;本章使用的是一套消费品牌第二季度运营数据，时间范围覆盖 4 月到 6 月。三份文件分别承担不同职责：`Q2经营数据.csv` 是订单事实表，`Q2用户明细.csv` 是用户维度表，`Q2渠道投放.csv` 是渠道花费和流量表。

> <font size=2>【名词解释】<b><font color=red>CSV</font>（Comma-Separated Values，逗号分隔值文件）</b>：一种用逗号分隔字段的表格文件格式，常用于导出订单、用户、投放等业务数据。</font>

&emsp;&emsp;三张表的关系决定了后续分析能不能成立。订单表可以通过 `用户ID` 连接用户表，补充注册渠道、会员等级、年龄段等用户维度；订单表还可以通过 `获客渠道` 和订单月份连接渠道投放表，进而计算投放成本、渠道成交额和渠道 ROI。

&emsp;&emsp;这里要先形成一个判断：经营看板不是把表格变成图表，而是把多个业务问题连起来。总销售额、订单量、客单价、复购率、渠道 ROI、月度趋势这些指标，只有在字段和表关系成立的前提下才有意义。


> <font size=2>【名词解释】<b><font color=red>GMV</font>（Gross Merchandise Volume，成交总额）</b>：一段时间内订单成交金额的总和，本章用它衡量第二季度整体销售规模。</font>

> <font size=2>【名词解释】<b><font color=red>ROI</font>（Return on Investment，投入产出比）</b>：投入成本和产出收益之间的比例，本章用它比较不同获客渠道的投放效率。</font>

### 1.3 让 Codex 给出数据看板方案

&emsp;&emsp;数据放进项目后，先不要直接让 Codex 生成页面。我们先打开<b>计划模式</b>，让它把三张表读一遍，确认字段、行数、数据质量和表之间能不能对上。这样后面生成看板时，方向会更稳。

&emsp;&emsp;给 Codex 的提示词可以同时写清楚业务目标和检查动作：我们关心什么问题，它需要先检查哪些表，最后要给出什么样的看板方案。

&emsp;&emsp;这里可以这样输入：


> 💬 **输入给 Codex 的提示词**

```text
项目里有三张运营表，我想做一个Q2季度经营看板，需要讲清楚"这季度生意做得怎么样、钱花得值不值"。
请你把三张表都读一遍，
1. 告诉我各有哪些字段、多少行、数据干不干净；
2. 看一下这三张表之间能不能对上；
3. 基于能对上的关系，告诉我能挖出哪些有价值的分析，我最关注的是总GMV、总订单数、各渠道的ROI，你也可以关注下其他的指标，然后给我一份看板方案：放哪些关键指标、配哪些图表。
```

### 1.4 审核 Codex 指标数据

&emsp;&emsp;Codex 给出看板方案后，不要马上执行。数据类任务最容易出现的问题是：页面看起来很完整，但底层指标算错了。只要关键数字不对，这个看板就不能拿来汇报。

&emsp;&emsp;这里先看 Codex 给出的方案。它推荐了总 GMV、总订单数、各渠道 ROI 等指标和图表，我们先判断这些内容是否覆盖了业务问题，再决定下一步怎么做。

```text
先不执行方案，我需要验证方案的数据正确性，按你给出的方案，把里面要用到的关键指标都算出来：
请读取三张表，输出“指标核验口径表”。不要创建文件。
指标顺序按照计算复查程度从易到难排序。
每个指标请给出：指标名称、使用表、字段、计算公式、人工验证方法、当前结果。
```

&emsp;&emsp;核验时尤其要看两个点。第一，简单指标先验，例如总订单数、总 GMV、客单价，这些数字很容易人工复查。第二，复杂指标再验，例如复购率和渠道 ROI，它们依赖去重、分组、关联和投放成本，最容易因为字段关系理解错而出问题。

&emsp;&emsp;核验表还要看“公式能不能被别人复算”。例如总 GMV 只需要汇总订单金额，客单价只需要用总 GMV 除以总订单数，这类指标的复查成本低，适合放在最前面。复购率涉及按用户去重并统计两单及以上用户，渠道 ROI 还要把订单表和渠道投放表按渠道、月份对齐，这类指标就要检查关联字段和时间口径。只要复查方法写不清，页面生成得再完整，也不能进入汇报材料。

### 1.8 小结

&emsp;&emsp;本章完成的不是不单单是““做出一个好看的看板”，而是掌握一套更稳妥的工作方法：**先让 Codex 理解数据，再让它设计方案；先核验指标口径，再执行生成；最后通过注释进行局部优化。**

&emsp;&emsp;把原始表交给 Codex 时，先让它读表和判断关系；看板方案出来后，先收住生成动作，转向指标核验；核验通过后，再让插件负责页面化；生成之后，用注释对局部内容做精修。

&emsp;&emsp;完成数据看板后，我们已经看到 Codex 可以处理结构化数据并生成网页交付。下一类任务换成信息图时，核心判断会从“指标是否可信”转向“文字是否清楚、生成方式是否匹配用途”。

## <center>第三章 用 Codex 做 PPT</center>

&emsp;&emsp;PPT 任务比信息图更复杂，因为它不是单张图，而是一组有顺序、有主线、有节奏的页面。Codex 可以生成 PPTX 文件，但如果直接把一大段原稿丢给它，结果很容易像“把文稿摘到幻灯片上”。

&emsp;&emsp;所以这一章先整理内容，再生成文件。先提炼汇报主线，把原稿变成成果表达；再设计<b>页纲</b>，明确每页标题、核心信息、数据和页面表达方式；最后确定视觉风格，生成 PPTX，并逐页检查。

### 3.3 提炼主线与内容

&emsp;&emsp;比较好的让codex制作PPT的步骤如下：

1. **先定主线**：主线通常是：今年围绕什么目标做了哪些事，带来了什么结果，沉淀了什么方法，明年准备怎么提升。
2. **把内容提炼成 PPT 成果表达**：从原始叙述中提炼目标、动作、结果和影响，用数据和结论表达价值。
3. **先做页纲**：拆成每一页：标题、核心信息、数据、页面表达方式。
4. **确定视觉风格**：公司规定模板就直接给codex模板，没有可以自己去网上寻找合适的模板。
5. **生成 PPTX**：每页只讲一个重点，少放大段文字，多用数字、表格、对比和简短结论。
6. **逐页检查**：生成后必须检查：文字有没有挤、有没有重叠、表格是否可读、数据有没有前后不一致、有没有模板占位文字残留。
7. **多轮优化**：第一版看结构，结构没问题之后就可以和codex进行对话来优化PPT。

### 3.7 小结

&emsp;&emsp;通过这个案例可以看到，Codex 确实可以帮助我们完成 PPT 的生成，但如果只是把一大段原稿直接丢给它，让它“生成一个 PPT”，最终效果往往更像是把文档内容搬到了幻灯片里，页面会偏文字堆叠，也不太像真正的汇报型 PPT。

&emsp;&emsp;更好的方式是把 Codex 当成一个“PPT 制作助手”，而不是简单的“文档转 PPT 工具”。我们需要先和它一起完成内容梳理：明确汇报主线，提炼关键成果，拆出页纲，再确定视觉风格，最后才进入 PPTX 生成阶段。这样生成出来的 PPT，结构会更清楚，数据表达也会更像汇报，而不是原稿摘抄。

&emsp;&emsp;在实际使用中，比较推荐按照“原稿整理 → 主线提炼 → 页纲设计 → 风格参考 → 生成 PPT → 注释修改 → 多轮优化”这个流程来做。尤其是页纲和风格参考非常关键，页纲决定了 PPT 讲什么，风格参考决定了 PPT 长什么样。后续再通过 Codex 的注释功能，对封面、图表、排版、文字密度等细节逐页修改，最终就能得到一份比较可用的汇报 PPT。

&emsp;&emsp;所以，Codex 做 PPT 的核心不是“一句话生成”，而是用对流程：先让内容变得适合 PPT，再让 Codex 按照明确的结构和风格去生成。

## <center>第四章 用 Record & Replay 录制并重复</center>

&emsp;&emsp;很多工作不是写文件，而是每天在同一个软件里重复点几步，比如打开应用、进入页面、点击签到、确认状态。这类任务步骤简单，但容易忘，也不值得每天手动做。<b>Record & Replay</b> 就适合处理这种固定路径的界面操作。

### 4.1 Record & Replay 插件

&emsp;&emsp;Record & Replay 是 Codex app 在 2026 年 6 月 18 日发布的新功能，属于 macOS 上的工作流录制能力。它可以让用户先手动演示一遍操作流程，然后 Codex 会把这套流程整理成一个可复用的 Skill，后续就可以通过自然语言让 Codex 重复执行类似任务。

&emsp;&emsp;OpenAI 官方 Codex changelog 里写的是：**2026-06-18，Codex app 26.616 新增 Record & Replay**，它是一个 macOS 功能，可以把用户演示的一套工作流转成可复用的 Skill。初始可用范围不包括 EEA、英国和瑞士，并且需要启用 Computer Use。 

&emsp;&emsp;可以直接在「插件」-「搜索」-「Record & Replay」-「添加插件」来安装这个插件

## <center>第五章 用 Goal 模式与 MCP 完成长任务</center>

&emsp;&emsp;普通提示词适合一次性回答，<b>Goal 模式</b>更适合长期任务。旅行规划就是典型例子：它不只是写一篇攻略，还要同时考虑预算、天气、路线、睡眠、交通、景点取舍、餐饮、来源和最终页面呈现。

### 5.1 Goal 模式

&emsp;&emsp;codex有一个goal模式（目标模式），它可以让 Codex **记住一个长期目标，并不断自我推进，直到它判断目标完成或遇到限制/阻塞**，让 Codex 朝一个明确目标持续工作“数小时甚至数天”，Codex 处于goal模式下会自己决定下一步做什么、任务是否已经完成。

> <font size=2>【名词解释】<b><font color=red>Goal</font>（Goal，目标）</b>：这里指 Codex App 里围绕一个长期目标持续推进的工作方式，适合多约束、需要持续检查的任务。</font>

&emsp;&emsp;Goal Mode 是 Codex 对 Loop Engineering 的一个产品化实现。这背后对应的思想就是现在 AI coding 圈说的 Loop Engineering：人不再负责一条条写 prompt，而是负责设计目标、约束、工具、检查点和停止条件，让 Agent 在循环中自己推进。

&emsp;&emsp;今天我们打算使用这个goal模式来帮我们设计一个旅游方案，我们给定自己的要求，让他自己来判断应该如何设计能满足我们的要求。

### 5.2 创建文件夹

&emsp;&emsp;旅游规划里面有很多变量：最主要的是时间、城市、预算、交通、酒店位置、景点开放时间、天气、餐馆吃饭时间地点价格等等。

&emsp;&emsp;goal模式来做旅游规划的优点是，它可以持续围绕目标检查：这个行程是否合理， 交通时间是否合理， 天气/开放时间有没有冲突， 预算有没有超， 有没有备用方案。

<tr><td>工具要求</td><td>说明哪些信息必须查证</td><td>地图 MCP、网页搜索、文件读取、天气来源</td></tr>
<tr><td>验收标准</td><td>列出完成结果必须包含什么</td><td>模块清单、自检清单、来源说明、最终文件名</td></tr>
</tbody>
</table>
</div>


&emsp;&emsp;普通提示词关注这一轮回答是否好，Goal 关注整个任务是否达标。写 Goal 时，最重要的是让 Codex 有明确的停止条件：预算是否达标，路线是否合理，来源是否说明，最终文件是否存在。没有验收标准，长任务很容易变成“生成了很多内容，但不知道是否完成”。

&emsp;&emsp;长任务还要把数据可信度分层。能通过工具或网页确认的信息，写成真实查询；只能根据经验估算的信息，明确标注估算；查不到或工具失败的信息，写出失败原因。这样即使最终结果不完美，读者也能判断哪些部分可以直接采用，哪些部分需要人工复核。


### 5.7 小结

&emsp;&emsp;Goal 模式适合多约束任务，但不等于可以放任它自由发挥。目标要清楚，约束要可检查，工具要指明，来源要写出来，失败和估算要诚实标注。大理旅行规划案例的价值就在于，它把这些要求都写进了任务说明。

&emsp;&emsp;我们把这个方法迁移到其他任务时，可以先把“完成标准”写出来，再倒推需要哪些资料和工具。比如做竞品分析，就要先写明需要哪些竞品、哪些维度、哪些来源和最终报告格式；做资料整理，就要写明输入文件、分类方式、引用规则和输出格式。Goal 的强项不是替代判断，而是围绕明确标准持续推进。

&emsp;&emsp;因此，写 Goal 的顺序可以反过来想：先写最终验收清单，再写约束，再写工具，最后写目标描述。验收清单越清楚，Codex 越容易判断自己是否还要继续执行；工具要求越清楚，Codex 越不容易用模型记忆替代真实查询。

&emsp;&emsp;这一章之后，五类任务已经完整走过：数据看板、信息图、PPT、录制自动化、Goal 长任务。最后一章不引入新功能，而是把这些方法收束成一张可以迁移的判断表。

### 6.4 下一次使用 Codex 的判断顺序

&emsp;&emsp;下一次遇到新任务时，可以先判断任务类型，再选择 Codex 用法。简单问答用普通对话；数据或文件任务先让 Codex 读材料并做计划；图片任务先选生成方式；重复界面任务考虑 Record & Replay；多约束长任务再使用 Goal 模式和 MCP。

&emsp;&emsp;整门课最后回到一个原则：不要只问 Codex “能不能生成”。更好的问法是：任务目标是什么，材料是否充分，关键事实怎么核验，生成后怎么检查，结果能不能复用。这个顺序稳定下来，Codex 就不只是帮我们完成单次任务，而是成为真实工作交付里可以反复使用的工具。

&emsp;&emsp;如果只记一条主线，可以记成“先收住，再放开”。收住，是先让 Codex 读材料、列方案、写检查点，不急着生成；放开，是在事实、结构和约束清楚之后，让 Codex 生成页面、图片、PPT、Skill 或长任务结果。这个节奏用在任何复杂任务里，都比一开始就要求最终成品更稳。
