&emsp;&emsp;在前置的《OpenAICodex快速入门实战（上）》和《OpenAICodex快速入门实战（下）》里，我们已经跑过 Codex App 的很多基础能力：安装登录、项目委派、计划模式、Git 存档、生图、AGENTS.md、多智能体、Computer Use、自动化、Skill、数据看板、信息图、PPT 和 Goal 长任务。这些内容证明了一件事：Codex App 不只是聊天工具，也不只是“帮我写一段代码”的工具，它已经可以围绕材料、目标和结果完成一整类任务。

## <center>第一章：使用 Codex App 阅读并运行开源项目</center>

&emsp;&emsp;进入正式项目操作前，我们先把 Git、GitHub、Codex App 和仓库的关系分清楚。第一章会从 GitHub 上的开源项目出发，先看懂它，再把它变成自己可以运行和修改的本地项目；如果这些对象的职责没有先摆正，后面的 Fork、Clone、提交、推送和 PR 审查就容易混成一团。

&emsp;&emsp;这一章的主线是 <font color=red>阅读项目、Fork 项目、Clone 到本地、配置启动、体验核心功能</font>。我们不会把 Git 命令扩展成系统课，只在对应操作点轻轻带一句“如果手敲会用到什么命令”。这样我们能把注意力放在真实开源项目接手流程上，而不是被命令细节打断。

&emsp;&emsp;本章选择 `Resume-Matcher` 作为案例。它不是一个零散代码片段，而是包含前端、后端、接口、文件处理、大模型配置和界面功能的完整应用。我们用它练习的不是“看一眼 README”，而是从陌生项目到本地可运行项目的完整第一轮接手。本章会进入 `API Key`、本地项目启动、前后端服务和页面体验；不同电脑上的目录、GitHub 账号和课堂现场路径可能不一样，操作时以课堂现场展示的路径和我们自己的 GitHub 账号为准。

### 1.1 先记住三个位置：本地、云端、项目里

&emsp;&emsp;后面会反复出现 Fork、Clone、Branch、Pull、Add、Commit、Push、Merge、PR 这些词。先不用急着背命令，我们先把<font color=red>三个位置</font>放清楚：Git 主要在本地记录版本，GitHub 是云端远程仓库和协作页面，Codex App 则打开我们当前的本地项目，帮我们读代码、改代码和跑检查。

&emsp;&emsp;先看 Git。Git 本质上是安装在我们电脑上的一个版本管理工具。一个项目被 Git 管理之后，项目目录里会出现一个隐藏的 `.git` 文件夹，里面保存着这个项目的版本历史、分支信息和提交记录。我们可以把它简单理解成：每次提交时，Git 都会给当前项目状态“拍一张快照”，记录这次代码长什么样。Git 本身也有官网：[https://git-scm.com](https://git-scm.com/)。如果电脑里还没有安装 Git，可以从这里下载安装，也可以让 Codex App 帮我们检查安装方式。

&emsp;&emsp;再看 GitHub。GitHub 可以理解成一个云端代码托管平台，它的核心作用是把代码仓库放到远程服务器上，方便我们备份代码、同步代码和进行团队协作。类似的平台不只有 GitHub，比如 GitLab、Gitee、Bitbucket、Azure DevOps 等，也都可以用来远程托管代码。

&emsp;&emsp;本课会主要以 <font color=red>GitHub</font> 为例来讲解完整开发流程。大家可以先打开 GitHub 官网：[https://github.com](https://github.com/)，点击页面上的 Sign up 注册账号。一般流程是填写邮箱、设置密码、选择用户名、完成邮箱验证，然后登录 GitHub。注册完成后，我们就可以在 GitHub 上 Fork 开源项目、创建自己的仓库、查看代码、发起 PR，并参与代码审查和多人协作。注册完成之后，可以打开 Codex App 连接 GitHub 插件，这样 Codex App 可以更好地帮我们处理 GitHub 相关操作。

### 1.3 使用 Codex App 阅读项目架构

&emsp;&emsp;在正式开始之前，我们先简单了解一下 开源项目。<font color=red>开源项目</font>就是开发者把项目的源代码公开出来，其他人可以查看、学习、运行，甚至在遵守开源协议的前提下进行修改和二次开发。比如 GitHub 上就有大量开源项目，有的是工具类项目，有的是完整应用，也有很多和 AI、前端、后端、数据分析相关的真实项目。

&emsp;&emsp;对我们来说，阅读开源项目是非常好的学习方式。因为它不是零散的代码片段，也不是一个小型示例，而是一个完整的软件工程项目，里面通常包含项目结构、依赖管理、接口设计、业务流程、前后端交互、测试和部署等内容。通过阅读这些项目，我们可以学习别人是如何组织代码、设计功能和解决实际问题的。

&emsp;&emsp;这次我们选择 `Resume-Matcher` 这个项目作为例子，主要有几个原因：首先，它是一个比较实用的 AI 应用，可以根据招聘岗位要求优化简历，和求职、AI 工具、文档处理这些场景都比较贴近；其次，从当时页面状态看，它有一定的社区关注度；最后，它同时包含前端、后端、AI 调用、文件解析、PDF 导出等模块，非常适合用来练习如何快速阅读一个真实开源项目的架构。

### 1.4 Fork 到自己的 GitHub 账号

&emsp;&emsp;通过前面的步骤，我们已经借助 Codex App 对 `Resume-Matcher` 项目有了一个整体认识：知道它是做什么的、后端使用什么技术栈、项目大概有哪些核心模块，以及如果后续要阅读源码，应该从哪些文件开始看。

&emsp;&emsp;不过，到这一步我们还只是“了解”这个项目，并没有真正开始修改代码。如果只是学习架构，看 GitHub 页面和让 Codex App 帮我们分析就够了；但如果后续想把项目跑起来、调试接口、修改功能，甚至基于它做二次开发，就需要先把这个项目变成我们自己可以操作的版本。

&emsp;&emsp;所以接下来，我们先把原项目 <font color=red>Fork 到自己的 GitHub 账号下</font>。这样我们就会拥有一份属于自己的项目副本，后面无论是下载到本地、提交代码，还是尝试二次开发，都不会影响原作者的仓库。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-004.png" width=80%></div>

&emsp;&emsp;点击界面右上角的 <b>Fork</b> 按钮，可以在我们的 GitHub 账号下创建一个该项目的副本。

&emsp;&emsp;点击 Fork 的原因是：我们通常没有权限直接修改别人的开源仓库，所以需要先把它复制到自己的 GitHub 账号下。这个 <font color=red>副本仓库</font> 我们有完整修改权限，可以自由学习、运行和二次开发，而且不会影响原项目。如果后续我们做出了有价值的修改，还可以通过 Pull Request 贡献回原项目；原项目维护者认为改动有价值时，才会合并我们的 PR。

## <center>第二章：使用 Codex App 对开源项目进行二次开发</center>

&emsp;&emsp;第一章结束时，我们已经把 `Resume-Matcher` 从 GitHub 页面推进到了本地可运行状态：自己的 Fork 仓库已经克隆到本地，前后端服务已经启动，`API Key`、主简历、职位定制简历和申请追踪页面也都已经体验过。也就是说，我们不再面对一个陌生仓库，而是面对一个已经能跑、能看、能操作的真实项目。

&emsp;&emsp;这一章要在这个基础上往前走一步：让 Codex App 帮我们完成一次<font color=red>小范围二次开发</font>。重点不是“让 AI 随便改代码”，而是按真实项目更稳妥的节奏推进：先检查基线，再创建功能分支；先看计划，再允许改动；先运行测试，再提交、推送和创建 PR。

&emsp;&emsp;课堂现场会重新从自己的 Fork 仓库继续操作。页面上如果出现某个固定账号、固定 PR 编号或固定提交记录，它们只是当时的演示状态，不是学员要照抄的目标。我们真正要复制的是流程：自己的 Fork 仓库、自己的 `feature/tracker` 分支、课堂现场创建出的 PR 编号，以及每一步之前的确认步骤。

### 2.1 确认开发基线

&emsp;&emsp;在开始二次开发之前，我们先不要急着让 Codex App 修改业务代码，而是先确认 <font color=red>开发基线</font>。现在这个项目已经成功跑起来了，而且我们已经完成了 `API Key` 配置、主简历上传、职位定制简历生成和申请追踪页面测试。也就是说，当前项目已经处在一个“可以使用”的状态。

&emsp;&emsp;但是，能跑起来并不代表项目已经适合直接开发。对于一个真实项目来说，除了页面能打开、接口能访问之外，我们还需要确认本地开发环境是否完整，以及后端测试、前端测试这些基础检查是否能够通过。

&emsp;&emsp;具体来说，在正式修改业务代码之前，我们要先确认当前项目处在什么状态：哪些功能是正常的，哪些测试是通过的，哪些问题是原本就存在的。这样后面我们修改代码时，如果出现新的报错，就能判断它到底是原项目本来就有的问题，还是本次二次开发引入的问题。

> <b>提示</b>：这里的基线不是额外负担，而是后面判断责任边界的依据。先知道起点，后面才知道改动带来了什么变化。

&emsp;&emsp;这里要注意，基线检查失败不一定代表项目代码有问题。很多时候只是本地环境还没有准备完整，比如依赖没有安装、虚拟环境没有配置好，或者测试工具缺失。

&emsp;&emsp;所以这一步我们可以让 Codex App 先检查当前项目状态。如果只是环境缺失，就让它按照项目文档或依赖文件补齐环境；如果需要修改项目源码或配置文件，则需要先说明原因，等我们确认后再修改。这样做的目的，是在真正二次开发之前，先得到一个相对稳定、可验证的 <font color=red>开发起点</font>。

&emsp;&emsp;所以在正式二次开发之前，我们先在当前项目中新建一个 Codex App 任务，让它做一次开发前检查，输入下面内容。运行后我们应该看到当前分支、工作区状态、依赖状态、服务状态和测试结果，如果有环境缺失，也会先被单独指出。

> 💬 <b>输入给 Codex 的提示词</b>

```text
请先做一次开发前基线检查，不要修改业务代码。
请确认：
1. 当前 Git 分支和远程仓库是否正确。
2. 工作区是否干净，有没有未提交改动。
3. 后端和前端依赖是否已安装。
4. 后端、前端能否正常启动。
5. 后端测试、前端测试是否通过。
如果检查过程中发现是本地环境缺失导致的问题，比如缺少依赖、虚拟环境未安装完整可以直接按项目文档或依赖文件补齐环境。
如果需要修改项目源码或配置文件，请先说明原因，等我确认后再修改。
最后给我一个结论：现在是否适合开始二次开发。
```

> <b>Git 提示</b>：这里对应状态检查。作用是确认当前分支、远程仓库和工作区是否干净，判断能不能开始开发。

&emsp;&emsp;这段提示词把“可以自动处理”和“必须先确认”的边界分开了：补齐依赖可以执行，修改源码要先说明原因。这样 Codex App 的操作会更可控，我们也能在改业务代码之前先拿到项目状态结论。

### 2.2 明确改造目标

&emsp;&emsp;完成开发前基线检查后，我们已经确认当前项目可以正常启动，后端测试和前端测试也可以通过。接下来就可以正式进入二次开发阶段。

&emsp;&emsp;不过在让 Codex App 修改代码之前，仍然不要一上来就直接说“帮我改一下”。对于真实项目来说，需求越清楚，Codex App 后面定位代码、制定计划和修改功能时就越稳定。

> <b>提示</b>：需求先写边界，不写愿望。这里的边界就是只改申请追踪页面、只控制显示隐藏、保留原有数据。

&emsp;&emsp;所以这一节我们先明确本次二次开发目标。前面我们已经体验过 `Resume-Matcher` 的「申请追踪」功能。这个页面可以用来<font color=red>追踪每份定制简历</font>在求职流程中的进展，比如已保存、已投递、无回复、有回复、面试、已录用、已拒绝等。

&emsp;&emsp;从功能完整性上看，这个设计是比较全面的。但在实际使用中，并不是每个人都需要关注所有状态。比如我们现在更关注求职后期的几个关键阶段：哪些岗位进入了面试、哪些岗位还在等结果、哪些岗位已经录用。其他状态如果全部展示出来，页面信息就会比较多，不够聚焦。

&emsp;&emsp;所以这次我们选择改造「申请追踪」页面，目标是让它更适合自己的求职管理习惯。本次二次开发目标是：在「申请追踪」页面增加一个 管理按钮，让用户可以控制哪些状态模块显示，哪些状态模块隐藏。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-025.png" width=80%></div>

&emsp;&emsp;这里要注意，被隐藏的状态模块不应该删除数据，只是不在页面上展示。比如我们暂时不想看「已保存」「已投递」这些模块，但里面原本的数据仍然应该保留，后续如果重新打开显示，也应该还能看到。这个边界可以概括成 <font color=red>隐藏不删除</font>。

&emsp;&emsp;在正式修改之前，我们不建议直接在 `main` 分支上开发。`main` 通常代表项目的主线版本，最好保持稳定。所以这里我们先让 Codex App 基于当前 `main` 创建一个新的功能分支 `feature/tracker`，用来承载本次申请追踪页面的改造。

&emsp;&emsp;这样做的好处是：后续 Codex App 修改代码、调试功能、运行测试时，都只发生在这个功能分支上，不会直接影响 `main` 主线。等功能开发完成、测试通过之后，再决定是否把这个分支合并回 `main`。

&emsp;&emsp;这里的分支名 `feature/tracker` 也有含义：`feature` 表示这是一个新功能或功能改造分支，`tracker` 表示这次改造的对象是申请追踪页面。所以接下来我们先让 Codex App 创建并切换到 `feature/tracker` 分支，再开始让它制定修改方案。

### 2.3 使用计划模式

&emsp;&emsp;确认已经切换到 `feature/tracker` 分支后，我们不要马上让 Codex App 直接改代码。因为这次要改的是已有开源项目的功能页面，涉及前端页面展示、状态模块控制等内容。直接让 Codex App 开始写代码，虽然也可能完成，但中间容易出现两个问题：它可能还没完全理解项目结构，就直接开始修改文件；它也可能一次性改动太多，我们不方便判断每个改动是否合理。

&emsp;&emsp;所以更稳妥的方式是：先进入 <font color=red>计划模式</font>，让 Codex App 只做代码阅读和方案设计，先告诉我们它准备修改哪些位置、涉及哪些文件、前后端分别需要调整什么。等我们确认方案后，再让它进入正式开发。

### 2.4 提交本地版本

&emsp;&emsp;功能实现并通过测试后，我们还需要把这次二次开发保存成一个本地版本。到目前为止，Codex App 已经完成了代码修改和测试验证，但这些改动还只是保存在当前工作区里，并没有真正形成一次 <font color=red>Git 提交</font>。

&emsp;&emsp;如果此时继续做其他修改，或者后面同步远程 `main` 时出现冲突，就不太方便区分哪些是本次功能开发的改动，哪些是后续操作带来的变化。所以在推送到 GitHub 之前，我们先在当前 `feature/tracker` 分支上创建一次 本地 commit。这个 commit 就相当于给当前功能做一个版本快照。

&emsp;&emsp;这样做有几个好处：后续如果功能有问题，可以通过 commit 快速回退；提交到 GitHub 后，PR 里能清楚看到这次功能对应的提交记录；后面同步远程 `main` 或处理冲突时，当前功能改动已经有了明确边界；如果继续开发其他功能，也不会和这次申请追踪页面改造混在一起。

&emsp;&emsp;我们可以继续让 Codex App 执行本地提交。运行后我们会看到它先确认当前分支，再检查工作区改动范围，最后生成一次本地 commit，但不会推送代码。

### 2.5 推送功能分支

&emsp;&emsp;本地提交完成后，我们还不能马上推送功能分支。本课采用的策略是：推送前先获取远程 `main` 的最新状态，并检查当前 `feature/tracker` 是否落后于远程主线。

&emsp;&emsp;如果远程 `main` 没有新的变化，功能分支可以直接进入推送前确认；如果远程 `main` 已经有新的提交，本课采用的处理方式是把最新 `main` merge 到当前 `feature/tracker` 分支，再重新运行测试。

> <b>提示</b>：这里先同步 `main`，是为了让 PR 尽量基于最新主线创建，减少后续才发现冲突的概率。

&emsp;&emsp;这样做的目的，是让我们在推送前先知道功能分支和远程主线之间有没有差异。即使远程 `main` 有新提交，也不要静默处理；本课先合并到 `feature/tracker`，重跑测试，再判断是否<font color=red>具备推送条件</font>。

&emsp;&emsp;可以让 Codex App 继续执行下面的提示词。运行后我们应该看到它确认当前分支、获取远程 `main` 信息、判断是否落后；只有远程 `main` 有新提交时，才把最新主线合并到当前功能分支，并重新运行相关测试。

> 💬 <b>输入给 Codex 的提示词</b>

```text
请帮我同步远程 main，并准备推送当前功能分支。

要求：
1. 确认当前分支是 feature/tracker。
2. 拉取远程 main 的最新代码。
3. 如果远程 main 有新提交，再将最新 main 合并到当前 feature/tracker 分支。
4. 如果出现冲突，先说明冲突文件和原因，等我确认后再处理。
5. 如果执行了合并，合并完成后重新运行相关测试；如果没有新提交，也请运行相关测试确认可推送。
6. 测试通过后，告诉我当前分支是否可以推送。
7. 不要自动推送，除非我再次确认。
```

> <b>Git 提示</b>：这里用到的是 <code>Pull</code> 和 <code>Merge</code> 两类操作。<code>Pull</code> 的作用是拉取远程 `main` 最新代码，<code>Merge</code> 的作用是把最新主线合并到当前功能分支，降低 PR 冲突风险。

&emsp;&emsp;这段提示词把“同步主线”和“推送分支”拆开了。同步主线是为了减少 PR 阶段的合并风险；推送分支则需要我们再次确认，避免把还没检查完的代码上传到远程。

> <b>Git 提示</b>：这里对应 <code>Push</code>。作用是把本地功能分支上传到自己的 GitHub Fork，不会直接合并进 `main`。

&emsp;&emsp;这里对应的是 <code>Push</code> 推送操作，作用是把本地功能分支上传到 GitHub，并在远程仓库里建立对应分支。

&emsp;&emsp;这里提到的 PR，全称是 Pull Request，可以理解成“合并请求”。我们把 `feature/tracker` 分支推送到 GitHub 后，并不会直接把代码合并进 `main`，而是先创建一个 PR，把这次功能分支里的改动提交出来，让项目维护者或团队成员查看这次改了哪些文件、实现了什么功能、测试是否通过。

> <font size=2><b>【名词解释】PR（Pull Request，合并请求）</b> — 请求把一个分支的改动拿出来审查，并在确认后合并到目标分支。</font>

&emsp;&emsp;确认没有问题后，再通过 PR 把 `feature/tracker` 合并到 `main`。这样既能保留代码审查过程，也能避免直接修改主线分支带来的风险。

&emsp;&emsp;功能分支推送到 GitHub 后，接下来就可以创建 PR 了。在开源项目中，`main` 通常代表项目当前对外提供的主线版本。我们平时访问一个 GitHub 仓库，默认看到的一般也是 `main` 分支；下载代码、克隆项目、部署项目时，默认拿到的通常也是 `main` 分支的内容。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-035.png" width=80%></div>

&emsp;&emsp;所以<font color=red>功能开发一般不会直接改 `main`</font>，而是先在独立分支中开发，比如这次的 `feature/tracker`。等功能完成、测试通过、确认没有问题后，再通过 PR 把这个功能分支合并回 `main`。

&emsp;&emsp;合并完成后，目标仓库的 `main` 分支就会包含这次功能改动。也就是说，后续别人再访问这个仓库的 `main`，看到的就是已经加入「申请追踪管理按钮」之后的版本。

&emsp;&emsp;这里要注意，我们现在操作的是自己的 Fork 仓库。所以如果我们把 PR 合并到自己 Fork 仓库的 `main`，更新的是自己的主分支，不会直接影响原作者的 `srbhr/Resume-Matcher`。如果是真正给原开源项目贡献代码，则需要把 PR 提交到原作者仓库，等维护者审核并合并后，原项目的 `main` 才会更新。

&emsp;&emsp;GitHub 已经在仓库首页提示 `feature/tracker had recent pushes`，并提供了 `Compare & pull request` 按钮。我们可以手动点击这个按钮创建 PR。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-033.png" width=80%></div>

&emsp;&emsp;进入 PR 创建页面后，需要重点确认两个分支：`base` 分支表示要合并到哪里，一般是自己 Fork 仓库的 `main`；`compare` 分支表示从哪里合并过来，这里是自己 Fork 仓库的 `feature/tracker`。

&emsp;&emsp;我们这里需要把 `base` 分支改成自己仓库的 `main` 分支。否则 PR 可能会提交到原作者仓库的 `main` 分支上，这和本章课堂现场自己练习的目标不一致。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-036.png" width=80%></div>

&emsp;&emsp;修改到我们的仓库后，这里还可以看到 GitHub 提示 `Able to merge`。页面上显示 `Able to merge`，说明 GitHub 当前没有检测到合并冲突，可以继续创建 PR。

&emsp;&emsp;所谓合并冲突，通常发生在两个分支同时修改了同一个文件里的同一段代码，Git 无法自动判断应该保留哪一边。比如，`main` 分支上有人把「申请追踪」页面的按钮区域改了一版；与此同时，我们的 `feature/tracker` 分支也在同一个位置新增了「管理」按钮。等我们想把 `feature/tracker` 合并回 `main` 时，Git 就可能不知道这一段到底应该采用 `main` 的写法，还是采用 `feature/tracker` 的写法，这时就会产生冲突。

&emsp;&emsp;出现冲突并不代表代码坏了，只是需要开发者手动确认：哪些内容保留，哪些内容调整，最后再重新提交解决结果。所以在创建 PR 前看到 `Able to merge`，就说明当前这个功能分支和目标 `main` 分支之间<font color=red>没有明显冲突</font>，后续可以正常创建 PR。

&emsp;&emsp;确认分支方向没有问题后，填写 PR 标题和描述。标题可以直接使用这次提交的信息。

```text
feat: add tracker status visibility management
```

&emsp;&emsp;这个标题会出现在 PR 列表和提交记录里，所以最好让它直接说明本次功能。这里的标题含义是：新增申请追踪状态可见性管理能力。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-037.png" width=80%></div>

&emsp;&emsp;进入 PR 创建页面后，如果看到描述框里已经有一段默认内容，这通常是项目提供的 PR 模板。这个内容不是 GitHub 默认自带的，而是这个项目作者自己在仓库里配置的 PR 模板。很多开源项目会放一个 `.github/pull_request_template.md` 之类的文件。创建 PR 时，GitHub 就会自动把这段模板填进描述框，提醒贡献者按项目规范说明改动内容、测试方式、截图或代码片段、检查清单等。

&emsp;&emsp;我们只需要知道它大概分成几块即可，这主要是方便提交 PR 之后，维护者根据描述快速判断这个 PR 的基础信息。

```text
Related Issue：关联的问题编号，如果没有就写 None
Description：说明这次改了什么
Type：选择改动类型
Proposed Changes：列出具体改动
How to Test：说明怎么测试
Screenshots / Code Snippets：如果有界面变化或关键代码片段，可以补截图或片段；没有可以留空
Checklist：勾选已经完成的检查项
Additional Information：补充信息；如果模板已有占位内容，按页面提示保留或补充
```

&emsp;&emsp;PR 标题和描述填写完成后，点击页面底部的 `Create pull request` 即可创建 PR。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-038.png" width=80%></div>

&emsp;&emsp;这里 GitHub 还提供了 `Create draft pull request` 选项。Draft PR 表示草稿状态，通常用于功能还没有完全完成、暂时不准备合并，只是先提交出来给别人预览。本次功能已经开发完成，并且测试通过，所以一般来说可以直接选择 `Create pull request`。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/L2-pr-diff-review.png" width=80%></div>

&emsp;&emsp;在创建 PR 页面下方，还可以看到这次功能分支相对 `main` 的代码差异，也就是 diff 审查。这里 GitHub 会列出本次 PR 包含的提交数量、修改文件数量，以及每个文件里具体新增和删除了哪些代码。绿色表示新增代码，红色表示删除代码。

&emsp;&emsp;在正式创建或合并 PR 之前，我们可以先简单检查几件事：修改文件是否符合预期，比如这次主要应该是看板组件、多语言文案和测试文件；有没有误改无关文件；有没有把调试代码、临时日志、密钥、缓存文件提交进去；新增和删除的代码量是否大致符合这个功能的复杂度；关键逻辑是否和前面计划一致，比如这里只是隐藏状态模块，不删除原有申请数据。

&emsp;&emsp;这一步就是<font color=red>代码审查的入口</font>。真实团队里，其他成员也会在 PR 的 diff 页面查看改动，并在有问题的代码行上留言。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-039.png" width=80%></div>

&emsp;&emsp;除了手动在 GitHub 页面点击 `Compare & pull request` 创建 PR，也可以让 Codex App 直接帮我们创建。不过如果项目本身提供了 PR 模板，最好明确告诉 Codex App：按照项目默认模板填写描述，不要自己重新设计格式。这样创建出来的 PR 会更符合项目维护者的要求。

&emsp;&emsp;这里要注意，Codex App 可以帮我们生成和创建 PR，但创建前仍然需要<font color=red>人工确认几个关键信息</font>：目标仓库是不是自己的 Fork，`base` 分支是不是 `main`，`compare` 分支是不是 `feature/tracker`，以及 PR 描述有没有和实际改动一致。

&emsp;&emsp;我们可以输入下面的提示词。运行后 Codex App 可以帮我们准备 PR，但它不能替代我们确认分支方向和 diff 内容。

> 💬 <b>输入给 Codex 的提示词</b>

```text
可以帮我创建 PR，但请按照当前 GitHub 页面里的默认 PR 模板来填写描述，不要自己改模板结构。

要求：
1. base 分支是我的 Fork 仓库 main。
2. compare 分支是我的 Fork 仓库 feature/tracker。
3. PR 标题使用：feat: add tracker status visibility management
4. PR 描述你按照模板来填写
5. 不要自动合并 PR。
```

> <b>Git 提示</b>：这里对应 <code>PR</code>。作用是在 GitHub 上发起合并请求，让 `main` 接收功能分支前先经过审查。

&emsp;&emsp;这段提示词把目标仓库、目标分支、来源分支、标题、描述格式和合并操作都限定清楚了。运行后课堂现场会得到一个自己的 PR，编号以当次 GitHub 页面显示为准。

## <center>第三章：使用 Codex App 进行多分支协同开发</center>

&emsp;&emsp;第二章里，我们已经走通了一次单功能二次开发：先基于 `main` 创建 `feature/tracker`，再让 Codex App 计划、修改、测试、提交、推送，并通过 PR 合并回 `main`。这条流程已经能支撑一个功能的开发，但真实项目往往不会等一个功能完全合并后才开始下一个功能。

&emsp;&emsp;这一章把上一章的单分支流程升级成 <font color=red>两个工作树（worktree）并行推进</font>。worktree 可以先理解为同一 Git 仓库拆出的多份互不干扰的本地工作目录。我们会用一个 GitHub 账号和多个工作树模拟两个开发者：开发者 A 做“面试时间记录”，开发者 B 做“面试题记录”。重点不是账号数量，而是理解不同功能如何在不同分支和工作区里隔离开发，最后又如何有序合并回稳定主线。

&emsp;&emsp;示例页面里的账号、PR 编号和提交哈希来自当时的演示状态。课堂现场会重新从自己的 Fork 仓库、自己的<font color=red>工作树（worktree）</font>和当次创建出的 PR 编号开始；我们要复制的是流程，而不是示例中的固定数字。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/L3-worktree-parallel.png" width=80%></div>

### 3.1 多分支协同的动机

&emsp;&emsp;上一章的单分支流程可以压缩成一条线：`main → 创建功能分支 → 开发功能 → 提交代码 → 创建 PR → 合并回 main`。对于一个人开发一个小功能来说，这条线已经足够清晰；但真实项目往往不会等一个功能完全合并后再开始下一个功能，更常见的是多个功能同时推进。比如在 `Resume-Matcher` 项目里，可能有人正在优化「申请追踪」页面，有人正在优化「简历导出」体验，还有人正在改进「模型配置」页面。只要这些改动都直接落在 `main` 上，稳定性、改动边界和审查回退都会变得很难控制。

<div align=center><font size=2>表 3-1：直接改 main 与多分支协同的差别</font></div>

<div align="center">
<table width="88%">
  <thead>
    <tr>
      <th>风险点</th>
      <th>都直接改 <code>main</code> 时</th>
      <th>使用功能分支和 PR 时</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>主线稳定性</td>
      <td>未完成、未测试的功能容易进入主线，别人拉到的代码可能不能正常运行。</td>
      <td><code>main</code> 只接收确认过的改动，功能开发先留在独立分支里。</td>
    </tr>
    <tr>
      <td>改动边界</td>
      <td>导出、模型配置、申请追踪等改动混在一起，后面很难看清每个功能改了什么。</td>
      <td>一个功能对应一个分支和一条 PR，diff 更容易审查。</td>
    </tr>
    <tr>
      <td>审查和回退</td>
      <td>某个功能出问题时，很难只回退这个功能而不影响其它功能。</td>
      <td>按功能审查、按功能合并，必要时也更容易按功能回退。</td>
    </tr>
  </tbody>
</table>
</div>

&emsp;&emsp;所以在真实开发中，更推荐的方式是：<b>每个功能都创建一个独立分支，在自己的分支里完成开发、测试和提交。等这个功能确认没有问题后，再通过 PR 合并回 `main`</b>。也就是说，`main` 分支负责保持稳定，功能分支负责承载具体开发；本章后面所有同步和审查步骤，都是为了让每个功能在提交前基于最新 `main`，在合并前保留清晰的功能边界。

&emsp;&emsp;这一章我们就来模拟这样一个<font color=red>多分支协同开发</font>流程。虽然现在我们只有一个 GitHub 账号，但这不影响演示；重点不是账号数量，而是理解不同功能如何在不同分支中并行推进，以及这些分支最后如何通过 PR 有序合并回主线。

### 3.2 同步 main 并明确两个需求

&emsp;&emsp;第二章的 PR 合并以后，远程 `main` 已经变化。本节先把本地 `main` 拉回最新状态，再从同一个稳定起点拆出两个新需求。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/L3-merge-timeline.png" width=80%></div>

&emsp;&emsp;前面我们已经说明了为什么真实项目中需要使用<font color=red>多分支协同开发</font>。接下来就可以正式进入第三章的实操部分。

&emsp;&emsp;我们已经把 `feature/tracker` 分支通过 PR 合并到了 GitHub 远程仓库的 `main` 分支。也就是说，<b>远程仓库里的 `main`</b> 已经包含了「申请追踪页面管理按钮」这个功能。

&emsp;&emsp;不过需要注意的是，我们本地的 `main` 分支<b>还没有同步到最新状态</b>。

> <b>提示</b>：远程 `main` 和本地 `main` 是两个位置，网页上合并完成不代表本地目录自动更新。

&emsp;&emsp;因为上一章的合并操作是在 GitHub 页面上完成的。GitHub 上的远程 `main` 已经更新了，但<font color=red>本地 `main` 分支还停留在合并之前</font>的版本。如果这个时候直接基于本地旧的 `main` 创建新分支，就可能导致新功能分支缺少上一章刚刚合并进去的代码。

&emsp;&emsp;所以在开始第三章之前，我们先新开一个 Codex 任务窗口，并选择 `main` 分支，让 Codex 帮我们检查当前 Git 状态，并把本地 `main` 更新到远程最新版本。

&emsp;&emsp;可以输入下面这段提示词：

> 💬 <b>输入给 Codex 的提示词</b>

```text
请先帮我同步当前项目的 main 分支到远程最新状态，并做一次基础开发基线检查。

要求：
1. 先确认当前 Git 仓库状态。
2. 如果当前不在 main 分支，请先确认工作区是否干净，再切换到 main。
3. 拉取远程 main 的最新代码。
4. 确认本地 main 是否已经包含上一章合并进来的申请追踪页面改动。
5. 告诉我当前 main 分支的最新 commit hash。
6. 按照项目当前启动方式，启动后端和前端服务。
7. 检查后端健康接口是否正常。
8. 检查前端页面是否可以正常访问。
9. 尽量运行和申请追踪页面相关的测试；如果测试较多，可以优先运行最相关的测试。
10. 如果有前端 lint 或格式检查，也请运行一次。
11. 最后给我一个结论：当前最新 main 是否适合作为多分支协同开发的起点。
12. 不要创建新分支。
13. 不要修改业务代码。
14. 不要提交 commit。
15. 不要推送代码。
```

&emsp;&emsp;接下来我们需要确定本章的多分支开发需求。

&emsp;&emsp;上一章我们已经给「申请追踪」页面增加了状态模块显示管理能力。现在这个页面已经可以让我们根据自己的求职流程，选择哪些状态模块显示，哪些状态模块隐藏。

&emsp;&emsp;不过在实际使用时，我发现「面试」这个状态还可以继续优化。

> <b>提示</b>：这里的两个需求都来自同一个业务页面，所以后面出现相近文件改动是合理现象。

&emsp;&emsp;现在如果一份简历进入了面试阶段，我们只能知道它处于「面试」状态，但页面里并没有一个明确的位置用来记录面试时间。这样后面如果面试机会多了，就不太方便安排时间，也不方便回顾自己接下来要参加哪些面试。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-046.png" width=80%></div>

&emsp;&emsp;所以<font color=red>第一个开发需求</font>是：在申请记录进入「面试」状态后，支持记录对应的面试时间。比如可以选择具体日期和时间，后续在申请追踪页面中直接看到这场面试安排。

&emsp;&emsp;除了面试时间之外，我还希望这个页面可以记录面试题。

&emsp;&emsp;因为真实求职过程中，我们面试完一家公司后，往往会遇到一些值得复盘的问题。比如对方问了哪些技术点、哪些项目细节、哪些业务场景。如果这些内容只是零散记录在聊天软件、备忘录或者文档里，后面复盘起来会很麻烦。

> <b>提示</b>：需求 A 关注时间安排，需求 B 关注面试复盘；它们关联同一条申请记录，但验收目标不同。

&emsp;&emsp;所以<font color=red>第二个开发需求</font>是：给申请记录增加面试题记录能力。我可以手动输入某家公司问过的面试题，并且在页面右上角增加一个「面试题」按钮。点击之后，可以统一查看所有已经记录过的面试题，同时看到这些题目分别来自哪些公司。

&emsp;&emsp;这样后续准备面试时，我们就可以根据历史面试题更有针对性地复盘。

&emsp;&emsp;这两个需求都和「申请追踪」页面有关，但它们的侧重点不同：

```text
需求 A：记录面试时间，帮助安排后续面试
需求 B：记录面试题，帮助复盘和准备后续面试
```

&emsp;&emsp;这里需要说明一下，在真实团队中，如果两个需求都集中在同一个页面、同一个业务模块里，通常不会随意拆给两个完全独立的开发者处理。

&emsp;&emsp;比如「面试时间」和「面试题记录」都属于申请追踪页面里的面试管理能力，它们之间存在一定关联：都和同一份申请记录有关，也都可能涉及申请记录的数据结构、页面展示和编辑入口。因此在真实项目中，这类功能更常见的做法是由同一个开发者或同一个小组统一设计和开发，避免页面交互和数据结构各做各的。

&emsp;&emsp;不过这一章我们是为了演示<font color=red>多分支协同开发</font>，所以会有意把它拆成两个功能分支：

```text
feature/interview-time
feature/interview-questions
```

&emsp;&emsp;这样做的目的不是说真实开发一定要这样分工，而是为了让大家看到：当两个分支都围绕同一个模块开发时，它们更容易修改到相近的文件。前一个分支合并进 `main` 之后，后一个分支如果还停留在旧的 `main` 基础上，就可能缺少前一个功能的代码，甚至在合并时出现<font color=red>冲突</font>。

### 3.3 创建两个独立 Codex 会话

&emsp;&emsp;明确两个开发需求之后，接下来就要为它们分别创建<font color=red>独立的开发环境</font>。

&emsp;&emsp;这里我们不建议让两个功能都在同一个 Codex 会话、同一个项目目录里连续开发。因为这样虽然也能完成，但不太能体现真实团队协作的状态。

&emsp;&emsp;在真实团队中，开发者 A 和开发者 B 通常是在各自的本地目录或各自的开发环境里工作。A 修改自己的功能分支，B 修改自己的功能分支。两个人不会共用同一个工作区，否则文件改动、依赖安装、服务端口、Git 分支状态都会互相影响。

&emsp;&emsp;所以这一章我们使用两个 Codex 会话来模拟两个开发者：

```text
Codex 会话 A：负责 feature/interview-time
Codex 会话 B：负责 feature/interview-questions
```

&emsp;&emsp;每个会话都需要使用独立的 worktree。可以简单理解为：worktree 是 Git 提供的一种“同一个仓库，多份工作目录”的能力。它允许我们基于同一个 Git 仓库，同时检出多个不同分支，并且每个分支都有自己独立的文件目录。

> <b>Git 提示</b>：这里对应的是 <code>Worktree</code> 和 <code>Branch</code>。作用是从 `main` 拉出独立功能分支，并放到另一个本地目录里开发。

&emsp;&emsp;这样做的好处是：

```text
会话 A 修改面试时间功能时，不会影响会话 B 的文件；
会话 B 修改面试题功能时，也不会污染会话 A 的工作区；
两个分支可以同时存在、同时开发、分别提交；
后续合并时再通过 PR 进入 main。
```

&emsp;&emsp;这里要注意，<font color=red>不要让两个 Codex 会话共用同一个 worktree</font>，也不要让两个 worktree 检出同一个本地分支。否则很容易出现文件状态混乱，甚至 Git 直接不允许同一个分支同时被多个 worktree 使用。

&emsp;&emsp;我们可以先创建第一个工作树，用来模拟开发者 A，负责「面试时间」功能。

&emsp;&emsp;这里需要把两种方式分清楚：如果课堂希望工作树固定放在 `/Users/xiaoyang/codex_workspace/resume_matcher` 的平级目录，我们就在主工作区的 Local 会话里让 Codex 执行普通 Git worktree 创建操作；如果使用 Codex App 自带的 Worktree 模式，它也能隔离任务，但位置由 Codex 管理，通常在 `$CODEX_HOME/worktrees`，不能写成通过提示词指定到课程平级目录。

&emsp;&emsp;本章为了让两个工作树目录固定在课堂演示路径下，采用第一种方式。我们先在主工作区的 Codex 会话中输入：

> 💬 <b>输入给 Codex 的提示词</b>

```text
你现在模拟开发者 A，负责开发「面试时间记录」功能。

要求：
1. 请在主工作区 /Users/xiaoyang/codex_workspace/resume_matcher 中操作。
2. 先确认当前项目的 main 分支已经是最新状态。
3. 不要在当前 main 工作目录里直接开发。
4. 请用普通 Git worktree 创建独立工作树，不使用 Codex App 自动管理的 Worktree 位置。
5. 执行类似 git worktree add -b feature/interview-time ../resume_matcher_interview_time main 的操作。
6. worktree 目录名使用：resume_matcher_interview_time。
7. 在这个 worktree 中创建并切换到新分支：feature/interview-time。
8. 确认这个分支的起点是最新 main。
9. 确认新 worktree 工作区干净。
10. 不要修改业务代码。
11. 不要提交 commit。
12. 不要推送代码。
13. 最后告诉我：
   - worktree 的目录位置
   - 当前分支名称
   - 当前分支基于哪个 main commit 创建
```

> <b>Git 提示</b>：这里对应 <code>Worktree</code> 和 <code>Branch</code>。作用是给开发者 A 创建独立目录和功能分支，避免污染主工作区。

&emsp;&emsp;执行完成后，主工作区里的 Codex 会为开发者 A 准备一个独立的工作目录。创建完成后，我们再把这个新目录作为独立项目或独立会话打开，后续开发就发生在 `feature/interview-time` 分支所在的工作树里。

### 3.5 开发者 A 先提交 PR

&emsp;&emsp;接下来我们开始模拟真实团队中的<font color=red>提交流程</font>。

&emsp;&emsp;假设开发者 A 的「面试时间记录」功能先完成开发和测试，那么开发者 A 就先进入提交和 PR 流程。

&emsp;&emsp;这里要注意，即使开发者 A 是第一个准备提交 PR 的人，也不应该直接把分支推送出去就创建 PR。更稳妥的做法是：先把自己的功能保存成一次本地 commit，然后同步最新 `main`，确认当前功能分支仍然可以基于最新主线正常工作，再推送分支并创建 PR。

> <b>提示</b>：A 分支这次没有遇到<font color=red>冲突</font>，不代表这一步可以省略；它仍然要先确认自己基于最新主线。

&emsp;&emsp;因为在真实团队中，开发者 A 开发期间，远程 `main` 也可能已经发生变化。如果 A 不同步最新 `main`，就可能出现一种情况：自己本地测试没问题，但 PR 创建后才发现和最新主线有冲突，或者合并后测试失败。

&emsp;&emsp;所以，提交 PR 前的基本流程是：

```text
功能开发完成
→ 本地 commit
→ 拉取最新 main
→ 合并最新 main 到当前功能分支
→ 重新运行测试
→ 推送功能分支
→ 创建 PR
```

&emsp;&emsp;这一步的核心目的，是确保开发者 A 的功能分支不是基于旧主线提交的，而是已经和最新 `main` 对齐过。

&emsp;&emsp;第三，先把面试时间功能保存成一次本地 commit。这样当前功能就形成了一个清晰的<font color=red>版本快照</font>，后续无论是同步 `main`、处理冲突，还是创建 PR，都有明确边界。

### 3.6 审查并合并开发者 A 的 PR

&emsp;&emsp;在真实团队中，通常不会让开发者写完代码后马上自己合并，而是会由其他成员进行<font color=red>代码审查</font>。代码审查的目的不是重新写一遍代码，而是确认这次改动是否符合需求、是否只修改了应该修改的文件、是否通过测试，以及是否会影响已有功能。

&emsp;&emsp;所以这一节我们用另一个 Codex 会话来模拟 Reviewer，也就是<font color=red>代码审查</font>者。

&emsp;&emsp;这里可以使用主工作区的 Codex 会话，或者新开一个单独的 Codex 会话。它不负责继续开发「面试时间记录」功能，而是站在审查者的角度，检查开发者 A 提交的 PR。

> 💬 <b>输入给 Codex 的提示词</b>

### 3.7 B 同步 main 后提 PR

&emsp;&emsp;开发者 A 的 PR 合并后，B 分支面对的<font color=red>主线已经变化</font>。本节的关键是先同步主线，再把两个功能合并到同一份代码里，而不是直接把旧分支推上去。

![image-20260702163736168](https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/image-20260702163736168.png)

&emsp;&emsp;现在的 `main` 不再只是上一章的申请追踪管理按钮功能，而是已经包含了开发者 A 新增的「面试时间记录」功能。但是开发者 B 的 `feature/interview-questions` 分支是在 A 合并之前创建的，起点还是<font color=red>旧的 `main`</font>，里面并不包含 A 刚刚合并进去的代码，所以开发者 B 不能直接提交 PR。

> <b>提示</b>：B 的关键步骤不是“晚一点推送”，而是先把已经变化的 `main` 合入自己的功能分支。

&emsp;&emsp;如果 B 此时直接把 `feature/interview-questions` 推送并创建 PR，可能会出现两个问题。

&emsp;&emsp;第一，可能出现 Git <font color=red>冲突</font>。因为 A 和 B 都修改了申请追踪页面相关代码，如果它们改到了同一个文件、同一段逻辑，Git 就可能无法自动合并。

&emsp;&emsp;第二，即使 Git 没有<font color=red>冲突</font>，也可能出现逻辑问题。比如 A 增加了 `interview_at` 字段，B 增加了 `interview_questions` 字段，如果两个功能都修改了申请记录的数据结构和保存逻辑，就需要确认这两个字段能否同时存在，保存时会不会互相覆盖，页面展示是否正常。

> <b>提示</b>：Git 能自动合并文件，不代表业务逻辑一定合并正确，所以同步后仍然要重新跑测试。

&emsp;&emsp;所以，开发者 B 在提交 PR 之前，必须先把最新的 `main` 同步到自己的功能分支中。

&emsp;&emsp;也就是说，B 的提交流程应该是：

```text
B 功能开发完成
→ 本地 commit
→ 拉取最新 main
→ 合并最新 main 到 feature/interview-questions
→ 处理可能出现的冲突
→ 重新运行测试
→ 确认 A 的面试时间功能没有被破坏
→ 确认 B 的面试题功能仍然正常
→ 推送 feature/interview-questions
→ 创建 PR
```

&emsp;&emsp;这里和开发者 A 的流程很像，但 B 多了一个关键点：此时最新 `main` 已经包含了 A 的功能，所以 B 同步 main 的过程，实际上就是把 A 已经合并进主线的代码同步到自己的分支里。

&emsp;&emsp;我们回到开发者 B 的 Codex 会话，也就是 `feature/interview-questions` 分支所在的 worktree，输入下面这段提示词：

> 💬 <b>输入给 Codex 的提示词</b>

```text
现在开发者 B 的「面试题记录」功能已经开发完成，请准备提交 PR。

要求：
1. 确认当前分支是 feature/interview-questions。
2. 确认当前 worktree 是 /Users/xiaoyang/codex_workspace/resume_matcher_interview_questions。
3. 确认工作区只有面试题记录功能相关改动。
4. 运行相关测试和代码检查。
5. 提交信息使用：feat: add interview question tracking。
6. 提交完成后告诉我 commit hash。
7. 拉取远程 main 的最新代码。
8. 将最新 main 合并到当前 feature/interview-questions 分支。
9. 如果出现冲突，先列出冲突文件和原因，等我确认后再处理。
10. 合并完成后重新运行相关测试。
11. 确认面试题记录功能仍然正常。
12. 确认刚刚进入 main 的面试时间记录功能没有被破坏。
13. 确认上一章的状态模块显示管理功能没有被破坏。
14. 测试通过后，将 feature/interview-questions 分支推送到我的 GitHub fork 仓库。
15. 不要直接合并到 main。
16. 如果可以创建 PR，请先给我 PR 标题和 PR 描述草稿，等我确认后再创建。
```

> <b>Git 提示</b>：这里同样串起 <code>Add</code>、<code>Commit</code>、<code>Pull</code>、<code>Merge</code>、<code>Push</code> 和 <code>PR</code>。区别是此时 `main` 已经包含开发者 A 的改动，<code>Pull</code> 拉取最新主线后再 <code>Merge</code> 到当前分支，更容易暴露需要处理的冲突。

&emsp;&emsp;这里仍然是先提交本地 commit，再同步最新 `main`。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/OpenAICodex%E8%BF%9B%E9%98%B6%E5%AE%9E%E6%88%98/2026-07-02/screenshot-056.png" width=80%></div>

&emsp;&emsp;从 Codex 的结果可以看到，开发者 B 在合并最新 `main` 时确实出现了冲突。

&emsp;&emsp;这也符合我们的预期。因为 A 和 B 虽然开发的是两个不同功能，但它们都围绕「申请追踪」页面的面试管理能力展开，所以很容易修改到同一批文件。

&emsp;&emsp;A 分支主要新增的是 `interview_at` 字段，用来记录面试时间；B 分支主要新增的是 `interview_questions` 字段，用来记录面试题。两边都需要修改申请记录的数据结构、后端接口 schema、前端详情弹窗和相关测试，因此 Git 无法自动判断应该如何合并。

&emsp;&emsp;这里要注意，出现<font color=red>冲突</font>并不代表代码坏了，而是 Git 需要我们明确告诉它：两边的改动应该如何组合。

> <b>提示</b>：处理冲突时不要只追求“冲突标记消失”。本章的正确结果是同时保留 `interview_at` 和 `interview_questions`，并重新验证面试时间、面试题和状态模块显示管理三个功能。

&emsp;&emsp;在这个案例里，正确的处理方是两个功能都要保留。最终合并后的申请记录既应该支持 `interview_at`，也应该支持 `interview_questions`。

&emsp;&emsp;所以我们继续让 Codex 处理<font color=red>冲突</font>，但要明确告诉它不要简单选择某一边，而是合并两边逻辑。

> 💬 <b>输入给 Codex 的提示词</b>

```text
请处理当前 merge 冲突，但不要简单选择某一边，需要合并两边逻辑。

要求：
1. 保留 main 中已经合并的 `interview_at` 面试时间功能。
2. 保留当前分支中的 interview_questions 面试题记录功能。
3. 冲突文件中如果同时涉及 Application / ApplicationUpdate 类型、后端 schema、数据库补齐逻辑、接口测试和详情弹窗，需要合并两边逻辑，让两个字段同时存在。
4. 解决冲突后，先告诉我每个冲突文件是如何处理的。
5. 运行相关后端测试、前端测试、类型检查和 lint。
6. 确认以下功能都正常：
   - 面试时间可以保存和展示；
   - 面试题可以添加、删除和统一查看；
   - 申请追踪状态模块显示管理功能没有被破坏。
7. 冲突解决并测试通过后，请启动后端和前端服务，让我在浏览器里查看效果。
8. 如果当前 worktree 的默认端口已经被其他服务占用，请自动换一个可用端口，并告诉我前端访问地址和后端接口地址。
9. 解决冲突后不要自动推送。
10. 如果需要创建冲突解决 commit，请先说明，不要自动提交。
```

> <b>Git 提示</b>：这里对应 <code>Merge</code> 冲突处理。作用是把 `main` 和当前功能分支的改动组合起来，而不是简单选择某一边。

&emsp;&emsp;这说明这次冲突不能简单理解为“保留 A”或“保留 B”，而是需要把两个分支的逻辑合并到一起：既保留 `interview_at` 面试时间字段，也保留 `interview_questions` 面试题字段。

&emsp;&emsp;不过此时 Git 还提示：

```text
All conflicts fixed but you are still merging
```

&emsp;&emsp;此时 Git 已经提示<font color=red>冲突</font>文件都处理完了，但还处在 merge 未完成状态。所以我们不需要重新开始，只需要完成这次 merge commit。

&emsp;&emsp;完成 merge commit 后，`feature/interview-questions` 分支就正式包含了最新 `main` 中的面试时间功能，同时也保留了自己开发的面试题功能。接下来就可以把这个分支推送到远程，并创建 PR。

> 💬 <b>输入给 Codex 的提示词</b>

```text
当前冲突已经解决，页面也能同时看到面试时间和面试题功能。

请继续完成 merge commit，然后推送当前分支并准备创建 PR。

要求：
1. 确认当前分支是 feature/interview-questions。
2. 确认当前状态是 All conflicts fixed but you are still merging。
3. 检查所有冲突文件已经没有冲突标记。
4. 运行相关测试和代码检查。
5. 确认以下功能都正常：
   - 面试时间可以保存和展示；
   - 面试题可以添加、删除和统一查看；
   - 申请追踪状态模块显示管理功能没有被破坏。
6. 测试通过后，完成本次 merge commit。
7. merge commit 完成后告诉我新的 commit hash。
8. 将 feature/interview-questions 分支推送到我的 GitHub fork 仓库。
9. 不要直接合并到 main。
10. 如果可以创建 PR，请先给我 PR 标题和 PR 描述草稿，等我确认后再创建。
```

> <b>Git 提示</b>：这里对应 <code>Add</code>、<code>Merge Commit</code> 和 <code>Push</code>。<code>Add</code> 标记冲突文件已经处理完，<code>Merge Commit</code> 完成合并提交，<code>Push</code> 再把当前分支上传到 GitHub。

> <b>Git 提示</b>：这里对应 <code>PR</code>。作用是把 `feature/interview-questions` 的改动提交到 GitHub 页面，并说明它已经同步最新 `main`。

&emsp;&emsp;和开发者 A 的 PR 不同，B 的 PR 已经包含了一次同步最新 `main` 后产生的 <font color=red>merge commit</font>。也就是说，B 分支不是直接基于旧的 `main` 提交，而是在合并了 A 的面试时间功能之后，再把自己的面试题功能提交出来。

&emsp;&emsp;这正是<font color=red>多分支协同开发</font>中最重要的一点：

```text
后提交的人，必须先同步最新 main；
如果出现冲突，要解决冲突并重新测试；
确认所有相关功能都正常后，再创建 PR。
```

### 3.8 审查并合并开发者 B 的 PR

&emsp;&emsp;上一节中，开发者 B 已经把 `feature/interview-questions` 分支推送到了 GitHub，并创建了 PR。

&emsp;&emsp;这个 PR 和开发者 A 的 PR 有一个明显不同：开发者 A 的 PR 比较简单，只包含「面试时间记录」功能本身；而开发者 B 的 PR 不仅包含「面试题记录」功能，还包含了同步最新 `main` 后产生的 merge commit。

&emsp;&emsp;也就是说，开发者 B 的分支里已经包含了开发者 A 合并进 `main` 的面试时间功能，并且在解决冲突后，把<font color=red>两个功能合并到了一起</font>。

> <b>提示</b>：审查 B 的 PR 时，最重要的是看两个功能能否同时存在，而不是只看面试题按钮有没有出现。

&emsp;&emsp;所以审查这个 PR 时，我们不能只看「面试题记录」功能本身，还要重点确认两个功能是否能同时正常工作。

&emsp;&emsp;我们可以继续使用 Reviewer 会话来审查开发者 B 的 PR。输入下面这段提示词：

> 💬 <b>输入给 Codex 的提示词</b>

```text
请帮我审查刚刚创建的 feature/interview-questions PR。

要求：
1. 确认 PR 的 base 分支是 main，compare 分支是 feature/interview-questions。
2. 查看本次 PR 修改了哪些文件。
3. 检查改动是否主要和「面试题记录」功能相关。
4. 检查这个 PR 是否已经同步了包含 `interview_at` 的最新 main。
5. 检查冲突解决是否合理：既保留 `interview_at` 面试时间功能，也保留 `interview_questions` 面试题功能。
6. 检查是否误提交了无关文件，比如配置文件、缓存文件、数据库文件或密钥。
7. 查看测试结果；如果有必要，可以重新运行和申请追踪页面相关的测试。
8. 给出审查结论：这个 PR 是否可以合并。
9. 如果可以合并，请先告诉我合并前需要确认的事项，不要直接合并。
```

> <b>Git 提示</b>：这里对应 <code>Diff/PR</code> 审查。作用是确认面试题功能、面试时间功能和上一章功能能否同时保留。

&emsp;&emsp;这里的<font color=red>审查重点</font>和开发者 A 的 PR 不完全一样。

&emsp;&emsp;开发者 A 的 PR 主要检查它有没有正确实现「面试时间记录」。而开发者 B 的 PR 除了检查「面试题记录」功能，还要重点检查它在同步最新 `main`、解决冲突之后，是否保留了开发者 A 已经合并进去的功能。

&emsp;&emsp;也就是说，合并开发者 B 的 PR 前，至少要确认：

```text
1. PR 方向正确：feature/interview-questions → main；
2. PR 已经包含最新 main，也就是已经包含 `interview_at` 面试时间功能；
3. 冲突解决方式不是简单选择某一边，而是保留了 `interview_at` 和 `interview_questions`；
4. 面试题可以正常添加、删除和统一查看；
5. 面试时间仍然可以正常保存和展示；
6. 状态模块显示管理功能没有被破坏；
7. 没有提交 .env、数据库、缓存、日志、密钥等不该提交的文件；
8. GitHub 显示可以合并。
```

&emsp;&emsp;不过 Reviewer 也指出了两个需要后续修正的小问题：一是部分文档没有同步补充上一版的面试时间字段说明，二是测试文件里存在重复导入。这些问题不会阻塞当前 PR 合并，但说明代码审查不仅要看功能能不能跑，还要关注文档、测试和代码整洁度。

&emsp;&emsp;如果是在真实团队中，我们可以根据团队规范选择两种处理方式：一种是在合并前继续让开发者 B 补一个小提交，把这些问题一起修掉；另一种是在确认不影响功能的情况下先合并，再单独开一个后续任务清理。

> 💬 <b>输入给 Codex 的提示词</b>

```text
确认可以合并这个 PR。
请将 feature/interview-questions 合并到 main。
```

> <b>Git 提示</b>：这里对应 <code>Merge</code>。作用是把审查通过的 `feature/interview-questions` 合并进 `main`。

&emsp;&emsp;合并完成后，远程 `main` 会再次发生变化。

```text
合并开发者 A 后：
main = 申请追踪管理按钮功能 + 面试时间记录功能

合并开发者 B 后：
main = 申请追踪管理按钮功能 + 面试时间记录功能 + 面试题记录功能
```

### 3.9 本章回顾

&emsp;&emsp;到这里，本章的<font color=red>多分支协同开发</font>流程就完整走通了：

```text
从最新 main 创建两个 worktree；
两个 Codex 会话分别开发不同功能；
开发者 A 先提交 PR 并合并；
开发者 B 同步 A 合并后的最新 main；
B 解决冲突、重新测试、提交 PR；
Reviewer 审查并合并 B；
最终 main 同时包含两个功能。
```

&emsp;&emsp;这就是一个比较完整的<font color=red>多分支协同开发</font>流程。

&emsp;&emsp;本章可以收束成一句话：<font color=red>多分支协同开发</font>不是把事情变复杂，而是把“谁在改什么、基于哪个主线、什么时候合并、合并后如何验证”这几件事拆清楚。开发者 A 和开发者 B 可以并行推进，但每个人提交 PR 前都要确认自己的分支已经和最新 `main` 对齐；合并 PR 前也要看方向、改动范围、diff、测试结果、敏感文件和 GitHub 的合并状态。

&emsp;&emsp;到最后，我们拿到的是一条可以复用的流程：先从稳定 `main` 创建独立 worktree，再让不同 Codex 会话分别开发；先让 A 合并，B 再同步最新主线；遇到冲突时同时保留两边业务逻辑；最终让 `main` 同时包含申请追踪状态模块显示管理、面试时间记录和面试题记录三个功能。备课实测中，最终后端相关测试为 52 个通过，前端相关测试为 4 个测试文件、24 个测试通过。

## <center>第四章：课程回顾：从一次演示变成可复用方法</center>

&emsp;&emsp;第三章结束时，我们已经把多分支协同开发完整走通：从最新 `main` 创建两个独立 `worktree`，让两个 Codex 会话分别开发面试时间和面试题功能，再按顺序通过 PR 合并回主线。到这里，课程已经不只是跑通一个开源项目，而是走完了从接手、改造到协同合并的<font color=red>完整工程路径</font>。这一章不再新增实践任务，也不再引入新的案例项目，而是把前面三章的操作提炼成一套可复用方法。

&emsp;&emsp;换句话说，Codex App 在这门课里的位置不是替代 Git、GitHub、测试和代码审查，而是放进这些流程里，参与读项目、改功能、跑检查和看改动。最终能不能合并，仍然要看事实、看运行结果、看 diff、看 PR 方向和测试结果。

&emsp;&emsp;第一章解决的是“陌生开源项目从哪里下手”。我们没有一上来就改代码，而是先让 Codex App 阅读 GitHub 仓库，理解项目用途、结构、入口文件和阅读顺序；接着 Fork 到自己的 GitHub 账号，再 Clone 到本地；最后配置前后端服务、验证健康检查，并体验主简历、岗位描述、大模型优化和申请追踪这条业务主线。第二章解决的是“已有项目怎样做小范围二次开发”：先做开发前基线检查，再创建 `feature/tracker` 功能分支；进入计划模式让 Codex App 先读代码、给方案、列测试；实现完成后先验收页面和测试，再提交本地版本、推送分支、创建 PR、查看 diff，最后确认 PR 方向和合并条件。

> <b>提示</b>：第一章和第二章的差别很关键。第一章证明项目能跑，第二章证明改动可控；一个解决接手问题，一个解决开发问题。

&emsp;&emsp;第三章解决的是“多个功能同时推进时怎样不互相污染”。我们从最新 `main` 创建两个独立 `worktree`，让开发者 A 和开发者 B 分别处理面试时间和面试题功能。A 先合并后，B 必须同步最新 `main`，处理冲突并重新测试，确认 `interview_at` 和 `interview_questions` 同时保留，再把 B 的 PR 合并回主线。

&emsp;&emsp;这三章串起来后，可以压缩成五个步骤：<font color=red>读、改、验、审、合</font>。它们不是口号，而是一个真实项目从陌生到可合并的推进顺序。

&emsp;&emsp;这张表也能反过来帮助我们判断 Codex App 的边界。Codex App 可以辅助读代码、定位文件、修改实现、运行命令、总结 diff 和准备 PR，但它不能替我们跳过工程判断。比如，它说测试通过，我们仍然要看测试命令和结果；它说可以合并，我们仍然要看 PR 的 base、compare、diff 和权限；它说冲突已经解决，我们仍然要确认两个业务功能都保留下来。

> <b>提示</b>：本课里最重要的分界线是“辅助执行”和“工程判断”。Codex App 可以帮我们执行很多操作，但是否进入下一步，要由运行结果、diff、测试和 PR 状态共同决定。

&emsp;&emsp;所以这套方法的核心不是“让 Codex App 一次做完”，而是把每一步都放进<font color=red>可检查的位置</font>：读项目时看结构，改功能时看计划，验证时看结果，审查时看 diff，合并时看主线状态。这样即使换成另一个开源项目，我们仍然知道该从哪里开始、在哪些位置停下来确认、哪些步骤必须留下证据。

&emsp;&emsp;把整门课压缩到最后，我们需要记住四条原则：事实优先，项目是否可用要看页面、接口、测试和运行结果；审查优先，代码是否适合合并要看 diff、PR 方向、敏感文件和测试记录；隔离优先，一个功能用一个功能分支，并行功能用独立 `worktree`；复用流程，换项目时可以换仓库、换页面、换功能，但“读、改、验、审、合”的顺序不变。

&emsp;&emsp;为了把复盘压缩成记忆锚点，这里只保留五个步骤。下一次接手陌生项目时，我们先想起这五个停顿点。

&emsp;&emsp;最终我们可以用三句话检查自己是否真正掌握了这门课。第一，遇到陌生开源项目，我们先让 Codex App 读结构、用途、入口和运行方式，再 Fork、Clone 和跑通。第二，准备改功能时，我们先创建分支或 worktree，让 Codex App 进入计划模式，确认范围后再动手。第三，改完以后，我们先看运行结果和测试，再看 diff 和 PR，最后才合并进稳定 `main`。

&emsp;&emsp;这些判断会让 Codex App 从“帮我操作一次”的工具，变成可以嵌入工程流程的助手。它能参与阅读、修改和验证，但不会取消我们对<font color=red>事实、审查、隔离和合并顺序</font>的责任。

> <b>提示</b>：以后换一个开源项目时，不必照搬本课的路径、账号和演示状态。我们只需要保留这条顺序：先读清楚，再小步改，改完验证，合并前审查，并行时隔离。

&emsp;&emsp;到这里，《OpenAICodex进阶实战》的主线就收束完成了。开场篇先说明本课为什么要把 Codex App 放进 Git、GitHub 和真实工程流程；第一章把开源项目读懂并跑起来；第二章完成一次单功能二次开发；第三章把流程升级到多分支协同；第四章把这些步骤压缩成“读、改、验、审、合”。这套方法的价值在于，它不依赖某一个固定 PR 编号、某一次页面状态或某个账号，而是可以迁移到下一次真实项目接手和二次开发中。
