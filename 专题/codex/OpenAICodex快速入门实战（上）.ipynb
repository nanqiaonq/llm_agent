# <center>OpenAI Codex 快速入门实战</center>

&emsp;&emsp;这门课我们从零开始,把 OpenAI Codex App 真正用起来——不是泛泛地讲它有多强,而是实打实地装好它、跟它对话、把一个个真实的小项目交给它做出来。我们会先花一点篇幅认识 Codex 到底是个什么东西(它跟我们印象里"会写代码的 AI"不太一样),然后就一头扎进操作:装好登录、做一个能增删改查的待办清单网页、用计划模式搭一套带前后台的餐厅点餐收银系统、用 git 给代码存档和回退、让它自己生成一张品牌 logo,给它立一份项目规约 <font color=red>AGENTS.md</font> 让它每次都按我们的规矩办事;再往后,我们还会让多个 agent 同时开干、用 Computer Use 操控电脑上的软件、配置每天自动跑的定时任务,甚至用手机远程指挥电脑端的 Codex。

## <center>第一章 先认识一下 Codex</center>

&emsp;&emsp;动手之前,我们先把"Codex 到底是什么"弄明白。很多人一听 Codex,脑子里浮现的是"一个更强的代码补全",结果一上手处处别扭。这一章不长,就把它今天的真实身份、跑的什么模型、怎么个工作法,几句话讲清楚,后面操作起来就顺了。

### 1.1 此 Codex 非彼 Codex

&emsp;&emsp;今天的 Codex,和 2021 年那个 Codex,其实是同名不同物的两样东西。最早那个 Codex 是 OpenAI 2021 年 6 月推出的代码补全模型,当过 GitHub Copilot 最初版本的引擎,2023 年 3 月就被正式弃用了。到 2025 年 5 月,OpenAI 把"Codex"这个名字重新拿出来用,指的是一套全新的、能自己干活的编程 agent。两个 Codex 之间没有任何技术上的传承,共用的只是一个名字而已。

&emsp;&emsp;所以我们现在说 Codex,基本都是指 2025 年这套新东西。这里有个小提醒:网上搜 Codex 资料时,大量 2021 到 2023 年的老文章还在讲那个早就退役的旧模型,看一眼发布时间就能分辨——2023 年以前的内容基本都过时了,别被带回"它只是个写代码的小工具"的老印象里。把这条演进按时间摆开,"同名不同物"就更直观了。

### 1.2 它今天跑什么模型

&emsp;&emsp;模型是 Codex 的"大脑",我们先认准它现在用的是哪一颗。Codex 的模型底座换过好几次,2025 年 5 月初上线时用的是 codex-1,后来一路换到 GPT-5.3-Codex、GPT-5.4,到 2026 年 4 月 GPT-5.5 发布后,成了 Codex App 当前的默认主力。我们今天在 App 里点开就用的,就是 GPT-5.5。

&emsp;&emsp;顺带澄清一个常见的口误:<b>并没有一个叫"GPT-5.5-Codex"的单独型号</b>。带 `-Codex` 后缀的命名到 GPT-5.3-Codex 那一代就停了,后面的 GPT-5.4、GPT-5.5 都是直接用通用模型,没有再出"Codex 专属版"。我们在 App 的模型菜单里看到的是 GPT-5.5、GPT-5.4 这些通用型号。

&emsp;&emsp;还有一个数字值得记一下,就是它的<b><font color=red>上下文窗口</font></b>——简单说就是它一次能"看见"多少内容,装满了就开始忘前面的。本机实测,GPT-5.5 在 Codex App 里的上下文窗口是 <b>272000</b> 个 token(2026 年 6 月本机实测值)。这个数字网上有好几种说法,有人说 40 万、有人按 API 文档说 100 万,但我们这门课讲的是桌面 App,以 App 里实测的 272000 为准就好。

> <font size=2>**「名词解释」<font color=red>上下文窗口</font>(context window)** — 模型一次能处理的最大文本量,以 token 计;装满之后早先的内容会被挤出去"忘掉",直接决定它能同时记住多少代码和对话。</font>

### 1.3 codex自动规划执行的agent

&emsp;&emsp;如果只用一句话概括 Codex 的工作方式,可以这么说:我们把目标讲清楚,它负责去项目里执行,最后我们来检查结果。这个过程不像普通聊天那样问一句答一句,更像是把一件具体的活交给它做。

&emsp;&emsp;比如我们说“帮我做一个待办清单页面”,Codex 不只是回一段代码给我们。它会先理解需求,再去看当前项目结构,创建或修改文件,必要时还会启动服务、查看报错、继续调整。等它做完之后,会把改了哪些文件、实现了什么效果、还有没有需要注意的地方告诉我们。我们要做的,不是盲目相信它,而是看结果、看 diff、试运行,确认没问题再保存下来。

&emsp;&emsp;所以用 Codex 时,最重要的不是把每一步命令都背下来,而是学会把任务说清楚。简单任务可以直接交给它,复杂任务可以先让它列计划,中途发现方向不对也可以打断它、补充要求。后面几个实战章节,其实都是在练这件事:怎么把一个真实需求讲明白,再怎么检查它交回来的结果。

### 1.4 Codex 不止一个入口

&emsp;&emsp;还有一个容易混淆的地方:Codex 不是只有一个入口。我们这门课主要用的是桌面 App,因为它有图形界面,能看到任务列表、文件改动、运行过程,对刚上手的人最直观。除此之外,也有 ChatGPT 里的 Codex 入口、终端里的 Codex CLI,以及编辑器里的相关扩展。

&emsp;&emsp;这些入口的定位不完全一样。桌面 App 更适合跟着界面一步步操作,适合做课程演示和本地项目实践;CLI 更偏开发者和自动化场景;编辑器扩展则更适合边写代码边让它帮忙。我们这门课先抓住桌面 App 这一条主线,把基本用法跑通,以后再去理解其他入口就容易多了。

> <font size=2>**「名词解释」<font color=red>CLI</font>(Command-Line Interface,命令行界面)** — 在终端里用文字命令操作程序的方式,跟图形界面相对。Codex CLI 是终端里的 agent 入口,更适合开发者和自动化场景;本课不用它,用图形界面的桌面 App。</font>

## <center>第二章 下载与安装 Codex App</center>

&emsp;&emsp;认识完了,这一章我们把 Codex App 装到自己电脑上。整个过程就三步:从官方入口下载、按自己的系统装好、第一次打开用 ChatGPT 账号登录。如果手上已经有一个能正常登录、能正常付费的 ChatGPT 账号,这一章会很轻松。

### 2.1 官方下载入口

&emsp;&emsp;Codex App 只认一个官方下载入口,记住这一个就行:

```
https://openai.com/zh-Hans-CN/codex/get-started/
```

&emsp;&emsp;打开后是"开始使用 Codex"的下载页,它会自动认出我们的系统,直接给出对应的版本,点「下载 macOS 版本」或「下载 Windows 版本」即可。这条习惯看着不起眼,但很重要:网上那些"Codex 增强版""Codex 远程工具"之类的第三方包,都不在 OpenAI 的保护范围内,只从官方页面拿安装包,能避开一大类风险。

## <center>第三章 初识 Codex App 界面</center>

&emsp;&emsp;装好登录进来了,这一章我们先把界面认熟、把几个该设的设好,免得后面边操作边找按钮。我们会看清主界面分几块、把语言设成中文、选对工作模式、看一眼自己的额度和计费、认识模型菜单,最后把 Codex 的权限档位搞明白。

### 3.1 主界面三大区域

&emsp;&emsp;界面分三大块。<b>左侧是「导航栏」</b>,自上而下是新对话、搜索、插件、自动化、Codex 移动版等入口;下半段的「对话」区是历史会话列表,刚装好是空的;左下角是设置入口。<b>中间是工作对话区</b>,我们跟 Codex 发命令、它回我们,都在这儿。<b>右侧是当前对话的侧边栏</b>,展示对话内容、以及它产出的文件和来源,右上角有个"显示/隐藏侧边栏"的按钮,可以收起或展开它。

&emsp;&emsp;关于额度,有几件事现在知道了,后面就不会一惊一乍。<b>第一,从 2026 年 4 月起,Codex 按 token 用量计费</b>(不再按消息条数算),任务越复杂、来回越多,烧得越快。<b>第二,有几类操作格外费额度</b>:生图大约是普通文字任务的 3 到 5 倍;电脑操控每走一步都要把屏幕截图喂进去,步数一多消耗就涨得很猛;多个 agent 同时干活时各自独立计费。<b>第三,额度是按"每 5 小时的滚动窗口"恢复的</b>——某一刻用掉的额度,会在 5 小时之后自动腾出来。所以真撞到上限也别急着升档,排开任务、过一阵再发就行。

> <font size=2>**「名词解释」<font color=red>token</font>(令牌,模型处理文字的最小单位)** — 模型把文字切成的小块,大致一个汉字对应 1 到 2 个 token;按 token 计费,就是按这些小块的总数算钱。</font>

### 3.5 选择模型

&emsp;&emsp;回到主界面,点对话区下方的「5.5 中」,能看到当前有哪些模型、以及它们对应的推理强度。这个「5.5 中」的意思是:用 GPT-5.5 模型、推理强度选"中"。

## <center>第六章 用 Git 管理代码版本</center>

&emsp;&emsp;点餐系统做出来了,接下来我们想再美化美化。但改着改着难免担心:万一改坏了、或者甲方反而更喜欢"第一版",想退回去却退不了。这就需要给代码<b>存档</b>——用 git 管起来,想回哪个版本都行。这一章我们就在点餐系统上,让 Codex 帮我们用 git 管理版本。

### 6.1 让 Codex 用 git 管理版本

### 6.6 git 不是万能的:.gitignore

&emsp;&emsp;git 能回滚代码,但<b>一般回滚不了数据库里的数据</b>。因为工程上,git 通常只管代码文件、配置模板、前端文件、说明文档、数据库建表脚本这些;它不管 `restaurant.db` 这类数据库文件、用户真实下的订单、运行时生成的数据。我们点开项目里的 `.gitignore` 文件就能看明白。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch6-gitignore%E6%96%87%E4%BB%B6-76663c18.png" width="80%"></div>

&emsp;&emsp;这个 `.gitignore` 是 Codex 帮我们建 git 仓库时就一起建好的,作用是告诉 git:<b>哪些文件不要纳入版本管理</b>。里面大致是这些:

```
.venv/          不提交 Python 虚拟环境
__pycache__/    不提交 Python 缓存
*.py[cod]       不提交 Python 编译缓存
.pytest_cache/  不提交测试缓存
.ruff_cache/    不提交代码检查缓存
restaurant.db   不提交本地数据库
*.db            不提交所有数据库文件
.DS_Store       不提交 Mac 系统隐藏文件
```

&emsp;&emsp;所以记住一句话:<b>git 主要负责回滚代码,数据库数据一般不跟着 git 走</b>。真实项目里,数据库是靠备份和迁移脚本来管的,而不是把数据库文件直接塞进仓库。

> <font size=2>**「名词解释」<font color=red>.gitignore</font>(git 忽略清单)** — 一份放在项目里的文本文件,逐行列出"不要纳入版本管理"的文件和目录(虚拟环境、缓存、数据库文件等);Codex 建仓库时会自动生成一份合理的默认清单。</font>

## <center>第八章 给 Codex 立规矩:AGENTS.md</center>

&emsp;&emsp;前面几章我们都是"现说现做",每开一个新项目、新会话,都得重新交代一遍背景。我们想要的是:让 Codex 一进项目,就知道"这个项目该怎么干"。这正是这一章的主角——<b>AGENTS.md</b>。它是一份放在项目里的规约文件,相当于给 Codex 的"项目说明书"。

### 8.1 为什么需要 AGENTS.md

&emsp;&emsp;`AGENTS.md` 一般放在<b>项目根目录</b>,它解决的核心问题是:让 Codex 每次进项目时,都知道"这个项目应该怎么做事"。没有它,Codex 就更容易"靠猜"干活,常见的麻烦有这么几类:

1. <b>每次都要重新解释项目</b>——比如怎么启动、前后端在哪、接口在哪、用什么命令测试。

2. <b>容易改错地方</b>——可能把逻辑写到不合适的文件里,或者误改了配置、数据库、构建产物。

3. <b>代码风格不统一</b>——这次一种写法,下次又换一种。

4. <b>不知道哪些文件不能碰</b>——比如 `.env`、数据库文件、上传的文件、第三方库目录。

5. <b>不知道改完要怎么验证</b>——可能只改代码,不跑测试、不启动、不检查报错。

6. <b>跨对话记不住规则</b>——新开一个任务后,它不一定记得我们之前定过的约定。

&emsp;&emsp;`AGENTS.md` 就像给 Codex 的项目操作手册,把上面这些一次性写清楚,让它<b>少问、少猜、少乱改</b>。一份典型的 AGENTS.md 一般会包含这几块内容:项目说明(用途和技术栈)、安装与启动(依赖怎么装、服务怎么起)、测试命令(改完跑什么来验证)、代码规范(命名、格式、目录约定)、安全与限制(哪些文件别动、哪些命令别跑)、提交规范(是否允许提交、提交信息格式)。

### 8.2 让 Codex 自己生成 AGENTS.md

&emsp;&emsp;AGENTS.md 看着内容不少,但我们其实不用自己一字一句写——<b>直接让 Codex 在当前项目下生成就行</b>。在点餐系统项目里输入"你总结下当前项目适合放进 agents.md 的内容,让我先看一下"。

&emsp;&emsp;AGENTS.md 是<b>单个项目</b>的规则。除此之外,我们还能设一份<b>所有项目都遵守</b>的全局指令。点「项目」→「个性化」,能看到一项「自定义指令」。

&emsp;&emsp;这里适合写一些我们个人长期的编程偏好,比如:

- 回答尽量简洁,先给结论。

- 修改代码时优先保持项目现有结构,不要过度设计。

- 不要主动引入新依赖,除非确实必要。

- 不要自动执行 git push。

- 只有我明确要求时才提交 commit。

- 改完代码后说明改了哪些文件、为什么改、如何验证。

- 遇到不确定的地方,先基于现有代码做最小改动。

### 8.5 个人偏好 vs 项目规则

&emsp;&emsp;到这儿,我们手上就有了两份规约,分工很清楚:

- <b>个性化里的"自定义指令" = 我们个人的长期偏好</b>,跟着我们这个人走,所有项目通用。

- <b>项目里的 AGENTS.md = 当前这个项目的具体规则</b>,跟着项目走,换个项目就换一份。

&emsp;&emsp;官方对"长期规则"的建议也是这个思路:把那些需要长期生效的规则,放到个人级或项目级的 AGENTS.md 里,而不是每次重新在对话里写一遍。其实这份全局的自定义指令,落到本机就是一个文件,比如在 macOS 上就是 `/Users/你的用户名/.codex/AGENTS.md`。

&emsp;&emsp;走到这里,我们就把 Codex 从"每次现说现做",带到了"立好规矩、它自动遵守"。回头看前面这一路:我们认识了 Codex、装好登录、熟悉了界面,然后用待办清单、点餐系统两个项目走通了派活、计划、审批、引导,又用 git 学会了存档和回退、用生图给项目添了门面、用 AGENTS.md 给它立了长期规矩。这些都是<b>一个任务交给一个 agent</b> 来做。从下一章开始,我们进阶一步,看看怎么让<b>多个 agent 同时开干</b>,把效率再往上提。

## <center>第九章 多 Agent 并行干活</center>

&emsp;&emsp;前面这几章,我们用 Codex 做点餐系统、管版本、生图、立规约,每件事都是一个任务一个来,排着队干。这一章我们换个节奏——让多个 agent 同时开工。先在<b>单个会话里</b>让它一边并行思考、并行分析,再升级到<b>多个会话加 worktree</b>,让好几个 agent 各干各的、同时往项目里写代码。两种模式干的活不一样,我们一个一个来看。

> <font size=2>**「名词解释」<font color=red>SubAgent</font>(子智能体)** — 由主 agent 临时派出去干一小块活的子任务执行单元。在 Codex 里,一个会话内最多能同时拉起 8 个 SubAgent 并行处理,各干各的,最后由主 agent 汇总。</font>

### 9.1 单会话内并行思考

&emsp;&emsp;之前我们一直是做单一的修改,一次说一件事。可要是想一口气改很多地方,单个 agent 排队执行就慢了。打个比方:做一个蛋糕,一个面点师从头到尾做下来当然行,但如果叫上几个人——一个打蛋糕胚、一个备奶油、一个摆水果,活儿被拆开同时干,时间自然省下来一大截。

&emsp;&emsp;Codex 支持我们直接在一个会话框里并行执行任务,一个会话内它最多能拉起 8 个并行 SubAgent。我们新开一个会话,把要改的几件事一次说清楚,顺带点名让它用几个 agent 并行处理:

```
1.前端页面我觉得菜单中每个菜的模块太大了,帮我改小一点;

2.菜的种类也太少了,再加 5 个菜;

3.给每个菜加一个详情,给出用料以及预计出餐速度。

请使用 3 个 agent 来并行处理这个任务。
```

&emsp;&emsp;看它的任务执行过程输出,就能发现它确实已经按我们说的,拉起了 3 个 agent 同时开始处理了。

&emsp;&emsp;这种主 agent 分派、子 agent 并行、主 agent 汇总的协作方式,叫 supervisor 协作模式,OpenAI 官方的说法是「subagent workflows」。可以这么理解:主 agent 像个工头,把大活拆成几块分给几个 SubAgent 同时干,等大家都交活了,再由它统一收拢、整合成最终结果。

&emsp;&emsp;对照着看就清楚了:这次 Codex 把任务拆给了 3 个并行 SubAgent。一个负责分析前端菜单卡片的渲染逻辑和样式改动点,一个负责分析数据库结构、菜单数据初始化以及后端 API 要扩展哪些字段,还有一个负责分析项目怎么启动、接口怎么验证、测试方案怎么定。三个 agent 都以<b>只读模式</b>做代码调研和方案设计,谁都不直接动文件;等它们各自交了结论,再由 supervisor 统一收拢,把代码改动、验证和结果整合一并完成。

&emsp;&emsp;这也是单会话并行的特点:它们的并行主要发生在<b>思考和调研</b>这一段,真正落笔改代码的还是 supervisor 一个人。所以这种方式特别适合那种需求偏复杂、要读不少代码、要先理清方案的场景。

### 9.3 多会话并行写入与 worktree

&emsp;&emsp;上面那套并行,本质上是<b>思考并行</b>,不是<b>写入并行</b>。接下来我们再往前走一步:开多个会话,让好几个 agent 同时往项目里写文件。

&emsp;&emsp;并行写入虽然更快,但麻烦也跟着来:前一个 agent 刚写好的文件,转头就被另一个 agent 的改动覆盖了。比如 A 改了 `main.py`,B 也改了 `main.py`,两边互不知情,最后要么互相覆盖、要么直接冲突。要让它们各写各的、互不踩脚,我们就得请出 worktree。

> <font size=2>**「名词解释」<font color=red>worktree</font>(工作树)** — Git 提供的多工作目录机制,能为同一个仓库开出好几份相互独立的工作目录,各自挂在不同分支上。对 agent 开发来说,它最大的用处不是版本管理,而是让多个 agent 同时改同一个项目却互不覆盖文件。</font>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch9-worktree%E5%8E%9F%E7%90%86%E5%9B%BE-db2afa8e.png" width="80%"></div>

&emsp;&emsp;在 Codex 里用 worktree 很省事,我们直接新开一个会话,在「启动模式」里选<b>新工作树</b>就行。这里有一点要先记牢:worktree 解决的是「并行写入互相覆盖」这个问题,它<b>并不能完全消除「最终合并时的冲突」</b>。所以最稳妥的用法是——<b>一个 worktree 对应一个清晰的功能</b>,尽量别让不同 agent 去改同一批文件,合并时自然就省心。

&emsp;&emsp;这回我们想让它一口气加几个新功能,正好拆成三件互不重叠的活:

```
1.每个菜都从网上找到对应的图片,做成合适的大小插入菜单内;

2.加一个会员充值的功能,充值 100 元可以抵扣 110 元(虚拟充值即可),
  前端要有充值页面,后台要有会员剩余额度列表;

3.后端没有取消订单或给客户退款的功能,也需要加上去。
```

&emsp;&emsp;我们开三个会话,每个会话挂一棵自己的工作树,分别交给它们一件事去做。

### 9.4 逐个合并回主分支

&emsp;&emsp;三件活都干完了,接下来要把它们合并回主分支。这里有个讲究:<b>最好逐个合并,别一股脑全推上去</b>。原因很实在——要是 A、B、C 一起合,万一合完出了 bug,我们很难判断到底是谁引入的。更稳的流程是这样一步一停:

```
先合并 A
   ↓
跑测试 / 看页面
   ↓
没问题再合并 B
   ↓
再跑测试 / 看页面
   ↓
最后合并 C
```

&emsp;&emsp;我们就按这个节奏一个一个来。先合并第一件「每个菜都从网上找到对应的图片,做成合适的大小插入菜单内」。每个会话的标题都是 Codex 根据当次会话主题自动生成的,名字不一定跟我们脑子里想的一字不差——找到那个跟图片相关的会话(这里叫「为菜单添加菜品图片」),输入"合并到主分支,然后启动服务"。

&emsp;&emsp;接着合并第二件「加一个会员充值的功能」。找到「新增会员充值功能」会话框,同样输入"合并到主分支,然后启动服务"。

### 9.5 两种并行模式回顾

&emsp;&emsp;走完这一遍实践,我们就能把 Codex 的多 agent 能力看清楚了——它其实分成两种模式。

&emsp;&emsp;<b>单会话并行</b>更偏向<b>思考并行</b>。supervisor 把任务拆给几个 SubAgent 同时去分析、调研、出方案,最后再由它一个人统一动手改、统一验证。这种方式适合需求偏复杂、要读大量代码、要先想清楚方案的场景。

&emsp;&emsp;<b>多会话加 worktree</b> 则进一步做到了<b>写入并行</b>。每个会话有自己独立的工作区和分支,可以同时开发不同功能,效率提升很明显。只不过 worktree 解决的是开发阶段的隔离问题,并不能完全避免最终合并时的代码冲突,所以实际用的时候,还是建议守住「一功能一 worktree」的原则,再用逐个合并、逐个验证的方式把功能集成回主分支,整个过程更像是在跑一条 workflow。

&emsp;&emsp;回头看这次点餐系统的案例,我们几乎没有手敲一行代码,全靠任务拆分、并行执行、逐步合并,在不长的时间里就完成了菜单优化、菜品详情、菜品图片、会员充值、订单退款这一大串功能。这也正是 Agent Coding 跟传统补全式工具最大的不同——agent 不再只是帮我们写两笔代码,而是能像一名开发那样,自己分析需求、动手改项目、验证结果、完成交付。

&emsp;&emsp;等项目规模再大些,多 agent 加 worktree 这套模式,会越来越像一支真正的软件团队:几个 agent 各管一块模块开发,而我们更多是站在产品经理和技术负责人的位置上,管需求拆解、任务分配、代码审查和最终拍板。随着模型和工具链一路往前走,这种协作方式,很可能会成为 AI 编程里一种重要的日常工作流。

### 10.2 添加 Computer Use 插件

&emsp;&emsp;我们要装的 Computer Use,可以理解成给 Codex「操作电脑界面」的能力。它不再只是读代码、跑命令,而是能<b>看到屏幕上显示的内容</b>,再通过点击、输入、翻菜单、切窗口、用剪贴板这些方式,去操作 macOS 或 Windows 上的应用。所以它特别适合那些没有标准接口、只能靠图形界面一步步点完的活,比如测试桌面软件、操作浏览器页面、检查模拟器、调一些本地软件里的设置。

> <font size=2>**「名词解释」<font color=red>Computer Use</font>(计算机使用)** — 让 Codex 像真人一样看屏幕、动鼠标、敲键盘来操作电脑软件的能力,专治那些没有代码接口、只能手点的图形界面任务。</font>

&emsp;&emsp;它既然能像真人一样看屏幕、点鼠标、敲键盘,理论上连游戏界面也能操作。不过它更擅长节奏慢、规则清楚的任务,像解谜、棋类、回合制这类还行;动作类、射击类那种要靠高速反应的,就不太现实了。还有一点得提醒:它是直接操作我们这台机器的真实界面,所以碰到登录、支付、删除、发消息这种高风险动作时,一定要格外当心。

&emsp;&emsp;下不过 Mac 人机,这很正常。Codex 是个通用大模型,本身就不是专门的棋类算法;再加上它这一局里既要认棋盘、又要算棋、还要操作鼠标,加上对话越拖越长、上下文越积越多,这些都会拖累它的判断。反观对面那个人机,虽说未必有多强,但它好歹是专门下棋的程序,公开资料里叫 Sjeng 引擎——它会把当前所有合法走法都列出来,往后推演若干步,假设双方都尽量走最优,再用子力、王的安全、兵型、中心控制、有没有威胁这些指标给每个局面打分,最后挑评分最高的那一步落子。专业归专业,通用归通用,差距就在这儿。

### 10.5 让它放一首歌

&emsp;&emsp;前面两个软件都是 Mac 自带的,我们再试个第三方软件,看 Codex 能不能照样操控。输入「打开网易云音乐软件,查找并播放《Rainy Blue(蓝色阴雨)》歌曲」。

&emsp;&emsp;它把软件打开了,而且准确地搜到了我们要的那首歌。

### 10.6 小节

&emsp;&emsp;从上面三个案例能看清 Computer Use 的核心价值:它的本事不在「写代码」,而在让 Codex 拿到了操作真实电脑界面的能力。它能像我们一样看屏幕、点按钮、输文字、切窗口,所以只要任务是摆在图形界面里的,它就有机会接过去办。

&emsp;&emsp;三个例子各说明了一个侧面。闹钟那一关告诉我们,它很适合步骤明确、目标清晰的软件操作,像打开应用、填内容、改设置、建提醒;国际象棋那一局告诉我们,它能操作游戏界面,可不等于自带专业游戏能力,碰到要长线推演、精确判断、高频决策的任务,通用大模型照样会出错;网易云那次又告诉我们,哪怕界面上没有明摆着的按钮,它也能靠画面和常识推断出下一步。

&emsp;&emsp;说到底,Computer Use 最适合的是这么一类活:没有现成接口、只能靠手点界面、流程相对明确、风险还可控——比如测试桌面软件、操作本地工具、检查网页或模拟器、处理一些重复的办公流程。反过来,高风险的操作、要极快反应或极高精度的任务,就不太适合交给它。

&emsp;&emsp;一句话收个尾:Computer Use 让 Codex 从一个会读代码、会跑命令的 Agent,进一步变成了能操作电脑界面的 Agent。它把 Codex 能干的活的边界又往外推了一截,但这不代表它能稳稳当当搞定所有桌面任务——真用起来,权限、安全和任务复杂度这几样,我们心里始终得有根弦。

## <center>第十一章 定时自动化与自建 Skill</center>

&emsp;&emsp;前面几章的任务都是我们手动发起的——想让 Codex 干点啥,就打开会话敲一句话。这一章我们换个方式:让它<b>定时自动干活</b>,每天到点自己去网上搜 AI 圈的最新消息、整理成一份美观的 HTML 日报、再自动弹出浏览器给我们看。光自动还不够,我们还会把一套自己喜欢的网页风格<b>沉淀成一个可复用的 Skill</b>,让它以后每次产出都按这套风格来。把这两件事串起来,Codex 就从"叫一次干一次"变成了"定时、稳定、按固定质量帮我们完成一类活"。

> <font size=2>**「名词解释」<font color=red>自动化(Automation)</font>** — Codex 里的定时任务功能,给它一段固定指令和一个时间表(比如每天九点),到点它就自动发起一个会话把活干完,不用我们守在旁边手动触发。</font>

### 11.1 起个项目放产出

&emsp;&emsp;自动化任务要往哪写文件、在哪个项目空间里跑,得先有个落脚的地方。我们先在本地建一个文件夹,专门放每天生成的日报。

&emsp;&emsp;然后回到 Codex,把这个文件夹选中、当成一个新项目打开。后面的自动化任务就绑在这个项目上跑,它写出来的文件也就乖乖落在这个目录里。

### 11.2 套用模板还是手动设置

&emsp;&emsp;项目有了,我们来建第一个自动化任务。点「自动化」→「查看模板」,这里官方备了不少现成的模板,像每日简报、定期巡检这类常见活计,挑一个改改就能直接用,适合还没想好怎么写指令的时候。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch11-%E8%87%AA%E5%8A%A8%E5%8C%96%E6%9F%A5%E7%9C%8B%E6%A8%A1%E6%9D%BF-fd103ee9.png" width="80%"></div>

&emsp;&emsp;不过我们想要的活比较具体,直接点「手动设置」自己写一份更顺手。这份指令我们写得很细——因为自动化是它自己定时跑的,我们不在旁边盯着,所以得把"写到哪、找不到目录怎么办、链接和图片有什么要求、页面长什么样"全都一次交代清楚,免得它到点了自己拿不准乱来:

```
请每天检索过去 24 小时内 AI 圈的重要新闻，并生成一份图文并茂的中文 HTML 日报。

任务开始前请先检查：
1. 当前环境是否可以写入 /Users/xiaoyang/每日AI资讯
2. 如果目录不存在，请创建该目录
3. 如果无法写入该目录，请先停止任务并告诉我需要授予本机文件访问权限，不要只返回文字版日报
4. 当前是否可以在本机打开浏览器
5. 如果无法自动打开浏览器，请在完成后明确告诉我 HTML 文件路径

检索策略：
- 优先使用 Codex 可用的联网搜索和网页浏览能力完成检索、来源确认、原文链接提取和图片链接提取
- 只有在普通搜索/浏览无法获取页面内容、图片链接、发布时间，或必须使用已登录浏览器状态时，才使用 Chrome 插件或界面操作能力
- 不要只根据搜索结果摘要写新闻，重要新闻必须进入原文页面确认

保存与展示要求：
1. 将 HTML 文件保存到：/Users/xiaoyang/每日AI资讯
2. 文件名格式：AI日报-YYYY-MM-DD.html
3. 生成完成后，用默认浏览器打开该 HTML 文件
4. 如果需要用 Chrome 打开本地 HTML，请使用 Chrome 插件；如果 Chrome 无法访问 file:// 本地文件，请提示我开启 Codex Chrome 扩展的 Allow access to file URLs 权限

链接要求：
- 每条新闻必须有真实原文链接
- 新闻卡片整体可以点击跳转
- 标题、图片、「查看原文」按钮都能跳转到原文
- 所有链接使用 target="_blank" 打开
- 不要使用搜索结果页链接
- 没有可靠原文链接的新闻不要收录

图片要求：
- 优先引用原文页面中的代表性图片
- 图片直接使用原文图片 URL，不下载到本地
- 图片点击后跳转到原文
- 如果没有可靠原文图片，就不要强行配图
- 不要使用无来源图片、无关配图、AI 生成配图

事实校验要求：
- 每条新闻必须确认发布时间、来源媒体/机构、原文 URL
- 涉及模型发布、价格、API 更新、公司公告时，优先使用官方博客、官方文档、GitHub Release、论文页面或公司公告
- 如果只有二手转载，请标注「二手来源」；不确定的信息不要写成确定事实
- “新闻事实”和“为什么重要”分开写，不要把主观判断伪装成事实
- 如果无法打开原文确认发布时间和来源，则不要收录该新闻

内容要求：
- 覆盖模型动态、产品工具、Agent 与编程工具、开源项目、研究论文、行业商业
- 每条新闻包含：标题、摘要、为什么重要、发布时间、来源、原文链接、图片来源
- 不要凭记忆写，必须基于当天搜索结果
- 不要堆砌小新闻，优先筛选真正重要的消息

页面结构：
1. 每日 AI 资讯
2. 日期与更新时间
3. 今日最重要的 5 条
4. 模型动态
5. 产品工具
6. Agent 与编程工具
7. 开源项目
8. 研究论文
9. 行业商业
10. 今日最值得关注的一件事

HTML 设计要求：
- 单文件 HTML，CSS 内嵌
- 白底、蓝色主色调，整体像专业科技资讯日报
- 卡片式布局，桌面端多列，移动端单列
- 图片比例统一，使用 object-fit: cover，不能拉伸变形
- 标题、摘要、来源、时间层级清晰
- 页面适合直接阅读、截图和分享
- 不要使用花哨渐变、夸张阴影、紫色 AI 风格
```

&emsp;&emsp;指令里我们特意留了一手 Chrome 插件。有些新闻页面要登录才能看全,而我们本机的 Chrome 本来就是登录状态,挂上插件后,Codex 就能借着这个登录身份在 Chrome 里读网页、点按钮、切页面,不会被登录墙卡住。配置很简单,点「插件」→「chrome」加上就行。

&emsp;&emsp;指令写好、插件挂上,最后配运行参数:运行环境选「本地」(这样它才能往我们本机目录写文件),运行项目选刚建的「每日AI资讯」,时间定在每天九点。

### 11.4 把喜欢的风格做成 Skill

&emsp;&emsp;日报内容是齐了,可这个 HTML 的样子我们还是觉得太朴素了点。要是每天都重新在指令里描述一遍想要什么风格,太啰嗦了。这时候就轮到 Skill 上场了。

> <font size=2>**「名词解释」<font color=red>Skill(技能)</font>** — 把一套规则或风格写进一份 SKILL.md 文件让 Codex 读取复用。它就像提前写好的一份说明,以后任务一对上号,Codex 就自动加载这份文件、照里面的规矩办,我们不用每次重复交代。</font>

&emsp;&emsp;说白了,普通提示词是临时的一句话,比如:

```
帮我生成一个好看的每日 AI 日报 HTML
```

&emsp;&emsp;而 Skill 是提前写好的一份规则,把"好看"具体成一条条要求:

```
当用户要生成日报 HTML 时：
- 使用 Scatterbrain 风格
- 背景像软木板或纸张
- 新闻卡片像便利贴
- 每张卡片要有轻微旋转、阴影、图钉、胶带
- 标题、图片、查看原文按钮都要能点击
- 输出单个 HTML 文件
```

&emsp;&emsp;有了这份规则,以后我们不用每次都把一大段风格描述重写一遍。Codex 一看任务跟这个 Skill 的描述对得上,就会自动加载这份 SKILL.md,照里面的规矩产出。

&emsp;&emsp;那风格从哪来?我们去 https://github.com/zarazhangrui/frontend-slides 这个开源项目上淘一淘。它本来是做网页幻灯片的,但里头的视觉风格挺好看,我们想把它整理一下,变成日报 HTML 的风格。

&emsp;&emsp;然后把要求输进去。注意我们没让它直接复制一份 HTML 了事,而是明确要它读模板目录下的设计文件、总结出核心视觉规范、做成一个可复用的 Skill,连 SKILL.md 里该写哪些、参考资料和模板骨架放哪个目录都交代清楚了:

```
我喜欢这个 HTML 模板的视觉风格：
(https://github.com/zarazhangrui/beautiful-html-templates/tree/main/templates/scatterbrain)

请帮我把它创建成一个 Codex Skill。

Skill 名称：scatterbrain-html-daily

用途：
当我要求生成每日资讯、AI 日报、新闻简报、图文 HTML 日报、本地 HTML 报告时，使用这个 skill 的视觉风格。

要求：
1. 先读取该模板目录下的 design.md、template.html、template.json。
2. 总结 Scatterbrain 的核心视觉规范：便利贴、软木板、纸张纹理、图钉、胶带、手写感字体、轻微旋转、柔和阴影、彩色卡片。
3. 创建一个可复用的 skill，而不是只复制一次 HTML。
4. SKILL.md 要说明触发场景、页面结构、颜色、字体、新闻卡片布局、响应式规则、链接规则。
5. references/ 中保存该风格的设计规范摘要。
6. templates/ 或 assets/ 中保存可复用的 HTML 模板骨架。
7. 这个 skill 只负责 HTML 视觉呈现，不负责联网搜索新闻。
8. 以后我的每日定时任务生成 HTML 日报时，可以指定使用这个 skill。
```

### 11.5 让自动化用上这个 Skill

&emsp;&emsp;Skill 有了,接下来把它接进原来的自动化任务。点「自动化」→「编辑自动化」,打开之前那份指令。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch11-%E7%BC%96%E8%BE%91%E8%87%AA%E5%8A%A8%E5%8C%96-d1e8c775.png" width="80%"></div>

&emsp;&emsp;我们只动一处——把原来那段「HTML 设计要求」整段换掉,明确要它生成时套用 scatterbrain-html-daily 这个 Skill 的视觉规范,改完点右上角「立即运行」:

```
HTML 设计要求：
- 生成 HTML 时必须使用 scatterbrain-html-daily Skill 的视觉规范
- 单文件 HTML，CSS 内嵌
- 整体采用 Scatterbrain 风格：便利贴、软木板/纸张纹理、图钉、胶带、轻微旋转、柔和阴影、彩色新闻卡片
- 页面要适合直接阅读、截图和分享
- 新闻卡片可以有轻微错落感，但不能影响可读性
- 桌面端多列布局，移动端单列布局
- 图片比例统一，使用 object-fit: cover，不能拉伸变形
- 标题、摘要、来源、时间层级清晰
- 不要使用无关装饰影响新闻阅读
```

&emsp;&emsp;最终效果还不错,Codex 完美用上了我们给它的 Skill,生成了便利贴风格的 HTML——软木板背景、彩色卡片、轻微旋转,跟我们挑的那套开源风格对上了。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch11-skill%E6%95%88%E6%9E%9C%E4%BE%BF%E5%88%A9%E8%B4%B4-bb199bed.png" width="80%"></div>

### 11.6 自动化与 Skill 的分工

&emsp;&emsp;走完这个例子,我们其实把一件原本每天要手动做的活,拆成了两块交给 Codex。

&emsp;&emsp;一块是<b>自动化任务</b>:让 Codex 每天定时去搜过去 24 小时 AI 圈的重要新闻,逐条确认原文链接、发布时间、来源和图片,再生成一份本地 HTML 日报、自动弹出浏览器展示。这样我们就不用天天手动搜索、筛选、复制链接、整理排版了。

&emsp;&emsp;另一块是<b>自建 Skill</b>:当我们对默认产出的样子不满意,就把自己喜欢的网页风格整理成一个可复用的 Skill;之后只要在自动化任务里点名用它,Codex 就照着这套视觉规范来出页面。

&emsp;&emsp;说到底,自动化解决的是"重复执行"的问题——同一件活到点自己跑;Skill 解决的是"稳定复用一套方法或风格"的问题——每次产出都是同一个质量标准。两者一结合,Codex 就不只是帮我们干一次活了,而是能持续、定时、按固定标准帮我们完成一整类活。

## <center>第十二章 用手机远程指挥电脑端 Codex</center>

&emsp;&emsp;走到这里,Codex 的主要能力我们差不多都体验过一遍了——派活、计划、审批、引导、用 git 存档回退、生图、立规矩。这最后一章很短,我们看一个挺有意思的用法:人不在电脑前,也能用手机远程指挥电脑上的 Codex 干活。当前版本的 Codex 已经支持这么用了,设置起来也就几步。

### 12.1 连上手机和电脑

&emsp;&emsp;先说清楚一件事:手机端要用的不是另装一个 Codex App,而是我们平时就在用的 <b>ChatGPT</b>手机应用。所以动手前,先在手机上装好 ChatGPT、用我们这门课一直在用的那个账号登录——记得是同一个账号,后面电脑和手机能对上,靠的就是这个账号打通。

&emsp;&emsp;账号备齐了,我们回到电脑端 Codex,点左侧导航栏的「codex 移动版」→「管理连接」。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch12-%E7%A7%BB%E5%8A%A8%E7%89%88%E7%AE%A1%E7%90%86%E8%BF%9E%E6%8E%A5-58393b09.png" width="80%"></div>

&emsp;&emsp;点开之后会弹出一个二维码。这个二维码我们暂时不用管它,直接点「添加」就行;要留意的是下面那两个功能开关,记得都打开——它们决定了手机端能不能看到电脑上的项目、能不能远程发任务,关着的话后面会用不了。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch12-%E8%BF%9E%E6%8E%A5%E4%BA%8C%E7%BB%B4%E7%A0%81-220d168d.png" width="80%"></div>

&emsp;&emsp;电脑这边弄好,就轮到手机了。这时手机上的 ChatGPT 会收到一条授权请求,我们点「为此手机授权」就完成了配对。说白了,这一步就是电脑在确认:这台手机是我本人的,允许它来遥控我。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch12-%E6%89%8B%E6%9C%BA%E6%8E%88%E6%9D%83-b8ab29e9.png" width="80%"></div>

&emsp;&emsp;授权一过,手机上立刻就能看到电脑里的所有项目了——待办清单、点餐系统,我们前面建的这些都在。到这儿手机和电脑就算连通了,接下来我们就能拿手机来遥控电脑上的 Codex 写代码。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch12-%E6%89%8B%E6%9C%BA%E7%9C%8B%E9%A1%B9%E7%9B%AE-4e0d7786.png" width="80%"></div>

### 12.2 手机发任务,电脑端干活

&emsp;&emsp;连通之后用起来跟电脑端没太大差别。我们在手机上挑出之前那个 `Ordering_System` 项目,输入一句任务:「店名由『好味来』换成『一品食堂』,然后启动服务」。发出去之后有意思的地方来了——电脑端的 Mac 上同时冒出了这个会话,手机点一句、电脑跟一句,两边的输出是实时同步的。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch12-%E6%89%8B%E6%9C%BA%E5%8F%91%E4%BB%BB%E5%8A%A1%E5%90%8C%E6%AD%A5-785b5adf.png" width="80%"></div>

&emsp;&emsp;之所以是这个样子,是因为手机其实没在干活,真正干活的还是电脑端的 Codex。手机只是一个<b>远程的入口</b>:我们在手机上挑项目、写任务、看它跑到哪一步,而实际的读代码、改代码、跑命令、起服务,全都发生在电脑那一头。手机替我们按了一下按钮,活儿是电脑在出。

&emsp;&emsp;任务跑完,服务确实在 Mac 上起来了,地址是 `http://127.0.0.1:8000/`。不过这里有个小坑得提一句:这个地址手机是<b>打不开</b>的。原因不复杂——`127.0.0.1` 这个地址,每台设备指的都是「我自己这台机器」,它从来不会跨设备。电脑上跑起来的服务,这个地址只在电脑本机有效;手机去访问 `127.0.0.1`,问的是手机自己,而手机上根本没起这个服务,自然就打不开。所以手机能远程指挥电脑、能看 Codex 干活的全过程,但要真打开这个本地网页看效果,还得回到电脑跟前。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/codex-app/2026-06-16/ch12-%E6%9C%8D%E5%8A%A1%E5%9C%A8mac%E5%90%AF%E5%8A%A8-406e685e.png" width="80%"></div>

> <font size=2>**「名词解释」<font color=red>127.0.0.1</font>(本地回环地址,也叫 localhost)** — 一个特殊的网络地址,在任何一台设备上,它都只指「当前这台设备自己」,永远不会跨到别的机器。所以电脑上跑的本地服务,只有电脑本机能用这个地址访问,手机、别人的电脑都访问不到。</font>

### 12.3 本章小结

&emsp;&emsp;手机远控这件事,串起来就一句话:手机是入口,电脑是干活的地方。我们在手机上挑项目、派任务、盯进度、做确认,而真正的改代码、跑命令、起服务,始终落在电脑端的 Codex 上——手机不会把代码搬到自己身上跑,它只是替我们远程按下了那个「开始」。

&emsp;&emsp;这套用法最顺手的场景,是处理那些临时冒出来的小活。人在外面,突然想起来要改个文案、修个小毛病、把某个项目的服务启起来,或者就是想看一眼之前那个任务跑得怎么样了,这些都不必非得赶回电脑前。只要电脑端的 Codex 一直开着、保持能连上,我们在手机上就能随时发起任务、随时跟进。它远没到「在手机上完整开发」的地步,但作为一个随身的遥控器,应付临时需求已经相当够用了。

### 13.2 最后真正要记住的几件事

&emsp;&emsp;学完这些功能之后,不需要把每个按钮的位置都死记硬背。界面以后可能会变,功能名字也可能会调整,但使用 Codex 的基本思路是稳定的:先把需求说清楚,再让它去执行,最后认真检查它交回来的结果。

&emsp;&emsp;实际用的时候,可以记住几个简单习惯。第一,任务不要一次塞得太大,能拆就拆;如果事情比较复杂,先让它列计划。第二,不要只看它最后说“完成了”,还要看它改了哪些文件、有没有报错、页面或接口能不能正常运行。第三,重要节点及时保存版本,这样就算改坏了也能退回来。第四,权限不要一开始就放得太开,先看清它要做什么,熟悉之后再根据任务需要放宽。

&emsp;&emsp;把这几件事做好,Codex 就会从一个“看起来很厉害的新工具”,慢慢变成你日常做项目时真正用得上的助手。

### 13.3 接下来交给自己的项目

&emsp;&emsp;真正熟练使用 Codex,不是靠看完一门课就结束了,而是要拿自己的真实需求继续练。可以从很小的事情开始:改一个页面、整理一份表格、做一个本地小工具、给已有项目补一个功能。每次都按同样的流程走:先说清楚目标,让它动手做,再自己检查结果。

&emsp;&emsp;这门课到这里就收尾了。后面再遇到项目需求时,大家就可以试着把一部分工作交给 Codex,但同时保留自己的判断:需求是不是讲清楚了,改动是不是合理,结果是不是能用。能做到这一点,才算真正把 Codex 用进了自己的工作流里。
