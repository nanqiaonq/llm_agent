## <center>第一章 简单认识 Claude Code</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch01-01-chapter-overview-da43f4e2.png" width="80%" alt="Claude Code 第一章安装、启动、文件授权与待办软件实作总览"></div>

&emsp;&emsp;Claude Code 会直接读取和修改项目文件，也会在获得授权后执行终端命令。初次使用时，安装状态、启动目录和权限边界只要有一项没有确认清楚，后续任务就可能落到错误项目，或者在授权环节停住。第一章先建立可靠的基础流程，让我们从第一次启动开始就能判断 Claude Code 正在处理哪个项目、准备执行什么操作，以及结果应当到哪里检查。

&emsp;&emsp;这一章会完成官方安装与 `claude --version` 版本确认，使用 `cd` 进入项目目录，再通过 `claude` 启动交互会话；随后我们用“小故事”任务认识 Agent 与对话型大模型的差别和文件编辑授权，再把“待办软件”从需求选择完整做到浏览器验收。走完这条路径，我们会得到一个可继续修改的项目结果，也会建立“确认环境、提交任务、处理授权、检查成品”的基本操作顺序。

### 1.1 Claude Code 快速安装

#### 1.1.1 安装前提

&emsp;&emsp;安装 Claude Code 前，需要确保电脑已连接到能够正常访问 Claude 官方服务的网络环境，也就是俗称的“魔法网络”环境。

&emsp;&emsp;Claude Code 有两种快速安装方式：一种是在终端中直接执行官方安装命令，另一种是让编程 Agent 帮助完成安装和验证。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch01-shot-01-installation-options-1652ede2.png" width="80%" alt="在终端中直接安装或让 Agent 帮助安装 Claude Code"></div>

&emsp;&emsp;图中左侧展示在终端中直接输入安装命令，右侧展示向 Agent 说明安装要求，让 Agent 自动执行安装并确认结果。

#### 1.1.2 在终端中直接安装

&emsp;&emsp;macOS、Linux 和 WSL 使用相同的安装命令：

```bash
curl -fsSL https://claude.ai/install.sh | bash
```

&emsp;&emsp;Windows 用户打开 PowerShell，运行：

```powershell
irm https://claude.ai/install.ps1 | iex
```

&emsp;&emsp;这些命令会下载并执行 Claude Code 官方安装脚本。

#### 1.1.3 让编程 Agent 帮助安装

&emsp;&emsp;这里的编程 Agent，指可以理解任务、调用终端等工具并在授权范围内完成操作的编程助手，比如cursor、codex等；如果没有这类工具，直接使用 1.1.2 的终端安装方式即可。

&emsp;&emsp;如果正在使用支持 Agent 模式的编程工具，也可以直接向 Agent 提出安装要求。以 Mac 为例，可以输入：

> <b>输入给编程 Agent 的提示词</b>

```text
请在我的 Mac 上安装 Claude Code。

使用官方推荐的安装方式：
curl -fsSL https://claude.ai/install.sh | bash

安装完成后运行：
claude --version

确认安装是否成功，不要修改当前项目文件。
```

&emsp;&emsp;Agent 会按照要求执行官方安装命令，并在安装后检查 Claude Code 的版本。

#### 1.1.4 用 claude --version 确认版本

&emsp;&emsp;安装完成后，macOS、Linux 和 Windows 都可以运行下面的命令查看版本：

```bash
claude --version
```

&emsp;&emsp;图中终端返回的版本为 `2.1.210 (Claude Code)`，说明 Claude Code 已经成功安装。

### 1.2 用 cd 进入项目目录

&emsp;&emsp;运行 Claude Code 前，建议先在终端中进入需要处理的项目目录。Claude Code 会把当前目录作为项目上下文，因此它后续读取、修改或创建的任务文件，都与启动时所在的项目目录密切相关。

&emsp;&emsp;macOS 和 Linux 可以运行：

```bash
cd ~/code/claude_code_demo
```

&emsp;&emsp;Windows PowerShell 以项目位于 D 盘根目录为例，可以运行：

```powershell
cd D:\claude_code_demo
```

&emsp;&emsp;请根据自己电脑中项目的实际位置修改目录路径。

### 1.4 Agent 与对话型大模型的区别

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch01-02-agent-vs-chat-bc4c9407.png" width="80%" alt="Claude Code Agent 与对话型大模型的工作方式对比"></div>

&emsp;&emsp;普通的对话型大模型主要通过文字回答问题，而 Agent 不仅能够生成内容，还可以调用工具，在权限允许的范围内操作电脑中的项目文件、执行终端命令并完成实际任务。

&emsp;&emsp;例如，在 Claude Code 的输入区域中输入：

> 💬 输入给 Claude Code 的提示词

```text
写一个200字的小故事，存到当前文件夹下，用中文
```

&emsp;&emsp;输入的任务指令前会显示 `>` 提示符。Claude Code 理解任务后，会准备创建 `小故事.txt`，并预览即将写入文件的内容。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch01-shot-06-file-creation-approval-fc5b1b99.png" width="80%" alt="Claude Code Agent 创建文件前请求用户授权"></div>

&emsp;&emsp;由于当前界面左下角显示 `manual mode on`，Claude Code 不会直接创建文件，而是先询问是否允许创建 `小故事.txt`。界面提供三个选项：

1. `Yes`：只允许本次创建文件。
2. `Yes, allow all edits during this session`：允许 Claude Code 在当前会话中继续编辑文件，不再逐次确认。
3. `No`：拒绝本次文件创建操作。

&emsp;&emsp;本次选择第 2 项并确认后，Claude Code 会将故事写入当前项目目录。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch01-shot-07-file-created-edit-permission-68f32ed3.png" width="80%" alt="Claude Code 成功创建小故事文件并切换为 accept edits on"></div>

&emsp;&emsp;图中左侧的项目文件列表中已经出现 `小故事.txt`，交互区域也显示 `Wrote 11 lines to 小故事.txt`，说明文件已经成功创建。

&emsp;&emsp;同时，界面左下角的权限状态已从 `manual mode on` 变为 `accept edits on`。这表示在当前会话中，Claude Code 可以继续创建或编辑文件，无需每次都重新请求确认。

### 1.5 让 Claude Code 编写待办软件

&emsp;&emsp;除了前面章节演示的简单内容生成任务，Claude Code 还可以帮助我们编写代码、创建应用程序，这也是它最擅长的能力之一。

&emsp;&emsp;在输入区域中提出需求：

> 💬 输入给 Claude Code 的提示词

```text
帮我做一个待办软件
```

&emsp;&emsp;由于这个需求还比较宽泛，在本次演示中，Claude Code 没有立即开始写代码，而是主动询问软件形态和功能范围，帮助我们补充具体需求。需要注意的是，Claude Code 每次追问的问题、顺序和选项不一定完全相同，它会根据当前项目的上下文和用户提出的需求进行调整。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch01-shot-08-todo-app-type-0d8790e1.png" width="80%" alt="Claude Code 询问待办软件的实现形态"></div>

&emsp;&emsp;首先需要选择待办软件的形态。界面提供了网页版、命令行版和桌面应用等选项，也可以输入自定义方案或继续与 Claude Code 讨论。

&emsp;&emsp;不同选择会影响软件的技术实现和最终使用方式。本次选择的是“网页版（浏览器打开）”。

&emsp;&emsp;完成软件形态选择后，Claude Code 会继续询问需要包含哪些功能。

&emsp;&emsp;功能范围支持多项选择，界面提供的选项包括：

- 基础增删改查
- 优先级与分类
- 截止日期与提醒
- 数据持久化保存

&emsp;&emsp;也可以选择 `Type something` 补充自定义功能。选择完成后，通过 `Submit` 进入答案确认界面。

&emsp;&emsp;提交前，Claude Code 会汇总已经选择的软件形态和功能。本次确认的需求为：

- 软件形态：网页版（浏览器打开）
- 功能范围：基础增删改查、优先级与分类、截止日期与提醒、数据持久化保存

&emsp;&emsp;确认内容没有问题后，选择 `Submit answers` 提交答案。Claude Code 会根据这些具体要求设计并编写待办软件。

&emsp;&emsp;本次 Claude Code 选择创建一个单文件网页应用，将 HTML、CSS 和 JavaScript 写入 `todo.html`，并使用浏览器的 `localStorage` 持久化保存数据。由于当前处于 `accept edits on` 模式，Claude Code 写入 `todo.html` 时无需再次请求确认。

&emsp;&emsp;文件创建完成后，Claude Code 尝试执行终端命令，在默认浏览器中打开待办应用。不同系统使用的命令不同。

&emsp;&emsp;macOS：

```bash
open "$HOME/code/claude_code_demo/todo.html"
```

&emsp;&emsp;Linux：

```bash
xdg-open "$HOME/code/claude_code_demo/todo.html"
```

&emsp;&emsp;Windows PowerShell：

```powershell
Start-Process D:\claude_code_demo\todo.html
```

#### 1.5.1 使用 Esc 或 Ctrl+C 中途补充说明

&emsp;&emsp;Claude Code 正在执行任务时，如果发现当前方向需要调整，或者还有条件需要补充，可以按 `Esc` 或 `Ctrl+C` 中断本轮执行，不必等待它把当前任务全部完成。这里讨论的是任务执行过程中的中断，不是没有任务运行时连续按两次 `Ctrl+C` 退出会话。

&emsp;&emsp;中断后仍会留在当前会话中，已经保留的对话上下文不会因此消失。此时可以直接输入补充要求、修改目标或纠正方向，再让 Claude Code 按新的说明继续执行。

### 1.6 小结

&emsp;&emsp;本章先完成 Claude Code 的安装与启动准备。可以在终端执行官方安装命令，或把安装和验证要求交给支持 Agent 模式的编程工具协助处理；无论采用哪种方式，结束后都要运行 `claude --version`，用返回的版本信息确认工具可用。本次手动结果为 `2.1.210 (Claude Code)`，后续界面以此为基准。确认版本后，先用 `cd` 进入 `claude_code_demo`，再运行 `claude`。启动目录就是会话的项目上下文，因此还要在欢迎界面复核项目目录，避免后续读取、创建或修改文件时落到错误位置，也为接下来的任务建立明确范围。

&emsp;&emsp;基础操作要按“确认版本、进入目录、启动会话”的顺序执行：`claude --version` 回答是否安装成功，`cd` 决定当前处理哪个项目，`claude` 则从该目录进入交互会话。任一步遗漏，都可能让后续操作处于错误环境。程序打开后也不能立即提交任务，应先检查欢迎界面中的版本、模型、账户信息、项目目录、输入区域和运行模式。项目目录限定会话关联的文件范围，输入区域承接自然语言任务，运行模式提示当前授权方式；把这些信息逐项对照无误后，再开始输入需求，能减少版本不符、目录选错或权限理解错误带来的返工。

&emsp;&emsp;对话型大模型主要生成文字，Agent 能调用工具，把回答落实为文件或命令。“小故事”任务中，Claude Code 会创建并预览 `小故事.txt`，获得允许后才写入项目。权限状态决定工具边界：`manual mode on` 下编辑前需确认；切换到 `accept edits on` 后，后续文件编辑不再逐次询问，但不代表其他电脑操作也已放行。运行终端命令打开 `todo.html` 时，仍会出现 `This command requires approval`。因此要区分文件编辑与终端命令，查看界面列出的具体命令和授权选项，只批准当前任务确实需要且影响范围可判断的操作。

&emsp;&emsp;待办软件展示了从想法到可检查结果的完整实作链路。“帮我做一个待办软件”只是起点，Claude Code 会先询问软件形态和功能范围；本次选择网页版，以及增删改查、优先级与分类、截止日期与提醒、数据持久化等能力，并在答案确认界面提交。需求明确后，它把 HTML、CSS 和 JavaScript 写入单文件 `todo.html`，用 `localStorage` 保存数据。文件写入沿用已允许的编辑权限，调用系统命令打开浏览器则另行授权。页面打开后，还要创建“周会”任务并标为完成，再检查分类、截止日期、筛选、删除、清除已完成、自动保存和双击编辑，确认生成、授权与验收真正闭环。

&emsp;&emsp;完成本章后，可用五项能力检查结果：能说明两种安装方式并用 `claude --version` 验证版本；能进入目录后用 `claude` 启动会话，并核对欢迎界面的模型、账户、目录、输入区和运行模式；能解释对话型大模型与 Agent 在工具调用上的差别；能区分 `manual mode on`、`accept edits on` 与终端命令单独授权的边界；能复述待办软件从需求澄清、生成 `todo.html`、授权打开到浏览器验收的过程。基础流程可归纳为：确认工具与项目，启动会话，提交清楚任务，逐项处理授权，最后回到实际文件和页面检查结果；说不清的环节应回到对应步骤复核。

## <center>第二章 Claude Code 权限配置</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-01-permission-flow-d9bef09f.png" width="80%" alt="Claude Code 五种权限模式与执行风险关系"></div>

&emsp;&emsp;第一章的待办软件任务已经表明，Claude Code 对文件编辑和终端命令采用不同的授权边界。权限设置过于保守，会让连续修改频繁停下来等待确认；权限范围过宽，又可能让删除文件等高风险命令在没有二次确认的情况下执行。权限管理决定了任务可以自动推进到什么程度，也是使用 Claude Code 时控制效率与风险的关键。

&emsp;&emsp;这一章会以 Claude Code 2.1.210 的实际界面为基准，认识 `manual mode on`、`accept edits on`、`plan mode on`、`auto mode on` 和 `bypass permissions on` 五种模式，使用 `Shift+Tab` 切换前四种模式，再通过 `/exit` 和 `claude --dangerously-skip-permissions` 进入 Bypass Permissions 模式；随后我们会在 Plan 模式中查看方案、选择确认方式并执行任务，同时明确 Bypass Permissions 自动修改文件和执行终端命令时的风险边界。

### 2.1 五种权限模式与 Shift+Tab

&emsp;&emsp;在本次演示使用的 Claude Code 2.1.210 环境中，本章介绍以下五种权限模式：

- `manual mode on`
- `accept edits on`
- `plan mode on`
- `auto mode on`
- `bypass permissions on`

&emsp;&emsp;前四种模式可以在本次环境中通过连续按 `Shift+Tab` 切换；`bypassPermissions` 是需要使用专门参数启用的高风险模式，会在后文单独说明。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-01-permission-modes-e6a2e857.png" width="80%" alt="Claude Code 前四种权限模式的状态栏对照"></div>

&emsp;&emsp;图中按照从上到下的顺序，展示了前四种权限模式。不同模式决定 Claude Code 可以自动完成哪些操作，以及哪些操作需要先获得用户确认。第五种 `bypassPermissions` 模式需要退出当前交互界面并使用参数重新启动，因此使用后文的独立画面展示。

#### 2.1.1 Manual 模式

&emsp;&emsp;状态栏显示：

```text
manual mode on
```

&emsp;&emsp;这是人工确认较多的模式。Claude Code 可以读取项目内容，但在创建或修改文件、执行需要授权的终端命令前，通常会先请求用户确认。

&emsp;&emsp;这种模式的可控性较高，适合刚开始使用 Claude Code，或者正在处理重要、敏感项目时使用。

#### 2.1.2 Accept Edits 模式

&emsp;&emsp;状态栏显示：

```text
accept edits on
```

&emsp;&emsp;在这个模式下，Claude Code 可以自动创建和编辑当前工作范围内的文件，并自动执行部分常见的文件操作。

&emsp;&emsp;其他需要授权的终端命令仍可能要求用户确认。例如前面章节中用于打开网页的 `open` 命令，就会单独显示授权提示。

&emsp;&emsp;这种模式适合已经确认开发方向，希望 Claude Code 连续完成多次代码修改的场景。

#### 2.1.3 Plan 模式

&emsp;&emsp;状态栏显示：

```text
plan mode on
```

&emsp;&emsp;Plan 模式主要用于分析和规划。Claude Code 可以读取项目文件、了解代码结构并制定实施方案，但不会直接修改项目源文件。

&emsp;&emsp;这种模式适合复杂任务。可以先让 Claude Code 分析问题并给出方案，确认思路正确后，再切换到允许编辑的模式开始实施。

#### 2.1.4 Auto 模式

&emsp;&emsp;状态栏显示：

```text
auto mode on
```

&emsp;&emsp;Auto 模式具有更高的自动化程度。Claude Code 会尽量减少权限确认，并通过后台安全检查判断操作是否符合用户提出的任务。

&emsp;&emsp;Auto 模式目前属于研究预览功能，并不是所有账号、模型和运行环境都能使用。即使可以启用，也不代表所有操作都绝对安全，处理重要项目时仍然需要检查最终结果。

> 不同用户能够看到的模式可能不同，尤其是 Auto 模式会受到账号套餐、模型和相关配置的影响。本节以图中实际使用的环境为准。

#### 2.1.5 Bypass Permissions 全权限模式

&emsp;&emsp;除了前面介绍的权限模式，Claude Code 还提供了一个权限几乎完全开放的 `bypassPermissions` 模式。

&emsp;&emsp;该模式不会出现在默认的权限切换流程中。需要先离开当前交互界面，再使用专门的启动参数重新运行 Claude Code。

##### 2.1.5.1 使用 /exit 退出交互界面

&emsp;&emsp;在 Claude Code 中输入：

```text
/exit
```

&emsp;&emsp;也可以在没有任务运行时连续按两次 `Ctrl+C`：第一次清空当前输入，第二次退出并返回终端。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-02-exit-interactive-session-8a2dcd15.png" width="80%" alt="使用 exit 命令离开 Claude Code 交互界面"></div>

&emsp;&emsp;图中输入 `/exit` 后，会离开当前 Claude Code 交互界面并返回终端。

&emsp;&emsp;需要注意的是，图中的命令说明显示：

```text
Detach from this background session (it keeps running)
```

&emsp;&emsp;这表示当前连接的是后台会话时，`/exit` 只会与该会话分离，后台会话仍会继续运行。

##### 2.1.5.2 使用跳过权限参数重新启动

&emsp;&emsp;返回项目目录的终端后，运行：

```bash
claude --dangerously-skip-permissions
```

&emsp;&emsp;macOS、Linux 和 Windows PowerShell 使用的启动命令相同，因此只需写一次。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-03-skip-permissions-launch-1487f8b2.png" width="80%" alt="使用 dangerously-skip-permissions 参数启动 Claude Code"></div>

&emsp;&emsp;`--dangerously-skip-permissions` 会让 Claude Code 以 `bypassPermissions` 模式启动。进入会话后，底部状态栏会显示：

```text
bypass permissions on
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-04-bypass-permissions-mode-fc10b2bf.png" width="80%" alt="Claude Code 进入 bypass permissions on 模式"></div>

&emsp;&emsp;图中底部状态栏已经显示 `bypass permissions on`，说明该模式已经启用。在这个模式下，Claude Code 执行工具调用时通常不会再显示权限确认，可以直接创建或修改文件、执行终端命令以及访问受保护的配置路径。

##### 2.1.5.3 使用时的安全风险

&emsp;&emsp;`bypassPermissions` 会跳过权限提示和常规安全检查，Claude Code 可能在没有确认的情况下执行高风险或不可逆操作。

&emsp;&emsp;出于安全考虑，只建议在隔离的容器、虚拟机或开发容器中使用，不建议直接在存有重要文件的日常电脑环境中开启。该模式也不能防止提示词注入或模型误操作。

&emsp;&emsp;即使在此模式下，针对文件系统根目录或用户主目录的极端删除操作，Claude Code 仍可能触发最后一道保护确认。

### 2.2 在 Plan 模式中规划并执行任务

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-02-plan-to-execution-0db15bf5.png" width="80%" alt="Claude Code 从 Plan 模式规划到批准执行的流程"></div>

&emsp;&emsp;Plan 模式适合在执行复杂任务前，让 Claude Code 先读取项目、分析需求并制定实施方案。

&emsp;&emsp;Plan 模式只有只读权限。Claude Code 可以读取文件、检索项目和执行只读操作，但无法创建、编辑或删除项目内容。只有用户确认方案，并让 Claude Code 退出 Plan 模式后，它才能切换到对应的权限模式开始修改。

&emsp;&emsp;发出任务前，首先查看界面左下角的权限状态，确认当前显示：

```text
plan mode on
```

&emsp;&emsp;如果当前不是 Plan 模式，可以按 `Shift+Tab` 切换，直到左下角出现 `plan mode on`。确认完成后，再输入本次任务：

> 💬 输入给 Claude Code 的提示词

```text
请你扩写小故事.txt到四百字，自行决定故事内容
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-05-plan-task-analysis-59ff44aa.png" width="80%" alt="Claude Code 在 Plan 模式中读取文件并生成扩写方案"></div>

&emsp;&emsp;图中左下角显示 `plan mode on`，说明当前仍处于 Plan 模式。Claude Code 首先读取 `小故事.txt`，确认原文内容和字数，然后分析扩写目标。

&emsp;&emsp;此时界面显示 `Updated plan`，表示实施方案已经生成，可以使用 `/plan` 查看方案内容。Claude Code 在规划阶段不会立即修改故事文件。

&emsp;&emsp;右下角还会出现根据当前任务生成的任务标签，例如 `expand-short-story`。标签内容由 Claude Code 根据任务自动生成，每次不一定完全相同。

#### 2.2.1 查看并确认实施方案

&emsp;&emsp;计划生成后，Claude Code 会展示准备执行的具体方案，包括任务背景、目标、扩写思路以及需要修改的文件。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-06-review-plan-dd30b805.png" width="80%" alt="查看 Claude Code 生成的扩写方案并选择执行方式"></div>

&emsp;&emsp;图中方案要求在保留原有结构、人物和主题的基础上，将故事扩写到约四百字。方案底部提供了多种处理方式：

1. `Yes, and bypass permissions`：批准方案，并切换到 Bypass Permissions 模式执行。
2. `Yes, manually approve edits`：批准方案，但文件修改仍需要人工确认。
3. `No, refine with Ultraplan on Claude Code on the web`：暂不执行，使用网页端 Ultraplan 继续完善方案。
4. `Tell Claude what to change`：不接受当前方案，告诉 Claude Code 需要修改哪些内容。

&emsp;&emsp;本次选择第 1 项：

```text
Yes, and bypass permissions
```

&emsp;&emsp;由于当前会话已经启用 Bypass Permissions 能力，所以可以直接使用该选项。

#### 2.2.2 批准方案并开始执行

&emsp;&emsp;选择第 1 项后，Claude Code 会退出 Plan 模式，并自动切换到该选项对应的权限模式。

&emsp;&emsp;本次状态栏由：

```text
plan mode on
```

&emsp;&emsp;自动变为：

```text
bypass permissions on
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-07-approve-plan-execute-cf578232.png" width="80%" alt="批准 Plan 方案后自动切换为 Bypass Permissions 并执行任务"></div>

&emsp;&emsp;图中红色内容表示原文，绿色内容表示扩写后新增或替换的文字。Claude Code 完成文件修改后，还执行了字数检查，确认正文落在预定的四百字区间内。

&emsp;&emsp;右下角同时显示或更新了当前任务标签，例如 `expand-watchmaker-story`。这个标签用于概括当前任务，也可以理解为 Claude Code 根据计划内容自动生成的会话名称。

&emsp;&emsp;批准 Plan 方案后，Claude Code 会按照所选确认项自动切换到相应的权限模式。具体选项和任务标签可能因 Claude Code 版本、当前已启用的权限模式和任务内容而有所不同。

#### 2.2.3 Bypass Permissions 自动执行终端命令

&emsp;&emsp;切换到 `bypass permissions on` 模式后，Claude Code 不仅可以直接修改文件，还能在无需用户再次确认的情况下自动执行终端命令。

&emsp;&emsp;本次发出的指令是：

> 💬 输入给 Claude Code 的提示词

```text
删除小故事.txt
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch02-shot-08-bypass-delete-file-fa0749d1.png" width="80%" alt="Bypass Permissions 模式自动执行终端命令删除文件"></div>

&emsp;&emsp;图中 Claude Code 将删除文件的要求转换成了 Bash 命令：

```bash
rm 小故事.txt && echo "已删除"
```

&emsp;&emsp;随后直接执行该命令并返回“已删除”的结果，中间没有出现权限确认。

&emsp;&emsp;这也是 Bypass Permissions 模式比 `accept edits on` 模式权限更高的体现：

- `accept edits on`：写入或修改文件时无需确认，但执行终端命令仍需确认。
- `bypass permissions on`：修改文件和执行终端命令都无需确认。

&emsp;&emsp;由于删除文件等高风险命令也可能被直接执行，使用该模式前需要确认当前项目目录和任务内容无误。

## <center>第三章 选择模型与推理强度</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-01-model-effort-def55c12.png" width="80%" alt="模型选择、推理强度与终端调用的关系"></div>

&emsp;&emsp;Claude Code 的模型和推理强度会共同影响响应速度、分析深度、回答质量和额度消耗。复杂调试、架构分析或长时间 Agent 任务通常需要更强的模型与更高的推理强度，简单问答和明确的小改动则更适合响应更快、消耗更低的配置。把模型能力、推理强度和任务难度对齐，可以在有限的等待时间与额度内获得更合适的结果，也能减少简单任务配置过重或复杂任务分析不足的情况。

&emsp;&emsp;这一章会先使用 `/model` 查看并切换当前账户可用的模型，并结合现有模型能力测评图理解不同模型的能力定位；随后使用 `/effort` 调整单次任务的推理强度，再通过 `claude -p` 从普通终端发起一次性调用、指定模型与推理强度、比较回答差异。最后，我们会把 `claude -p` 接入管道、脚本变量和 Codex 调用流程，形成可以继续组合的脚本工作流。

### 3.1 使用 /model 选择模型

&emsp;&emsp;在 Claude Code 交互会话中输入：

```text
/model
```

&emsp;&emsp;`/model` 用于查看当前账户可以使用的模型，也可以在不中断会话的情况下切换模型。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-shot-01-input-model-command-de49e63d.png" width="80%" alt="在 Claude Code 交互会话中输入 model 命令"></div>

&emsp;&emsp;图中执行命令后，Claude Code 保留了当前正在使用的 `Opus 4.8（1M context）`。再次进入 `/model` 的选择界面，就可以查看当前账户可用的全部模型。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-shot-02-select-model-ccede493.png" width="80%" alt="使用 model 命令查看并选择当前账户可用的 Claude 模型"></div>

- 按 `Esc` 取消选择。

### 3.2 使用 /effort 调整推理强度

&emsp;&emsp;选择模型以后，还可以使用 `/effort` 调整 Claude Code 处理任务时的推理强度：

```text
/effort
```

&emsp;&emsp;直接输入 `/effort` 会打开可视化调节界面，也可以在命令后面指定档位：

```text
/effort low
/effort medium
/effort high
/effort xhigh
/effort max
/effort ultracode
/effort auto
```

&emsp;&emsp;多个 Subagent 可以同时处理不同部分，因此在适合并行执行的任务中，整体完成速度可能更快；相应地，它也会使用更多 Token。`ultracode` 只作用于当前会话，关闭会话后不会继续保留。

### 3.3 使用 claude -p 进行终端调用

&emsp;&emsp;除了进入交互界面持续对话，也可以直接在普通终端中调用 Claude Code。

&emsp;&emsp;macOS、Linux 和 Windows PowerShell 使用相同的命令：

```text
claude -p "你是什么模型"
```

&emsp;&emsp;其中，`-p` 是单次调用模式。Claude Code 收到问题后会直接输出结果，然后返回普通终端，不会进入交互界面。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-shot-05-print-mode-query-32b5fcc1.png" width="80%" alt="使用 claude 命令的 print 模式直接提问"></div>

&emsp;&emsp;图中没有手动指定模型，因此 Claude Code 使用当前配置的默认模型 `Opus 4.8`。

&emsp;&emsp;Claude Code 仍然会读取当前项目及相关的 `CLAUDE.md` 配置，但配置文件中的文字不能真正改变正在运行的模型，实际模型由 Claude Code 的模型配置和命令参数决定。

#### 3.3.1 使用 --model 与 --effort

&emsp;&emsp;单次调用时，可以同时使用 `--model` 和 `--effort` 参数：

```text
claude -p --model sonnet --effort low "你是哪个模型"
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-shot-06-specify-model-effort-bcd86d2a.png" width="80%" alt="在终端单次调用中指定 Sonnet 模型和 low 推理强度"></div>

&emsp;&emsp;这条命令中：

- `-p`：执行一次任务并直接返回结果；
- `--model sonnet`：指定使用 Sonnet 模型；
- `--effort low`：将本次任务的推理强度设置为 `low`；
- 最后的引号内容：本次需要 Claude 完成的任务。

&emsp;&emsp;图中返回的实际模型是 `Claude Sonnet 5`。`sonnet`、`opus` 和 `haiku` 等名称属于模型别名，具体对应的模型版本可能随 Claude Code 更新而变化。

&emsp;&emsp;通用命令格式为：

```text
claude -p --model <模型> --effort <推理强度> "需要完成的任务"
```

#### 3.3.2 对比不同推理强度的回答

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-02-effort-comparison-802ad108.png" width="80%" alt="不同推理强度回答对比的观察维度"></div>

&emsp;&emsp;即使选择相同的模型，不同推理强度也可能影响回答质量。

&emsp;&emsp;下面两条命令都使用 Haiku 模型，并提出完全相同的问题，区别只在于推理强度：

```text
claude -p --model haiku --effort low "9.11 和 9.9 哪个大?只回一句话"

claude -p --model haiku --effort max "9.11 和 9.9 哪个大?只回一句话"
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-shot-07-effort-response-comparison-7ba1e75d.png" width="80%" alt="使用同一模型对比 low 和 max 推理强度的回答结果"></div>

&emsp;&emsp;第一次使用 `low` 时，模型回答 `9.11 大于 9.9`，这是错误的。第二次使用 `max` 时，模型先把 `9.9` 理解为 `9.90`，正确得出 `9.9` 大于 `9.11`。

&emsp;&emsp;本次手动演示中，`low` 与 `max` 得到了不同结果，说明推理强度可能影响分析过程；单次对比不能证明某档位的一般正确率，重要结果仍需独立验证。

#### 3.3.3 将 claude -p 嵌入脚本和工作流

&emsp;&emsp;`claude -p` 会接收引号里的任务，并把回答直接输出到终端。例如，macOS、Linux 和 Windows PowerShell 都可以使用下面这条命令：

```text
claude -p "总结当前项目的主要功能"
```

&emsp;&emsp;命令输出结果后就会退出，不会进入持续对话。这种一次性调用适合放进脚本或自动化工作流，让 Claude Code 完成一个明确步骤后，再由后续流程继续处理结果。

&emsp;&emsp;我们也可以在 Codex 中提出需求，让 Codex 使用 `claude -p` 调用 Claude Code，并询问它当前正在使用的模型。Codex 随后执行终端命令，获得 Claude Code 返回的 `Claude Opus 4.8` 模型信息。

> 💬 输入给 Codex 的提示词

```text
使用 `claude -p` 终端命令来调用 Claude Code，问下它现在是什么模型，并输出你的完整终端命令。
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch03-shot-08-codex-calls-claude-code-f0600aba.png" width="80%" alt="在 Codex 中通过终端命令调用 Claude Code"></div>

&emsp;&emsp;这说明只要一个工具能够执行终端命令，就可以把 Claude Code 作为外部命令加入自己的工作流程。Codex 负责安排任务和执行命令，Claude Code 负责处理收到的问题，然后 Codex 再读取它输出的结果并继续工作。

&emsp;&emsp;图中第一次调用返回了 `Not logged in`，说明当时的隔离运行环境无法读取本机的 Claude 登录状态。允许 Codex 在能够访问登录信息的环境中执行相同命令后，调用成功完成。

&emsp;&emsp;把 Claude Code 接入其他工具时，需要满足两个条件：

- 当前执行环境已经登录 Claude Code，或者配置了可用的认证方式；
- 调用方拥有执行 `claude -p` 终端命令的权限。

## <center>第四章 查看 Claude Code 的状态与用量</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch04-01-status-usage-2d1cd25b.png" width="80%" alt="Claude Code 状态、配置、用量与统计信息总览"></div>

&emsp;&emsp;Claude Code 的任务持续时间变长、同时维护的会话增多以后，只看最终回复不足以判断运行是否健康。版本、模型、工作目录和连接状态会影响当前会话，配置项决定自动压缩、权限与通知等行为，额度和上下文占用则决定任务还能持续到什么程度。把这些信息集中观察，能够更早发现目录不符、配置不合适、额度接近上限或上下文持续膨胀等情况。

#### 4.1.4 查看 Usage 用量

&emsp;&emsp;如果想知道“当前用了多少额度、什么时候恢复”，可以从 `Status` 使用右方向键切换到 `Usage`。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch04-shot-04-usage-panel-cd3610ea.png" width="80%" alt="查看 Claude Code 当前会话和本周的 Usage 用量"></div>

&emsp;&emsp;页面顶部显示当前会话的消耗：

- `Total cost`：当前会话的成本统计；
- `Total duration (API)`：等待模型返回结果的时间；
- `Total duration (wall)`：会话实际经过的总时间；
- `Total code changes`：本次会话新增和删除的代码行数。

&emsp;&emsp;`Usage by model` 会分别统计每个模型的 `input`、`output`、`cache read` 和 `cache write`。其中，`cache read` 表示重复使用已有上下文，长会话中数值较高是正常现象。

&emsp;&emsp;页面中部的进度条用于查看额度：

- `Current session`：当前 5 小时额度窗口的使用比例和重置时间；
- `Current week (all models)`：所有模型合计的本周用量；
- `Current week (Fable)`：Fable 模型单独计算的本周用量。

&emsp;&emsp;这里的 `Current session` 指当前短周期额度窗口，不只是当前这一次对话。

&emsp;&emsp;页面下方还会分析哪些使用习惯正在增加额度消耗。图中提示，大部分用量发生在上下文超过 150k 的长会话中。任务仍有关联时可以使用 `/compact` 压缩上下文；切换到新任务时可以使用 `/clear` 清空旧上下文。

&emsp;&emsp;这些分析主要根据当前电脑上的本地会话估算，不一定包含其他设备或网页端产生的完整用量。

```text
/config
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch04-shot-07-input-config-command-11f65eda.png" width="80%" alt="输入 config 命令并查看可直接选择的配置项"></div>

&emsp;&emsp;图中显示的部分配置包括：

- `agentPushNotifEnabled`：是否启用 Agent 推送通知，可选 `true` 或 `false`；

- `autoCompact`：是否在上下文较长时自动压缩；
- `autoInstallIdeExtension`：是否自动安装 IDE 扩展。

&emsp;&emsp;这里显示的是部分候选配置，实际项目可能会随 Claude Code 版本和当前环境有所不同。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch04-02-statusline-flow-981a00fb.png" width="80%" alt="Claude Code 状态栏从配置需求到结果确认的流程"></div>

#### 4.3.2 由 Subagent 完成配置

&emsp;&emsp;发出需求后，主会话没有直接修改配置，而是启动了一个名为 `statusline-setup` 的 Subagent。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch04-shot-10-statusline-subagent-07d9d19d.png" width="80%" alt="主会话启动 statusline-setup 子代理配置状态栏"></div>

## <center>第五章 管理 Claude Code 的上下文</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-01-context-lifecycle-539de519.png" width="80%" alt="Claude Code 上下文与会话的管理生命周期"></div>

&emsp;&emsp;Claude Code 的对话持续变长时，历史消息、工具结果、项目规则和技能会不断占用上下文空间。旧内容积累过多，不仅会增加额度消耗，还可能分散模型对当前任务的注意力，让会话逐渐偏离最初目标；退出或清空对话以后，如果没有明确的恢复入口和会话名称，跨会话衔接也容易丢失任务进度、已完成修改与剩余计划。

&emsp;&emsp;这一章会先使用 `/context` 和 `/context all` 查看上下文组成，再用 `/btw` 在主任务执行期间提出不写入主会话历史的临时问题；随后通过 `/compact` 压缩当前任务、用 `/clear` 开始新任务，并分别使用 `/resume`、`claude --resume` 和 `claude -c` 恢复历史会话。我们还会用 `/rename` 命名会话、通过 `claude -r` 查找记录，最后认识会话文件的存储位置以及 `/codex:transfer` 的迁移方式。

### 5.1 使用 /context 查看当前上下文

&emsp;&emsp;前面章节介绍的用量页面可以查看会话的使用量和成本。随着对话不断进行，历史消息会持续进入后续请求的上下文，Claude Code 使用的系统提示词、工具、Agent、记忆文件和技能等内容也会占用上下文空间。

&emsp;&emsp;如果不管理上下文，可能出现三个问题：

- 内容逐渐接近模型的上下文窗口上限，无法继续完整放入更多信息；
- 大量旧内容可能分散模型的注意力，使它难以专注于当前任务；
- 每次请求需要处理的内容越来越多，用量和成本也可能随之增加。

&emsp;&emsp;在 Claude Code 交互会话中输入：

```text
/context
```

&emsp;&emsp;可以查看当前上下文的整体使用情况。图中主要包含：

- 当前使用的模型和模型 ID；
- 已使用的 Token 数量、上下文总容量和占用百分比；
- `System prompt`：系统提示词占用；
- `System tools`：系统工具占用；
- `Custom agents`：自定义 Agent 占用；
- `Memory files`：记忆文件占用；
- `Skills`：技能占用；
- `Messages`：当前会话消息占用；
- `Free space`：上下文窗口剩余空间。

#### 5.1.2 使用 /btw 提问，不增加主会话历史

&emsp;&emsp;处理任务时，我们有时会临时想到一个问题，但又不希望这个问题干扰正在进行的工作。此时可以使用：

```text
/btw <问题>
```

&emsp;&emsp;`btw` 是 `by the way` 的缩写，可以理解为“顺便问一下”。

&emsp;&emsp;先在主会话中发出待办清单样式修改任务：

> 💬 输入给 Claude Code 的提示词

```text
帮我把当前的待办清单背景变成暖色卡片吧
```

&emsp;&emsp;图中 Claude Code 正在修改待办清单的卡片背景，并继续检查项目文件和页面效果。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-03-main-task-style-update-cba48560.png" width="80%" alt="Claude Code 正在执行待办清单样式修改任务"></div>

&emsp;&emsp;即使 Claude Code 仍在处理主任务，我们也可以输入一个旁支问题：

> 💬 输入给 Claude Code 的提示词

```text
/btw 现在项目里有哪些文件
```

&emsp;&emsp;这个问题会独立运行，不会中断正在进行的主任务。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-04-btw-during-main-task-a3f0a0cf.png" width="80%" alt="主任务执行期间使用 btw 提出旁支问题"></div>

&emsp;&emsp;回答会显示在单独的浮层中。`/btw` 可以看到当前会话中已有的对话、文件信息和先前结论，但这次问题及其回答不会加入主会话历史，因此不会继续增加主会话的上下文占用。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-05-btw-answer-overlay-8da47d78.png" width="80%" alt="btw 在独立浮层中根据现有上下文回答问题"></div>

&emsp;&emsp;需要注意，`/btw` 没有工具权限，不能重新读取文件、执行终端命令或搜索资料。图中的回答来自当前对话里已经出现过的信息，不一定代表磁盘上的最新状态。如果需要实时检查项目目录，仍应使用正常任务或执行：

```text
! ls -la
```

&emsp;&emsp;`/btw` 适合快速确认一个已有信息，并且每次只返回一次回答。如果发现这个问题还需要继续讨论，可以按 `f` 将它分叉出去；否则按 `Esc` 关闭浮层，继续原来的任务。

&emsp;&emsp;关闭 `/btw` 的回答浮层后，原来的主任务仍在继续执行，不会因为这次旁支提问而被打断。

&emsp;&emsp;图中可以看到，Claude Code 已经完成了待办清单的样式修改：页面从原来的冷色背景变成了暖色卡片风格。这说明 `/btw` 会独立处理临时问题，主任务仍会按照原来的流程继续运行并完成。

&emsp;&emsp;因此，`/btw` 特别适合在长任务执行过程中临时确认一个已有信息：我们可以得到快速回答，同时不打断主任务，也不会把旁支问答加入主会话历史。

> `/btw` 不会增加主会话历史，但仍然会调用模型。由于它会复用主会话的提示词缓存，额外用量通常较少，但并不是完全零消耗。

### 5.2 使用 /compact 与 /clear 管理上下文

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-02-context-compression-21421f13.png" width="80%" alt="Claude Code 压缩上下文与清空上下文的差异"></div>

&emsp;&emsp;上下文不会一直增长到完全用满后才报错。Claude Code 默认启用自动压缩，会在上下文接近模型窗口上限时主动处理旧内容：先清理较早的工具输出，必要时再把历史对话整理成摘要，为后续任务释放空间。

&emsp;&emsp;不同模型的上下文窗口不同，自动压缩的触发位置也会有所区别：

<div align="center">
<table width="80%">
<thead><tr><th>场景</th><th>自动压缩的位置</th></tr></thead>
<tbody>
<tr><td>标准 200K 上下文窗口</td><td>以 200K 作为自动压缩计算容量，在接近窗口上限时触发；没有统一公开的精确 Token 数值</td></tr>
<tr><td>Sonnet 5 默认 1M 上下文窗口</td><td>本课 Claude Code 2.1.210 手动实测中，约在 <code>967K tokens</code> 时自动压缩，约占上下文窗口的 96.7%；具体触发点受版本和环境配置影响，不应视为长期稳定常量</td></tr>
</tbody>
</table>

&emsp;&emsp;自动压缩能够避免大多数上下文超限错误，但摘要不一定能保留早期对话的全部细节。重要、需要长期遵守的项目规则，应写入 `CLAUDE.md`，不能只依赖聊天记录。

#### 5.2.1 使用 /compact 压缩当前对话

&emsp;&emsp;不需要等到系统自动处理，我们也可以输入：

```text
/compact
```

&emsp;&emsp;主动压缩当前对话。Claude Code 会把已有历史整理为更短的摘要，然后继续当前会话。

&emsp;&emsp;这里主要被压缩的是 `/context` 中的 `Messages`。它不只包含用户与 Claude 的对话，还包含会话中读取的文件内容、执行命令产生的输出以及工具调用结果。压缩时，较早的工具输出可能先被清理，其余历史内容会被整理为摘要，而不是逐字保留。

&emsp;&emsp;`/compact` 后面还可以补充希望保留的重点，例如：

> 💬 输入给 Claude Code 的提示词

```text
/compact 保留当前任务、已修改文件和剩余计划
```

&emsp;&emsp;这样可以告诉 Claude Code 在生成摘要时重点保留哪些信息。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-07-compact-command-4e1094dc.png" width="80%" alt="输入 compact 命令手动压缩上下文"></div>

&emsp;&emsp;确认命令后，Claude Code 开始整理现有对话。图中会显示 `Compacting conversation` 和当前处理进度，完成后便可以继续输入任务。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-08-compacting-conversation-1dddf157.png" width="80%" alt="Claude Code 正在压缩当前对话"></div>

&emsp;&emsp;压缩保留的是历史内容的摘要，而不是逐字保留全部消息，因此重要要求仍建议写入项目文件。

#### 5.2.2 使用 /clear 清空上下文

&emsp;&emsp;如果准备开始一个完全不同的任务，可以输入：

```text
/clear
```

&emsp;&emsp;`/clear` 会清空当前会话的消息历史，效果类似重新开始这段对话。系统提示词、工具、`CLAUDE.md`、记忆和已经加载的 Skill 仍可能占用上下文，因此状态栏中的 `ctx` 会明显下降，但不保证回到 `0%`。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-09-clear-context-41b6eecb.png" width="80%" alt="输入 clear 命令清空当前上下文"></div>

&emsp;&emsp;清空当前上下文并不会删除之前的会话记录。旧会话仍然保存在磁盘中，之后可以通过 `/resume` 重新打开。

&emsp;&emsp;简单区分：

<tbody>
<tr><td><code>/compact</code></td><td>仍要继续当前任务，只是希望压缩历史内容并释放上下文空间</td></tr>
<tr><td><code>/clear</code></td><td>准备开始新的任务，不再需要当前对话内容参与后续回答</td></tr>
</tbody>
</table>
</div>

### 5.3 使用 /resume 恢复会话

&emsp;&emsp;执行 `/clear` 或退出 Claude Code 后，历史会话并不会立即消失。Claude Code 会把会话记录保存在磁盘中，我们可以通过多种方式继续之前的对话。

#### 5.3.1 使用 /resume 选择历史会话

&emsp;&emsp;在 Claude Code 交互会话中输入：

```text
/resume
```

&emsp;&emsp;确认命令后会进入会话恢复流程。如果中途取消，界面会显示 `Resume cancelled`；再次输入 `/resume` 即可重新选择。

#### 5.3.2 使用 claude --resume 精确恢复

&emsp;&emsp;退出 Claude Code 时，终端通常会显示类似下面的提示：

```text
Resume this session with:
claude --resume <会话ID>
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-13-resume-id-command-8b3f758b.png" width="80%" alt="退出 Claude Code 后显示会话恢复命令"></div>

&emsp;&emsp;此时即使进入消息历史为空的新对话，旧会话仍保留在磁盘中，需要时仍可恢复。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-14-blank-session-after-clear-044a2b58.png" width="80%" alt="clear 后进入消息历史为空的新对话"></div>

&emsp;&emsp;提示中的一长串字符是本次会话的唯一 ID。直接复制并运行终端给出的完整命令：

```bash
claude --resume <会话ID>
```

&emsp;&emsp;即可准确返回刚刚关闭的那段会话。每个会话的 ID 不同，实际使用时应以自己终端显示的命令为准。

#### 5.3.3 使用 claude -c 继续最近会话

&emsp;&emsp;如果只想继续当前项目目录下最近使用的会话，不需要复制会话 ID，可以运行：

```bash
claude -c
```

&emsp;&emsp;这里的 `c` 是 `continue` 的缩写，表示继续最近一次会话。图中依次演示了使用完整会话 ID 精确恢复，以及输入 `claude -c` 继续最近一次会话。

### 5.5 会话文件的存储与迁移

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-03-session-migration-b2802d97.png" width="80%" alt="Claude Code 会话文件的存储、读取与迁移关系"></div>

&emsp;&emsp;Claude Code 会持续把会话记录保存在用户目录下：

```text
~/.claude/projects/<项目路径编码>/<会话ID>.jsonl
```

&emsp;&emsp;其中：

- 项目文件夹名称由项目的绝对路径转换而来；
- JSONL 文件名通常就是会话 ID；
- 文件中保存了用户消息、Claude 的回答、工具调用、工具结果等会话记录；
- 前面退出会话时看到的 `--resume` 参数，对应的就是这里的会话 ID。

&emsp;&emsp;图中可以看到，对应项目的目录下保存了多个 JSONL 文件，选中的文件名与前面得到的会话 ID 一致。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-19-session-jsonl-files-01cc0808.png" width="80%" alt="查看项目目录中以会话 ID 命名的 JSONL 文件"></div>

&emsp;&emsp;查看文件信息，可以进一步确认它位于用户目录的 `.claude/projects/` 下，并归属于当前项目对应的文件夹。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch05-shot-20-session-file-location-4ab90939.png" width="80%" alt="确认 Claude Code 会话文件的存储位置"></div>

#### 5.5.1 让其他 Agent 了解历史任务

&emsp;&emsp;JSONL 文件保存了较完整的任务过程。只要其他 Agent 有权读取这个文件，就可以让它先整理之前的任务目标、已经完成的内容、修改过的文件和下一步计划，再继续工作。

&emsp;&emsp;例如可以告诉其他 Agent：

> <b>输入给其他 Agent 的提示词</b>

```text
请读取这个 Claude Code 会话文件，整理当前任务的目标、已完成内容、
修改过的文件和未完成事项，然后继续完成任务：

~/.claude/projects/<项目路径编码>/<会话ID>.jsonl
```

&emsp;&emsp;需要注意，读取会话文件可以帮助其他 Agent 理解任务进度，但不代表它会自动变成该工具原生的可恢复会话。不同 Agent 使用的会话格式可能不同。

#### 5.5.2 （拓展）会话迁移到 Codex

&emsp;&emsp;Codex 官方提供了 Claude Code 插件，可以把当前 Claude Code 会话转换为一个可继续使用的 Codex 任务。

&emsp;&emsp;安装并配置插件后，可以在 Claude Code 中运行：

```text
/codex:transfer
```

&emsp;&emsp;也可以明确指定会话文件：

```text
/codex:transfer --source ~/.claude/projects/<项目路径编码>/<会话ID>.jsonl
```

&emsp;&emsp;转换完成后会返回类似下面的命令：

```bash
codex resume <session-id>
```

&emsp;&emsp;运行该命令，即可在 Codex 中继续处理原来的任务。

&emsp;&emsp;因此，这里的“迁移”是由官方工具读取并转换 Claude Code 会话，而不是直接把 JSONL 文件复制到 Codex 的会话目录。

&emsp;&emsp;完整说明请参考 [Codex plugin for Claude Code 官方 README](https://github.com/openai/codex-plugin-cc/blob/main/README.md)。

> JSONL 会话文件是本地明文数据，可能包含代码、终端输出甚至敏感信息，不要随意发送给不可信的工具或服务。

## <center>第六章 使用 CLAUDE.md 固定项目规则</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch06-01-claude-md-rules-6671f5ab.png" width="80%" alt="项目规则、全局规则与自动记忆的作用层级"></div>

&emsp;&emsp;Claude Code 新会话的消息上下文会重新开始，不会自动保留上一段对话中的全部口头约定。项目结构、运行命令、团队规范和禁止修改的范围如果只在聊天中说明，每次重新启动都可能需要重复输入，执行方式也容易前后不一致；跨项目通用的回答习惯和个人偏好同样需要稳定载体，才能在不同会话中持续生效。

&emsp;&emsp;这一章会先使用 `/init` 分析当前项目并生成项目级 `CLAUDE.md` 初稿，再补充长期有效的项目说明，并在新会话中检查规则是否加载；随后我们会区分项目级与全局 `CLAUDE.md` 的作用范围，把团队规则和个人通用偏好放到合适位置。最后，我们会通过 `/memory` 查看当前项目的记忆入口，认识自动记忆的保存目录、写入方式与新会话加载机制。

### 6.1 使用 /init 创建项目说明

&emsp;&emsp;每次新建 Claude Code 会话时，对话上下文都会重新开始。如果没有专门的项目说明，我们可能需要反复介绍项目结构、运行方式和开发规范。

&emsp;&emsp;`CLAUDE.md` 是项目的长期说明文件。Claude Code 进入项目后会自动读取它，从而提前了解：

- 项目用途和整体结构；
- 使用的技术和运行方式；
- 编码风格与命名规范；
- 测试、构建和检查命令；
- 重要限制以及需要避免的操作。

&emsp;&emsp;这样可以减少重复说明，也能降低 Claude Code 使用错误命令或不符合项目规范的实现方式的概率。

&emsp;&emsp;需要注意，`CLAUDE.md` 提供的是项目背景和操作指导，并不是权限控制文件。内容应尽量简洁，只保留长期有效且无法轻易从代码中看出的规则，避免占用不必要的上下文。

#### 6.1.1 使用 /init 生成 CLAUDE.md

&emsp;&emsp;在 Claude Code 交互会话中输入：

```text
/init
```

&emsp;&emsp;`/init` 会读取当前项目中的已有文件，分析项目结构、技术栈和运行方式，然后在项目根目录生成一份初始的 `CLAUDE.md`。

&emsp;&emsp;它不会创建新的代码项目，也不会初始化 Git、安装依赖或修改项目功能。

&emsp;&emsp;图中，Claude Code 先检查了当前目录，识别出这是一个由 `todo.html` 构成的零依赖单文件前端应用。由于项目中没有构建配置、包管理文件和测试命令，它没有凭空补充这些内容，而是根据实际文件整理项目约定。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch06-shot-01-init-project-analysis-04bf67e8.png" width="80%" alt="使用 init 分析当前项目并生成 CLAUDE.md"></div>

&emsp;&emsp;随后，Claude Code 开始生成 `CLAUDE.md`。生成后的文件包含项目概览、运行与调试方式、数据存储说明和核心架构等内容。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch06-shot-02-generated-claude-md-9052288f.png" width="80%" alt="查看 init 生成的 CLAUDE 项目说明"></div>

&emsp;&emsp;`/init` 生成的是一份可以继续完善的初稿。生成后应当检查其中是否存在不准确或容易过时的描述，并补充团队规范、固定测试命令、禁止修改的目录等重要信息。

&emsp;&emsp;以后在这个项目目录中启动 Claude Code 时，它会自动读取 `CLAUDE.md`。因此，可以把两者简单理解为：

### 6.2 在新会话中加载 CLAUDE.md

&emsp;&emsp;生成项目级 `CLAUDE.md` 后，我们还可以继续编辑它，把希望 Claude Code 长期遵守的项目规则写进去。

&emsp;&emsp;例如，在文件末尾增加下面的回答规范：

```markdown
## 回答规范

- 回答必须用中文
- 每次回答结束要带上当前时间
```

&emsp;&emsp;图中在项目级 `CLAUDE.md` 中加入了两条规则，分别用于约束回答语言和结尾内容。

&emsp;&emsp;项目根目录和用户目录中的 `CLAUDE.md` 会在 Claude Code 启动会话时加载。修改文件后，为了确保新规则被完整读取，可以退出当前会话并在同一项目目录中启动一个新会话。

&emsp;&emsp;如果规则没有按预期生效，可以先确认终端所在目录是否为项目目录，再检查 `CLAUDE.md` 的文件名和保存位置。

### 6.3 配置全局 CLAUDE.md

&emsp;&emsp;前面创建的是项目级 `CLAUDE.md`，适合记录当前项目的架构、运行方式、测试命令、代码规范和注意事项，只对当前项目生效。

&emsp;&emsp;如果一些规则希望应用到所有项目，可以写入用户级的全局文件。

&emsp;&emsp;macOS 和 Linux：

```text
~/.claude/CLAUDE.md
```

&emsp;&emsp;Windows PowerShell：

```text
C:\Users\<用户名>\.claude\CLAUDE.md
```

&emsp;&emsp;例如，可以在全局文件中要求 Claude Code：

- 优先采用简单方案，避免过度设计；
- 修改代码时尽量缩小改动范围；
- 遇到不确定的需求时先说明假设；
- 完成任务前进行必要验证；
- 使用固定语言和回答风格。

&emsp;&emsp;图中展示的是一份全局 `CLAUDE.md` 的部分内容，其中包含简洁至上、第一性原理、事实为本和最小改动等通用规则。

&emsp;&emsp;全局文件不适合记录某个项目独有的文件路径、技术栈和运行命令，这些内容应继续放在项目级 `CLAUDE.md` 中。

<tr><td>项目根目录中的 <code>CLAUDE.md</code></td><td>当前项目</td><td>项目架构、运行与测试命令、技术约定和项目注意事项</td></tr>

&emsp;&emsp;Claude Code 启动会话时，会先加载用户级规则，再加载当前项目的规则，两者都会进入模型上下文。它们并不是覆盖关系，但如果针对同一件事写了互相矛盾的要求，Claude 可能任意采用其中一条，因此最好让全局文件负责通用原则，让项目文件负责具体项目约定。

&emsp;&emsp;下面是对应的中文版本。可以将其复制到全局 `CLAUDE.md` 中，也可以参考其中的思想，整理成更符合自己习惯的规则：

```markdown
# CLAUDE.md

用于减少大语言模型在编码过程中常见错误的行为准则。可根据需要与项目专属说明合并使用。

**取舍说明：**这些准则更偏向谨慎，而不是追求速度。对于简单任务，可以自行判断是否全部执行。

## 1. 编码前先思考

**不要想当然。不要掩饰困惑。明确说明取舍。**

开始实现前：

- 明确说明你的假设。如果存在不确定性，就询问。
- 如果存在多种理解方式，把它们列出来，不要默默选择其中一种。
- 如果存在更简单的方案，要指出来。必要时应对不合理需求提出异议。
- 如果某些地方不清楚，就暂停。明确指出困惑之处，并进行询问。

## 2. 简单优先

**只编写能够解决问题的最少代码，不做任何推测性设计。**

- 不添加用户没有要求的功能。
- 不为只使用一次的代码创建抽象层。
- 不添加用户没有要求的“灵活性”或“可配置性”。
- 不为不可能发生的情况添加错误处理。
- 如果你写了 200 行，但其实 50 行就能完成，就重新简化。

问问自己：

“资深工程师会不会觉得这个方案过度复杂？”

如果答案是会，那就继续简化。

## 3. 最小范围修改

**只修改必须修改的内容。只清理由你自己造成的问题。**

修改现有代码时：

- 不要顺手“优化”相邻代码、注释或格式。
- 不要重构没有问题的代码。
- 保持项目现有风格，即使你个人会选择其他写法。
- 如果发现无关的废弃代码，可以指出，但不要删除。

当你的修改产生无用代码时：

- 删除因为你的修改而变得无用的 import、变量或函数。
- 除非用户明确要求，否则不要删除原本就存在的废弃代码。

判断标准：

每一行改动都应该能够直接追溯到用户的需求。

## 4. 以目标驱动执行

**定义成功标准，并持续循环，直到验证完成。**

将任务转化为可以验证的目标：

- “添加参数校验”
  → “先为非法输入编写测试，然后让测试通过”

- “修复这个 Bug”
  → “先编写能够复现该 Bug 的测试，然后让测试通过”

- “重构 X”
  → “确保重构前后测试都能够通过”

对于多步骤任务，先给出一个简短计划：

1. [步骤] → 验证方式：[检查内容]
2. [步骤] → 验证方式：[检查内容]
3. [步骤] → 验证方式：[检查内容]

清晰的成功标准可以让你独立循环执行。

模糊的标准，例如“让它能用”，会导致执行过程中频繁需要用户确认。

---

如果这些准则有效，应该能看到以下结果：

- Diff 中不必要的修改更少；
- 因为过度设计而返工的次数更少；
- 澄清问题发生在开始实现之前，而不是出错之后。
```

## <center>第七章 使用 Skill 复用工作流程</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch07-01-skill-workflow-f2c65c41.png" width="80%" alt="Claude Code Skill 从定义到稳定触发的工作流"></div>

&emsp;&emsp;同一类任务如果每次都靠临时口述步骤、判断标准和注意事项，要求很容易在不同会话里被遗漏，执行顺序、处理尺度和结果格式也可能前后不一致。Skill 可以把这些内容沉淀到可复用的 `SKILL.md` 中，让 Claude Code 根据任务自动匹配，或者由我们明确指定，从而把一次性的说明变成能够反复触发的固定流程。

&emsp;&emsp;这一章会先认识 Skill 的作用、目录结构、标准格式和触发方式，再安装并使用 Humanizer-zh，通过 `/skills` 查看和禁用 Skill；随后创建一个三点总结 Skill，比较自动匹配、`CLAUDE.md` 规则与 Hook 三种调用方式，最后使用 `/plugin` 安装常用 Skill，并整理一组推荐选择。

### 7.1 认识 Skill

&emsp;&emsp;在 Claude Code 中，Skill 是一套可以重复使用的任务说明。我们可以把某类工作的操作步骤、判断标准和注意事项提前写进 `SKILL.md`，以后遇到相似任务时，让 Claude Code 直接读取并执行，不必每次重新输入一长段要求。

&emsp;&emsp;Skill 与前面介绍的两类长期信息并不完全相同：

- `CLAUDE.md` 主要保存长期生效的项目规则或用户规则；
- 自动记忆主要保存跨会话仍有价值的信息；
- Skill 更适合封装某一类具体任务的工作方法，需要时再调用。

&emsp;&emsp;Claude Code 通常只需先了解 Skill 的名称和描述，判断任务需要它时才读取完整内容。这种按需加载机制可以减少无关说明对上下文的占用。不过，Skill 真正被调用后，其任务说明仍会进入当前上下文，并不是完全不消耗上下文。

&emsp;&emsp;如果某项工作会反复出现，我们也可以直接让 Claude Code 帮忙创建一个项目级 Skill。图中提出的需求，是希望在项目中创建一套自动处理待办事项的固定流程。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch07-shot-01-create-project-skill-request-0e172932.png" width="80%" alt="向 Claude Code 提出创建项目 Skill 的需求"></div>

### 7.2 Skill 的标准格式

&emsp;&emsp;项目级 Skill 通常放在：

```text
<项目目录>/.claude/skills/<skill-name>/SKILL.md
```

&emsp;&emsp;用户级 Skill 通常放在：

```text
~/.claude/skills/<skill-name>/SKILL.md
```

&emsp;&emsp;一个常见的 `SKILL.md` 由两部分组成：顶部是 YAML 元数据，下面是 Markdown 格式的具体说明。例如：

```markdown
---
name: humanizer-zh
description: 去除中文文本中的 AI 写作痕迹。适用于用户要求润色、改写或降低 AI 味时。
---

# Humanizer-zh

## 工作步骤

1. 读取目标文章。
2. 按规则识别并修改明显的 AI 写作模式。
3. 保存结果并说明主要改动。
```

&emsp;&emsp;其中：

- `name` 是 Skill 的名称；
- `description` 说明它能做什么、应该在什么情况下使用，也是 Claude Code 判断是否需要自动调用它的重要依据；
- YAML 元数据之后的 Markdown 正文，负责记录完整步骤、规则、示例和验收标准。

&emsp;&emsp;上面展示的是最精简、最常用的写法，并不代表两个 `---` 之间只能出现 `name` 和 `description`。例如，后面要安装的 Humanizer-zh 使用了扩展格式：

```yaml
---
name: humanizer-zh
description: |
  去除文本中的 AI 生成痕迹。适用于编辑或审阅文本，
  使其听起来更自然、更像人类书写。
allowed-tools:
  - Read
  - Write
  - Edit
  - AskUserQuestion
metadata:
  trigger: 编辑或审阅文本，去除 AI 写作痕迹
  source: 翻译自 blader/humanizer，参考 hardikpandya/stop-slop
---
```

&emsp;&emsp;这些写法同样属于 YAML 元数据：

- `description: |` 是 YAML 的多行文本写法，适合放置较长的描述；它仍然属于同一个 `description` 字段；
- `allowed-tools` 表示 Skill 生效期间可以免去逐次确认的工具。它不是“只能使用这些工具”的限制，其他工具仍按 Claude Code 的正常权限规则处理；
- `metadata` 用来保存来源、作者、版本等附加信息。Humanizer-zh 中的 `trigger` 是作者自定义的元数据名称，不等同于 Claude Code 的强制触发开关；自动匹配仍主要依赖 `description`。

&emsp;&emsp;因此，7.2 中的示例是便于初学者理解的最小格式，Humanizer-zh 则是在这个基础上加入可选字段。安装第三方 Skill 时通常应保留原有 frontmatter，但也要注意部分扩展字段可能因客户端和版本不同而有兼容性差异。

&emsp;&emsp;除了 `SKILL.md`，同一目录还可以按需存放脚本、模板或参考资料。对于简单 Skill，仅使用一个 `SKILL.md` 就足够了。

### 7.5 创建自己的 Skill

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch07-02-skill-trigger-cycle-34f9c4d6.png" width="80%" alt="Claude Code Skill 从创建到触发验证的闭环"></div>

#### 7.5.4 使用 Hook 稳定触发 Skill

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch07-03-hook-trigger-83b44ea2.png" width="80%" alt="Hook 接收用户提示并稳定触发 Skill 的链路"></div>

&emsp;&emsp;前面的做法是在项目级 `CLAUDE.md` 中声明调用规则。它可以提高 Skill 的触发概率，但仍然需要 Claude Code 读取并遵循这条规则。当上下文很长或规则较多时，依然存在被忽略的可能。

&emsp;&emsp;如果希望关键词检测不再依赖模型自行判断，可以使用 Hook。

##### 7.5.4.1 什么是 Hook

&emsp;&emsp;Hook 是一种自动化触发机制。它会监听 Claude Code 的特定事件，并在事件发生时执行预先配置好的命令。例如：

- `UserPromptSubmit`：用户提交提示词时触发；
- `PreToolUse`：Claude Code 调用工具前触发；
- `PostToolUse`：工具执行完成后触发。

&emsp;&emsp;可以把 Hook 理解为安装在 Claude Code 工作流程中的自动触发器：先指定一个触发时机，当对应事件发生时，Claude Code 就会自动运行预先配置的命令或程序。

&emsp;&emsp;一个 Hook 通常包含三个部分：

- **触发事件**：决定什么时候运行，例如提交提示词、调用工具或结束会话；
- **处理程序**：决定触发后执行什么，本例执行的是一条终端命令；
- **处理结果**：可以记录信息、补充上下文，或者阻止某个操作继续执行。

&emsp;&emsp;Hook 的运行过程可以简单理解为：

```text
事件发生 → Hook 接收事件数据 → 执行检测命令 → 返回处理结果
```

&emsp;&emsp;它和 `CLAUDE.md` 的主要区别在于：`CLAUDE.md` 保存的是需要模型读取和遵循的文字规则，而 Hook 是 Claude Code 在指定事件发生时主动执行的自动化程序。因此，即使当前对话很长，Hook 中的检测代码仍然会在对应事件发生时运行。

&emsp;&emsp;本例的检测命令依赖 `jq`。配置 Hook 前，先用 `jq --version` 确认它可以运行；如果缺失，macOS 可以通过 Homebrew 的 `brew install jq` 安装，Ubuntu 或 Debian 可以依次使用 `sudo apt update` 和 `sudo apt install jq`，Windows 可以使用 `winget install jqlang.jq`。本课只提供这些安装入口，不执行命令。

&emsp;&emsp;最小排查顺序如下：

- 出现 `jq: command not found` 时，先完成安装，确认 `jq` 已进入 `PATH`，再重新打开终端和 Claude Code 会话；
- 出现 JSON 配置错误时，使用 `jq empty .claude/settings.local.json` 检查语法，重点查看括号、逗号和引号，同时保留文件中原有的其他设置；
- Hook 未触发时，确认 `.claude/settings.local.json` 位于当前项目、事件名为 `UserPromptSubmit`，再开启新会话，并检查 Hook 执行记录或 `additionalContext` 注入记录。

&emsp;&emsp;这里使用 `UserPromptSubmit`，让 Claude Code 每次收到用户输入时都运行一次检测。输入的要求是：

> 💬 输入给 Claude Code 的提示词

```text
创建一个 Hook，只要我的输入中包含“总结”二字，就要使用总结 Skill
```

&emsp;&emsp;Claude Code 检查现有配置后，把 Hook 写入 `.claude/settings.local.json`，同时保留文件中原有的权限设置。

&emsp;&emsp;这个 Hook 的主要逻辑是：

## <center>第八章 使用 MCP 扩展 Claude Code 的能力</center>

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch08-01-mcp-connection-ba7f0038.png" width="80%" alt="Claude Code 通过 MCP 连接外部地图工具的链路"></div>

&emsp;&emsp;模型本身不能天然访问实时地图、企业系统或其他外部专业服务。没有地图工具时，Claude Code 只能根据已有知识和常见速度估算路线；MCP 提供一套标准连接方式，让 Claude Code 能够调用 MCP Server，从外部服务取得实际路线、距离和预计用时。

&emsp;&emsp;这一章会先对比接入地图 MCP 前后的回答，再准备高德地图 API Key，并使用 `claude mcp add` 完成安装；随后通过 `/mcp` 检查连接状态，在新会话中验证工具调用和骑行查询结果，同时认识 `local`、`project`、`user` 三种范围与 `.mcp.json` 共享配置，最后整理常用 MCP 和社区目录的选择方法。

### 8.1 认识 MCP

&emsp;&emsp;MCP 的全称是 Model Context Protocol，可以理解为一种让 Claude Code 连接外部工具和数据服务的标准协议。

&emsp;&emsp;Claude Code 本身擅长理解和生成内容，但不会自动拥有所有专业服务的数据。例如，直接询问两地之间的骑行时间时，它可能只能根据地理常识、距离和平均速度进行估算。

&emsp;&emsp;接入地图 MCP 后，Claude Code 就可以调用地图服务提供的工具，查询实际道路上的路线、距离和预计用时。

```text
用户提出问题 → Claude Code 调用 MCP 工具 → MCP Server 查询外部服务 → 返回结果
```

&emsp;&emsp;MCP 和 Skill 的作用不同：

- Skill 保存可以重复使用的操作流程和提示规则；
- MCP 连接外部服务，为 Claude Code 提供新的工具和数据能力。

### 8.2 使用 claude mcp add 安装高德地图 MCP

#### 8.2.1 准备高德地图 API Key

&emsp;&emsp;接入前需要注册高德开放平台账号、创建应用并申请 API Key。创建 Key 时，服务平台应选择“Web 服务”。

&emsp;&emsp;具体步骤可以查看：[高德地图创建应用和 Key](https://lbs.amap.com/api/mcp-server/create-project-and-key)。

&emsp;&emsp;API Key 属于访问凭证。课程中的可复制文本和图片统一使用字面量占位符 `xx`，实际操作时将 `xx` 替换为自己申请的 Key；课件中不要改写为环境变量，也不要写入真实 Key。实际使用的 Key 不要提交到公开代码仓库。

&emsp;&emsp;准备好后输入：

> 💬 输入给 Claude Code 的提示词

```text
帮我装上高德地图的 MCP
```

1. `Streamable HTTP`：连接高德提供的远程 MCP 服务；
2. 本地 `stdio`：通过 `npx` 在本机运行 MCP Server。

&emsp;&emsp;这里选择官方推荐的 `Streamable HTTP`。这种方式不需要在本机长期运行 Node.js 进程，配置也更加简单。

&emsp;&emsp;选择接入方式后，Claude Code 会继续询问 API Key 和安装范围。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch08-shot-04-amap-scope-api-key-d712f5cd.png" width="80%" alt="设置高德地图 MCP 的安装范围并填写 API Key"></div>

&emsp;&emsp;如果只希望在当前项目中使用，应选择 `local` 范围。对应命令为：

```bash
claude mcp add --transport http --scope local amap "https://mcp.amap.com/mcp?key=xx"
```

&emsp;&emsp;实际操作时把 URL 中的字面量 `xx` 替换为自己的 Key。完整 URL 可能进入终端历史、本地 MCP 配置和截图；如果真实 Key 泄漏，应立即在高德开放平台轮换，并同步更新本地配置。课件中的 URL 继续保留 `xx`，不改用环境变量。

&emsp;&emsp;三个常见范围的区别是：

- `local`：仅当前项目、当前用户可用；
- `project`：当前项目使用，并可通过 `.mcp.json` 与团队共享；
- `user`：当前用户的所有项目都可以使用。

&emsp;&emsp;本例选择安装到当前项目，因此使用 `local`。

&emsp;&emsp;这一章会先使用 `Ctrl + V` 与文件拖入向 Claude Code 提交图片，再用 `!` 直接执行终端命令；随后比较 `/rewind` 和 Git 的恢复边界，使用 Subagent 并行处理独立任务，通过 `/loop` 创建循环任务、用 `/goal` 持续完成长任务，最后借助 `/help` 查看命令与快捷键，并用 `/doctor` 或 `claude doctor` 诊断运行环境。

### 9.3 使用 `/rewind` 恢复文件与会话

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-02-rewind-boundary-deeaf548.png" width="80%" alt="Claude Code rewind 对文件与会话状态的恢复边界"></div>

&emsp;&emsp;下面先让 Claude Code 修改 `人工智能.txt`：

```text
把人工智能.txt中的“人工智能”换成英文
```

&emsp;&emsp;进入 Rewind 界面后，可以在之前提交过的消息之间选择恢复点。每条消息下方还会显示对应的文件及代码变更数量，例如图中的 `+2 -2` 表示该操作新增了两行，同时删除了两行。

5. `Summarize up to here`：总结所选位置之前的对话。

&emsp;&emsp;恢复完成后，`人工智能.txt` 中的内容重新变回了“人工智能”。此前提交的修改要求也会回到输入框中，可以直接编辑后重新发送。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-11-restored-file-input-1da65733.png" width="80%" alt="文件内容和原始输入要求恢复完成"></div>

&emsp;&emsp;需要注意，`/rewind` 只能恢复 Claude Code 通过文件编辑工具记录下来的修改，不能撤销手动编辑或 Bash 命令产生的变化。例如：

```bash
rm file.txt
echo "" > file.txt
```

&emsp;&emsp;这类命令直接由操作系统执行。Claude Code 只会获得命令及其输出，不会为命令涉及的所有文件和外部状态保存完整的修改前副本，因此无法可靠地自动反向执行。

&emsp;&emsp;同样，脚本修改、数据库写入、API 调用和部署操作也不能依靠 `/rewind` 撤销。执行删除、覆盖或清空等操作前，仍应使用 Git、系统废纸篓或其他备份方式保护重要文件。`/rewind` 是会话级的便捷撤销功能，不能替代版本控制。

### 9.4 使用 Git 保存和恢复文件状态

&emsp;&emsp;`/rewind` 无法恢复手动操作或 Bash 命令造成的文件变化。如果希望更可靠地保存项目状态，可以使用 Git 进行版本管理。

&emsp;&emsp;首次提交前先完成一个紧凑检查点：

- 使用 `git --version` 确认 Git 可用；如果首次提交因缺少 `user.name` 或 `user.email` 失败，先按错误提示完成本机 Git 身份配置，再重新提交；
- 暂存前使用 `git status --short` 查看项目文件并检查 `.gitignore`，暂存后使用 `git diff --cached --name-only` 复核暂存清单；
- 从提交中排除 `.env`、真实 Key、会话文件、本机 Claude 配置以及不应入库的生成数据；
- 下方提示词中的“当前项目所有文件”，只指完成敏感信息与忽略规则检查后允许入库的示例文件。

&emsp;&emsp;首先让 Claude Code 创建 Git 仓库，并保存当前项目中的文件：

> 💬 输入给 Claude Code 的提示词

```text
创建git仓库，把当前项目所有文件存到git仓库里
```

&emsp;&emsp;Claude Code 会检查当前目录和 Git 状态，然后初始化仓库、添加文件并创建第一次提交。

&emsp;&emsp;图中，项目已经提交到 `main` 分支，本次提交编号为 `<提交ID>`。提交编号可以理解为这个项目状态的唯一标识，以后可以利用它查看或恢复当时的文件。

&emsp;&emsp;只有确认首次提交成功并看到提交 ID 后，才能继续下面的删除与恢复演示；如果提交失败，应停在这里处理错误，不能先删除文件。

> 💬 输入给 Claude Code 的提示词

```text
恢复到 <提交ID> 状态
```

&emsp;&emsp;Claude Code 检查后发现，当前 `HEAD` 本来就位于 `<提交ID>`，只是工作区中的 `人工智能.txt` 被删除了。因此不需要切换或重置整个仓库，只需从该提交中恢复被删除的文件。

&emsp;&emsp;需要注意，Git 无法恢复从未提交过的新文件。因此，在进行删除、批量替换或其他风险较高的操作前，应先创建一次提交，为当前项目保存一个明确的恢复点。

### 9.5 使用 Subagent 并行执行任务

&emsp;&emsp;Claude Code 可以把一个复杂任务拆成多个相互独立的子任务，再交给不同的 Subagent 同时处理。每个 Subagent 都拥有独立的上下文，完成任务后会把结果返回给主会话，由主代理负责读取、比较和汇总。

&emsp;&emsp;下面让 Claude Code 同时派出两个 Subagent，分别研究两个模型：

> 💬 输入给 Claude Code 的提示词

```text
你派两个subagent，模型用sonnet，思考强度用high，一个搜索kimi k3的相关资料然后写到当前文件夹中，一个搜索GPT5.6 sol模型的相关资料，也写到当前文件夹中，最多搜索三个网站，之后你比较下两者，给出孰优孰劣的结论
```

&emsp;&emsp;Claude Code 首先整理执行计划，然后同时启动两个后台 Subagent：

- 一个负责搜索 Kimi K3 的资料。
- 一个负责搜索 GPT5.6 sol 的资料。
- 两者完成后，由主代理读取文件并进行横向比较。

&emsp;&emsp;Subagent 的最终质量主要受到以下因素影响：

- 子任务是否清晰、具体，并且彼此独立。

- 搜索范围和可用工具是否足够。

- 主代理是否对多个结果进行了复核。

&emsp;&emsp;自 Claude Code v2.1.198 起，内置 `Explore` 子代理默认继承主会话模型（Claude API 上继承结果最高限制为 Opus）；`general-purpose` 子代理也默认继承主会话模型，两者均可受配置或调用参数影响。[Claude Code Subagent 官方文档](https://code.claude.com/docs/en/sub-agents)

### 9.6 使用 `/loop` 创建循环任务

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-03-loop-goal-5df2e966.png" width="80%" alt="Claude Code 循环任务与持续目标的执行关系"></div>

&emsp;&emsp;Claude Code 的 `/loop` 可以按照指定的时间间隔重复执行同一条提示词，适合设置临时提醒、轮询部署状态或定期检查长时间运行的任务。

&emsp;&emsp;例如，每分钟提醒一次喝水，并输出当前时间：

> 💬 输入给 Claude Code 的提示词

```text
/loop 1m 提醒我喝水，然后给出当前时间
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-23-loop-reminder-90e7f813.png" width="80%" alt="使用 loop 命令创建每分钟执行一次的喝水提醒"></div>

&emsp;&emsp;Claude Code 会把 `1m` 转换成对应的 Cron 表达式：

```text
*/1 * * * *
```

&emsp;&emsp;这里的表达式表示每分钟执行一次。任务创建成功后还会生成一个由 8 个字符组成的任务 ID，图中的任务 ID 为 `<任务ID>`。

&emsp;&emsp;在这次操作中，Claude Code 先立即执行了一次提醒，随后在每分钟到达计划时间时重新执行，输出新的喝水提示和当前时间。

&emsp;&emsp;`/loop` 的基本格式为：

```text
/loop 时间间隔 要重复执行的任务
```

&emsp;&emsp;常见时间间隔包括：

```text
1m    每分钟
5m    每5分钟
1h    每小时
```

&emsp;&emsp;`/loop` 的最小固定间隔为 1 分钟。循环任务会在当前 Claude Code 会话中运行，并在会话空闲时触发；如果 Claude Code 正在回答其他问题，任务会等当前回答结束后再执行。循环任务最多保留 7 天，之后会自动过期。

&emsp;&emsp;不再需要循环任务时，可以直接用自然语言要求 Claude Code 取消：

> 💬 输入给 Claude Code 的提示词

```text
取消喝水loop
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-24-cancel-loop-task-2884faed.png" width="80%" alt="取消正在运行的喝水循环任务"></div>

&emsp;&emsp;Claude Code 会根据任务 ID 调用 `CronDelete` 删除循环任务。图中显示任务 `<任务ID>` 已被取消，此后不会继续定时提醒。

&emsp;&emsp;对于使用 `/loop` 创建、正在等待下一次执行的任务，也可以按 `Esc` 停止。还可以用自然语言让 Claude Code 查看当前任务：

> 💬 输入给 Claude Code 的提示词

```text
列出当前所有定时任务
```

&emsp;&emsp;`/loop` 主要适合当前会话中的临时循环任务，需要电脑保持运行并维持相应会话。如果希望任务在关闭会话或电脑后仍能长期执行，可以使用 `/schedule` 创建云端定时任务，或者使用 Claude Code Desktop 的计划任务功能。[Claude Code 定时任务官方文档](https://code.claude.com/docs/en/scheduled-tasks)

### 9.7 使用 `/goal` 持续完成长任务

&emsp;&emsp;面对大型开发、资料整理或反复测试等长任务时，Claude Code 有时会完成一轮工作后停止，等待我们继续下达指令。`/goal` 可以为任务设置一个明确的完成条件，让 Claude Code 跨越多个回合持续工作，直到确认目标已经实现。

&emsp;&emsp;例如，让 Claude Code 搜索 Sonnet 5 的相关新闻，并与已有的 Kimi K3 资料进行比较：

> 💬 输入给 Claude Code 的提示词

```text
/goal 搜索下sonnet5的新闻来和当前kimi-k3来比较，最多搜索3个网站
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-25-start-goal-110a1b56.png" width="80%" alt="使用 goal 命令设置需要持续完成的研究目标"></div>

&emsp;&emsp;目标设置成功后，界面会显示：

```text
Goal set: 搜索下sonnet5的新闻来和当前kimi-k3来比较，最多搜索3个网站
```

&emsp;&emsp;右下角的 `◎ /goal active` 表示当前 Goal 正在运行。Claude Code 会读取项目中已有的 Kimi K3 资料，并继续联网搜索、整理信息和完成对比。

&emsp;&emsp;`/goal` 的基本格式为：

```text
/goal 完成条件
```

&emsp;&emsp;普通提示词主要说明“现在要做什么”，而 Goal 最好说明“做到什么程度才算完成”。例如：

> 💬 输入给 Claude Code 的提示词

```text
/goal 已从不超过3个网站搜集Sonnet 5的资料，与kimi-k3-资料.md完成对比，所有结论注明来源，无法确认的信息明确标为不确定
```

&emsp;&emsp;这种写法包含了任务范围、输出结果和验收条件，更容易让 Claude Code 判断目标是否真正完成。

&emsp;&emsp;Claude Code 每完成一个回合后，会使用一个独立的快速模型检查目标条件是否满足：

- 如果目标尚未完成，会自动开始下一轮工作。
- 如果目标已经完成，会停止继续执行并清除当前 Goal。
- 如果希望提前终止，可以输入：

```text
/goal clear
```

&emsp;&emsp;输入不带参数的 `/goal`，可以查看当前目标、运行时间、已经执行的回合数、Token 消耗和最近一次检查结果：

```text
/goal
```

&emsp;&emsp;经过持续搜索和整理，Claude Code 最终生成了 Kimi K3 与 Claude Sonnet 5 的对比表，并在表格中标注不同数据的可信度和存疑项。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-26-goal-comparison-result-17ae9c05.png" width="80%" alt="Goal 模式持续执行后完成资料对比"></div>

&emsp;&emsp;这份比较结果用于展示 Goal 模式能够持续推进任务并产出最终结果，表中的模型参数和结论仍需结合官方来源进一步核实，不能直接视为权威评测。

&emsp;&emsp;`/goal` 特别适合以下任务：

- 持续修复代码，直到测试和构建全部通过。
- 按照需求文档实现功能，直到所有验收条件完成。
- 拆分大型文件，直到每个文件满足规定的大小。
- 连续处理待办问题，直到指定列表清空。
- 睡觉前启动耗时较长、结束状态明确的工作。

&emsp;&emsp;需要注意，`/goal` 只负责让任务跨回合持续推进，不会改变 Claude Code 的权限设置。如果操作需要额外授权，Claude Code 仍然可能停下来等待确认。希望任务无人值守运行时，可以搭配 Auto mode，让已经允许范围内的工具操作自动执行。

&emsp;&emsp;开始长任务前，建议同时明确：

- 可以修改哪些目录和文件。
- 必须运行哪些测试作为验收依据。
- 哪些文件或功能不能改变。
- 最多执行多少回合或运行多长时间。
- 禁止删除、部署、推送或调用外部服务等高风险操作。

&emsp;&emsp;例如，可以在目标末尾增加：

> 💬 输入给 Claude Code 的提示词

```text
如果20个回合后仍未完成则停止，并总结当前进度和阻塞原因
```

&emsp;&emsp;`/goal` 使用的快速模型只负责判断完成条件，真正执行任务的仍然是当前会话选择的主模型。长时间运行会持续消耗 Token，因此应设置可验证的目标和合理的停止条件。[Claude Code Goal 官方文档](https://code.claude.com/docs/en/goal)

### 9.8 使用 `/help` 查看命令和快捷键

&emsp;&emsp;当我们忘记某个命令或快捷键时，可以输入 `/help` 打开 Claude Code 的帮助界面：

```text
/help
```

### 9.9 使用 `/doctor` 检查运行环境

&emsp;&emsp;当 Claude Code 出现配置未生效、MCP 无法连接、Skill 无法识别或运行状态异常等问题时，可以使用 `/doctor` 对当前安装和配置进行健康检查。

&emsp;&emsp;输入：

```text
/doctor
```

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-31-doctor-command-dda38416.png" width="80%" alt="在 Claude Code 中输入 doctor 诊断命令"></div>

&emsp;&emsp;输入 `/doc` 等部分内容时，Claude Code 会自动筛选并提示 `/doctor` 命令。按回车即可开始检查。

&emsp;&emsp;`/doctor` 会检查 Claude Code 的安装状态、配置文件、MCP 服务和上下文使用情况，并报告发现的异常。图中，Claude Code 首先说明会进行只读检查，再统一生成诊断报告。

<div align=center><img src="https://typora-photo1220.oss-cn-beijing.aliyuncs.com/DataAnalysis/Leo/claude-code-quickstart/2026-07-21/ch09-shot-32-doctor-health-check-f6c7cdeb.png" width="80%" alt="doctor 命令执行 Claude Code 环境健康检查"></div>

&emsp;&emsp;检查过程中，Claude Code 会读取相关配置文件并查看项目级和用户级目录，例如：

```text
.claude/agents/
~/.claude/agents/
.claude/skills/
~/.claude/skills/
```

&emsp;&emsp;它还会验证设置文件是否包含无效字段或格式错误，检查 MCP 连接、安装状态以及可能占用大量上下文的配置。图中显示读取了两个文件、列出了两个目录，并执行了七条诊断命令。

&emsp;&emsp;诊断完成后，结果会通过不同的状态图标显示。如果发现可以自动处理的问题，可以根据界面提示按 `f`，让 Claude Code 尝试修复。

&emsp;&emsp;`/doctor` 适合排查以下情况：

- Claude Code 安装或升级后运行异常。
- `settings.json` 中的配置没有生效。
- MCP 服务无法连接或没有提供工具。
- Hook、Skill 或 Subagent 没有被正确识别。
- 上下文占用异常，响应速度明显下降。
- 项目级配置和用户级配置发生冲突。

&emsp;&emsp;如果 Claude Code 已经无法正常启动，不能在会话中输入 `/doctor`，可以在普通终端中执行：

```bash
claude doctor
```

&emsp;&emsp;`/doctor` 主要用于发现本机安装与配置问题。API 服务故障、账号权限或模型暂时不可用等问题，还需要结合具体错误提示继续排查。[Claude Code 命令参考](https://code.claude.com/docs/en/commands)
